## Step 1 Fit function to the large ensemble dataset

```python
# ============================================================
# Hydropower attribution
# Step 1: Fit factual and counterfactual ensemble distributions
#
# The observed event threshold is read directly from each
# province's historical hydro-anomaly CSV:
#
#     year == 2022
#     month == 9
#     std_anomaly
#
# The ensemble samples are the 525 population-weighted
# September hydro standardized anomalies produced in Step 5.
# ============================================================

import os
import pickle
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy import stats
from fitter import Fitter


warnings.filterwarnings(
    "ignore",
    category=RuntimeWarning,
)


# ============================================================
# 1. User settings
# ============================================================

# ------------------------------------------------------------
# Factual and counterfactual 525-member outputs from Step 5
# ------------------------------------------------------------

FACTUAL_CSV = (
    "/home/users/weijia_6/Yangtze_code/gen_result/"
    "step5_gamma_spi6/"
    "hydro_std_anom_sep_factual_525_by_province.csv"
)

COUNTERFACTUAL_CSV = (
    "/home/users/weijia_6/Yangtze_code/gen_result/"
    "step5_gamma_spi6/"
    "hydro_std_anom_sep_counterfactual_525_by_province.csv"
)


# ------------------------------------------------------------
# Observed hydro-anomaly CSV files
#
# Files are expected to be named:
#
# guizhou_hydro_anomaly_and_std_in_reg.csv
# chongqing_hydro_anomaly_and_std_in_reg.csv
# sichuan_hydro_anomaly_and_std_in_reg.csv
# hubei_hydro_anomaly_and_std_in_reg.csv
# hunan_hydro_anomaly_and_std_in_reg.csv
# ------------------------------------------------------------

OBSERVED_HYDRO_DIRECTORY = (
    "/home/users/weijia_6/Yangtze_code/"
    "gen_data/hydropower_anomaly"
)


# ------------------------------------------------------------
# Province metadata
#
# csv_name:
#     Name used in the factual/counterfactual Step 5 CSV files.
#
# display_name:
#     Short name used in figures.
# ------------------------------------------------------------

PROVINCE_INFO = {
    "guizhou": {
        "csv_name": "Guizhou Province",
        "display_name": "Guizhou",
        "panel": "(a)",
    },
    "chongqing": {
        "csv_name": "Chongqing Municipality",
        "display_name": "Chongqing",
        "panel": "(b)",
    },
    "sichuan": {
        "csv_name": "Sichuan Province",
        "display_name": "Sichuan",
        "panel": "(c)",
    },
    "hubei": {
        "csv_name": "Hubei Province",
        "display_name": "Hubei",
        "panel": "(d)",
    },
    "hunan": {
        "csv_name": "Hunan Province",
        "display_name": "Hunan",
        "panel": "(e)",
    },
}


SELECTED_PROVINCES = [
    information["csv_name"]
    for information in PROVINCE_INFO.values()
]


# ------------------------------------------------------------
# Observed event definition
# ------------------------------------------------------------

EVENT_YEAR = 2022
EVENT_MONTH = 9

OBSERVED_YEAR_COLUMN = "year"
OBSERVED_MONTH_COLUMN = "month"
OBSERVED_ANOMALY_COLUMN = "std_anomaly"


# ------------------------------------------------------------
# Ensemble CSV columns
# ------------------------------------------------------------

PROVINCE_COLUMN = "province"
ENSEMBLE_COLUMN = "ensemble"
ENSEMBLE_VALUE_COLUMN = "hydro_std_anom_sep"
SCENARIO_COLUMN = "scenario"

EXPECTED_ENSEMBLES = 525


# ------------------------------------------------------------
# Output directory
# ------------------------------------------------------------

OUTPUT_DIRECTORY = Path(
    "/home/users/weijia_6/Yangtze_code/gen_result/"
    "attribution_hydro/step1_distribution_fitting"
)

OUTPUT_DIRECTORY.mkdir(
    parents=True,
    exist_ok=True,
)


# ============================================================
# 2. Candidate probability distributions
# ============================================================

# Hydro standardized anomalies can be positive or negative.
#
# These distributions can represent data over the real line,
# either naturally or through a fitted location parameter.
#
# The same family will ultimately be used for all five
# provinces and both scenarios.
CANDIDATE_DISTRIBUTIONS = [
    "norm",
    "logistic",
    "pearson3",
    "genextreme",
    "skewnorm",
    "johnsonsu",
    "t",
    "gennorm",
]

FIT_METRIC = "sumsquare_error"
FIT_TIMEOUT = 600
FITTER_BINS = 40


# ============================================================
# 3. Figure settings
# ============================================================

FACTUAL_COLOR = "#CC6677"
COUNTERFACTUAL_COLOR = "#4477AA"
EVENT_COLOR = "#555555"

FIGURE_DPI = 300

PDF_FIGSIZE = (
    11,
    3.2 * len(PROVINCE_INFO),
)

CDF_FIGSIZE = (
    11,
    3.2 * len(PROVINCE_INFO),
)

N_HISTOGRAM_BINS = 32
N_X_POINTS = 600

# Quantiles used to avoid an excessively wide range caused by
# unstable fitted tails.
PLOT_LOWER_QUANTILE = 0.001
PLOT_UPPER_QUANTILE = 0.999
PLOT_PADDING_FRACTION = 0.08


plt.rcParams.update(
    {
        "font.family": "DejaVu Sans",
        "font.size": 13,
        "axes.unicode_minus": False,
        "axes.linewidth": 1.0,
        "xtick.direction": "out",
        "ytick.direction": "out",
    }
)


# ============================================================
# 4. General helpers
# ============================================================

def assert_file_exists(path):
    """Require an input file to exist."""
    if not os.path.isfile(path):
        raise FileNotFoundError(
            f"File not found:\n{path}"
        )


def assert_directory_exists(path):
    """Require an input directory to exist."""
    if not os.path.isdir(path):
        raise FileNotFoundError(
            f"Directory not found:\n{path}"
        )


def clean_data(values):
    """Convert values to float and retain finite values only."""
    values = np.asarray(
        values,
        dtype=np.float64,
    )

    return values[
        np.isfinite(values)
    ]


def get_distribution_object(
    distribution_name,
):
    """Return a scipy.stats distribution object."""
    if not hasattr(
        stats,
        distribution_name,
    ):
        raise AttributeError(
            f"scipy.stats does not contain "
            f"'{distribution_name}'."
        )

    return getattr(
        stats,
        distribution_name,
    )


def build_observed_file_path(
    province_key,
):
    """Build one observed province CSV path."""
    filename = (
        f"{province_key}_"
        "hydro_anomaly_and_std_in_reg.csv"
    )

    return os.path.join(
        OBSERVED_HYDRO_DIRECTORY,
        filename,
    )


# ============================================================
# 5. Load the 525-member ensemble CSV files
# ============================================================

def load_ensemble_csv(
    path,
    expected_scenario,
):
    """
    Load one Step 5 long-format ensemble CSV.

    Expected columns:
        province
        ensemble
        hydro_std_anom_sep
    """
    assert_file_exists(
        path
    )

    data = pd.read_csv(
        path
    )

    data.columns = [
        str(column).strip()
        for column in data.columns
    ]

    required_columns = {
        PROVINCE_COLUMN,
        ENSEMBLE_COLUMN,
        ENSEMBLE_VALUE_COLUMN,
    }

    missing_columns = (
        required_columns
        - set(data.columns)
    )

    if missing_columns:
        raise KeyError(
            f"Missing columns in:\n{path}\n"
            f"Missing: {sorted(missing_columns)}\n"
            f"Available: {list(data.columns)}"
        )

    data = data.copy()

    data[PROVINCE_COLUMN] = (
        data[PROVINCE_COLUMN]
        .astype(str)
        .str.strip()
    )

    data[ENSEMBLE_COLUMN] = pd.to_numeric(
        data[ENSEMBLE_COLUMN],
        errors="coerce",
    )

    data[ENSEMBLE_VALUE_COLUMN] = pd.to_numeric(
        data[ENSEMBLE_VALUE_COLUMN],
        errors="coerce",
    )

    # Keep only selected provinces.
    data = data.loc[
        data[PROVINCE_COLUMN].isin(
            SELECTED_PROVINCES
        )
    ].copy()

    data[SCENARIO_COLUMN] = expected_scenario

    duplicated = data.duplicated(
        subset=[
            PROVINCE_COLUMN,
            ENSEMBLE_COLUMN,
        ],
        keep=False,
    )

    if duplicated.any():
        duplicate_rows = data.loc[
            duplicated,
            [
                PROVINCE_COLUMN,
                ENSEMBLE_COLUMN,
            ],
        ]

        raise ValueError(
            "Duplicated province-ensemble records were found "
            f"in:\n{path}\n"
            f"{duplicate_rows.head(20)}"
        )

    return data


def validate_ensemble_datasets(
    factual,
    counterfactual,
):
    """
    Check that each province contains 525 members in both
    scenarios and that ensemble keys match.
    """
    print(
        "\n" + "=" * 65
    )
    print(
        "Ensemble-data checks"
    )
    print(
        "=" * 65
    )

    for province in SELECTED_PROVINCES:
        factual_subset = factual.loc[
            factual[PROVINCE_COLUMN]
            == province
        ]

        counterfactual_subset = (
            counterfactual.loc[
                counterfactual[
                    PROVINCE_COLUMN
                ]
                == province
            ]
        )

        factual_members = (
            factual_subset[
                ENSEMBLE_COLUMN
            ]
            .dropna()
            .nunique()
        )

        counterfactual_members = (
            counterfactual_subset[
                ENSEMBLE_COLUMN
            ]
            .dropna()
            .nunique()
        )

        factual_finite = int(
            np.isfinite(
                factual_subset[
                    ENSEMBLE_VALUE_COLUMN
                ]
            ).sum()
        )

        counterfactual_finite = int(
            np.isfinite(
                counterfactual_subset[
                    ENSEMBLE_VALUE_COLUMN
                ]
            ).sum()
        )

        print(
            f"\n{province}"
        )

        print(
            "  Factual members:",
            factual_members,
        )

        print(
            "  Factual finite values:",
            factual_finite,
        )

        print(
            "  Counterfactual members:",
            counterfactual_members,
        )

        print(
            "  Counterfactual finite values:",
            counterfactual_finite,
        )

        if factual_members != EXPECTED_ENSEMBLES:
            raise ValueError(
                f"{province} has {factual_members} factual "
                f"members; expected {EXPECTED_ENSEMBLES}."
            )

        if (
            counterfactual_members
            != EXPECTED_ENSEMBLES
        ):
            raise ValueError(
                f"{province} has "
                f"{counterfactual_members} counterfactual "
                f"members; expected {EXPECTED_ENSEMBLES}."
            )

        if factual_finite < 100:
            raise ValueError(
                f"Too few finite factual values for "
                f"{province}: {factual_finite}."
            )

        if counterfactual_finite < 100:
            raise ValueError(
                f"Too few finite counterfactual values for "
                f"{province}: {counterfactual_finite}."
            )

    factual_keys = (
        factual[
            [
                PROVINCE_COLUMN,
                ENSEMBLE_COLUMN,
            ]
        ]
        .sort_values(
            [
                PROVINCE_COLUMN,
                ENSEMBLE_COLUMN,
            ]
        )
        .reset_index(drop=True)
    )

    counterfactual_keys = (
        counterfactual[
            [
                PROVINCE_COLUMN,
                ENSEMBLE_COLUMN,
            ]
        ]
        .sort_values(
            [
                PROVINCE_COLUMN,
                ENSEMBLE_COLUMN,
            ]
        )
        .reset_index(drop=True)
    )

    if not factual_keys.equals(
        counterfactual_keys
    ):
        raise ValueError(
            "The factual and counterfactual datasets do not "
            "contain identical province-ensemble combinations."
        )


def extract_province_sample(
    data,
    province,
):
    """Return the finite ensemble sample for one province."""
    sample = data.loc[
        data[PROVINCE_COLUMN] == province,
        ENSEMBLE_VALUE_COLUMN,
    ].values

    return clean_data(
        sample
    )


# ============================================================
# 6. Read the real September 2022 event anomaly
# ============================================================

def load_one_observed_province(
    province_key,
    province_name,
):
    """
    Read one historical province CSV and extract the actual
    September 2022 hydropower standardized anomaly.
    """
    input_path = build_observed_file_path(
        province_key
    )

    assert_file_exists(
        input_path
    )

    data = pd.read_csv(
        input_path,
        encoding="utf-8-sig",
    )

    data.columns = (
        data.columns
        .astype(str)
        .str.strip()
    )

    required_columns = {
        OBSERVED_YEAR_COLUMN,
        OBSERVED_MONTH_COLUMN,
        OBSERVED_ANOMALY_COLUMN,
    }

    missing_columns = (
        required_columns
        - set(data.columns)
    )

    if missing_columns:
        raise KeyError(
            f"{province_name} observed CSV is missing "
            f"columns: {sorted(missing_columns)}\n"
            f"File: {input_path}\n"
            f"Available columns: {list(data.columns)}"
        )

    data = data.copy()

    for column in required_columns:
        data[column] = pd.to_numeric(
            data[column],
            errors="coerce",
        )

    data = data.replace(
        [np.inf, -np.inf],
        np.nan,
    )

    event_rows = data.loc[
        (
            data[
                OBSERVED_YEAR_COLUMN
            ]
            == EVENT_YEAR
        )
        & (
            data[
                OBSERVED_MONTH_COLUMN
            ]
            == EVENT_MONTH
        )
    ].copy()

    if event_rows.empty:
        raise ValueError(
            f"No {EVENT_YEAR}-{EVENT_MONTH:02d} record was "
            f"found for {province_name}.\n"
            f"File: {input_path}"
        )

    finite_event_rows = event_rows.loc[
        np.isfinite(
            event_rows[
                OBSERVED_ANOMALY_COLUMN
            ]
        )
    ]

    if finite_event_rows.empty:
        raise ValueError(
            f"The {EVENT_YEAR}-{EVENT_MONTH:02d} "
            f"{OBSERVED_ANOMALY_COLUMN} value is missing for "
            f"{province_name}."
        )

    if len(finite_event_rows) > 1:
        unique_values = np.unique(
            finite_event_rows[
                OBSERVED_ANOMALY_COLUMN
            ].values
        )

        if len(unique_values) > 1:
            raise ValueError(
                f"Multiple different September 2022 event "
                f"values were found for {province_name}: "
                f"{unique_values}"
            )

        print(
            f"[WARNING] Duplicate September 2022 rows were "
            f"found for {province_name}; using the last row."
        )

    event_value = float(
        finite_event_rows[
            OBSERVED_ANOMALY_COLUMN
        ].iloc[-1]
    )

    return {
        "province_key": province_key,
        "province": province_name,
        "event_year": EVENT_YEAR,
        "event_month": EVENT_MONTH,
        "event_value": event_value,
        "source_file": input_path,
    }


def load_all_observed_events():
    """
    Load the real September 2022 anomaly for all provinces.
    """
    event_records = []

    print(
        "\n" + "=" * 65
    )
    print(
        "Observed September 2022 event values"
    )
    print(
        "=" * 65
    )

    for province_key, information in (
        PROVINCE_INFO.items()
    ):
        record = load_one_observed_province(
            province_key=province_key,
            province_name=information[
                "csv_name"
            ],
        )

        event_records.append(
            record
        )

        print(
            f"{record['province']}: "
            f"{record['event_value']:.4f}"
        )

    event_table = pd.DataFrame(
        event_records
    )

    event_values = dict(
        zip(
            event_table["province"],
            event_table["event_value"],
        )
    )

    return (
        event_table,
        event_values,
    )


# ============================================================
# 7. Extract Fitter result table
# ============================================================

def extract_fit_table(
    fitter_object,
    dataset_name,
):
    """
    Extract the Fitter goodness-of-fit table in a consistent
    format.
    """
    if (
        hasattr(
            fitter_object,
            "df_errors",
        )
        and fitter_object.df_errors is not None
    ):
        table = (
            fitter_object
            .df_errors
            .copy()
        )

    else:
        table = fitter_object.summary()

    if not isinstance(
        table,
        pd.DataFrame,
    ):
        table = pd.DataFrame(
            table
        )

    table = table.reset_index()

    table.columns = [
        str(column)
        .strip()
        .lower()
        for column in table.columns
    ]

    if "distribution" not in table.columns:
        if "index" in table.columns:
            table = table.rename(
                columns={
                    "index":
                    "distribution"
                }
            )

        else:
            table = table.rename(
                columns={
                    table.columns[0]:
                    "distribution"
                }
            )

    if FIT_METRIC not in table.columns:
        raise KeyError(
            f"Fitter output for {dataset_name} does not "
            f"contain '{FIT_METRIC}'.\n"
            f"Columns: {list(table.columns)}"
        )

    table[FIT_METRIC] = pd.to_numeric(
        table[FIT_METRIC],
        errors="coerce",
    )

    table["dataset"] = dataset_name

    return table


# ============================================================
# 8. Fit all candidate distributions to all ten samples
# ============================================================

def fit_all_candidates(
    factual,
    counterfactual,
):
    """
    Fit each candidate family independently to:

        5 provinces x 2 scenarios = 10 samples
    """
    fit_tables = []
    preliminary_fit_objects = {}

    for province in SELECTED_PROVINCES:
        samples = {
            "factual": extract_province_sample(
                factual,
                province,
            ),
            "counterfactual":
            extract_province_sample(
                counterfactual,
                province,
            ),
        }

        for scenario, sample in samples.items():
            dataset_name = (
                f"{province}__{scenario}"
            )

            print(
                "\n" + "=" * 65
            )
            print(
                f"Fitting {dataset_name}"
            )
            print(
                "=" * 65
            )

            print(
                "Sample size:",
                sample.size,
            )

            print(
                "Mean:",
                float(
                    np.mean(sample)
                ),
            )

            print(
                "Standard deviation:",
                float(
                    np.std(
                        sample,
                        ddof=1,
                    )
                ),
            )

            print(
                "Minimum:",
                float(
                    np.min(sample)
                ),
            )

            print(
                "Maximum:",
                float(
                    np.max(sample)
                ),
            )

            fitter_object = Fitter(
                sample,
                distributions=(
                    CANDIDATE_DISTRIBUTIONS
                ),
                timeout=FIT_TIMEOUT,
                bins=FITTER_BINS,
            )

            fitter_object.fit()

            fit_table = extract_fit_table(
                fitter_object,
                dataset_name,
            )

            fit_tables.append(
                fit_table
            )

            preliminary_fit_objects[
                dataset_name
            ] = {
                "data": sample,
                "fitter": fitter_object,
                "fit_table": fit_table,
            }

    all_fit_table = pd.concat(
        fit_tables,
        ignore_index=True,
    )

    # Remove unsuccessful/non-finite fits.
    all_fit_table = all_fit_table.loc[
        np.isfinite(
            all_fit_table[
                FIT_METRIC
            ]
        )
    ].copy()

    return (
        all_fit_table,
        preliminary_fit_objects,
    )


# ============================================================
# 9. Select one common distribution family
# ============================================================

def select_shared_family(
    all_fit_table,
):
    """
    Select one common family using the average rank of SSE
    across all ten datasets.

    Lower SSE and lower rank are better.
    """
    ranked_table = all_fit_table.copy()

    ranked_table["rank_sse"] = (
        ranked_table
        .groupby("dataset")[
            FIT_METRIC
        ]
        .rank(
            method="average",
            ascending=True,
        )
    )

    best_indices = (
        ranked_table
        .groupby("dataset")[
            FIT_METRIC
        ]
        .idxmin()
    )

    best_by_dataset = (
        ranked_table.loc[
            best_indices,
            [
                "dataset",
                "distribution",
                FIT_METRIC,
                "rank_sse",
            ],
        ]
        .rename(
            columns={
                "distribution":
                "best_distribution",
                FIT_METRIC:
                "best_sumsquare_error",
            }
        )
        .reset_index(drop=True)
    )

    winner_counts = (
        best_by_dataset[
            "best_distribution"
        ]
        .value_counts()
        .rename(
            "win_count"
        )
    )

    shared_summary = (
        ranked_table
        .groupby("distribution")
        .agg(
            datasets_fitted=(
                "dataset",
                "nunique",
            ),
            mean_rank_sse=(
                "rank_sse",
                "mean",
            ),
            median_rank_sse=(
                "rank_sse",
                "median",
            ),
            mean_sse=(
                FIT_METRIC,
                "mean",
            ),
            median_sse=(
                FIT_METRIC,
                "median",
            ),
            maximum_sse=(
                FIT_METRIC,
                "max",
            ),
        )
        .join(
            winner_counts,
            how="left",
        )
        .fillna(
            {
                "win_count": 0
            }
        )
    )

    expected_dataset_count = (
        len(SELECTED_PROVINCES)
        * 2
    )

    # Require successful fitting to all ten datasets.
    complete_summary = (
        shared_summary.loc[
            shared_summary[
                "datasets_fitted"
            ]
            == expected_dataset_count
        ]
        .copy()
    )

    if complete_summary.empty:
        raise RuntimeError(
            "No distribution was fitted successfully to all "
            f"{expected_dataset_count} datasets."
        )

    complete_summary = (
        complete_summary
        .sort_values(
            [
                "mean_rank_sse",
                "mean_sse",
                "median_sse",
                "maximum_sse",
            ],
            ascending=True,
        )
    )

    shared_family = (
        complete_summary.index[0]
    )

    print(
        "\n" + "=" * 65
    )
    print(
        "Shared distribution-family ranking"
    )
    print(
        "=" * 65
    )

    print(
        complete_summary.to_string()
    )

    print(
        "\nSelected shared family:",
        shared_family,
    )

    return (
        shared_family,
        ranked_table,
        complete_summary,
        best_by_dataset,
    )


# ============================================================
# 10. Refit all ten samples with the shared family
# ============================================================

def refit_using_shared_family(
    factual,
    counterfactual,
    shared_family,
):
    """
    Refit factual and counterfactual samples using the same
    selected family.
    """
    distribution_family = (
        get_distribution_object(
            shared_family
        )
    )

    fit_store = {}
    summary_records = []

    for province in SELECTED_PROVINCES:
        factual_values = extract_province_sample(
            factual,
            province,
        )

        counterfactual_values = (
            extract_province_sample(
                counterfactual,
                province,
            )
        )

        factual_parameters = (
            distribution_family.fit(
                factual_values
            )
        )

        counterfactual_parameters = (
            distribution_family.fit(
                counterfactual_values
            )
        )

        factual_distribution = (
            distribution_family(
                *factual_parameters
            )
        )

        counterfactual_distribution = (
            distribution_family(
                *counterfactual_parameters
            )
        )

        fit_store[province] = {
            "shared_family": shared_family,
            "factual": {
                "data": factual_values,
                "params": factual_parameters,
                "dist": factual_distribution,
            },
            "counterfactual": {
                "data": counterfactual_values,
                "params":
                counterfactual_parameters,
                "dist":
                counterfactual_distribution,
            },
        }

        summary_records.append(
            {
                "province": province,
                "shared_family": shared_family,
                "factual_sample_size":
                factual_values.size,
                "counterfactual_sample_size":
                counterfactual_values.size,
                "factual_empirical_mean": float(
                    np.mean(factual_values)
                ),
                "counterfactual_empirical_mean": float(
                    np.mean(
                        counterfactual_values
                    )
                ),
                "factual_empirical_std": float(
                    np.std(
                        factual_values,
                        ddof=1,
                    )
                ),
                "counterfactual_empirical_std": float(
                    np.std(
                        counterfactual_values,
                        ddof=1,
                    )
                ),
                "factual_parameters": repr(
                    factual_parameters
                ),
                "counterfactual_parameters": repr(
                    counterfactual_parameters
                ),
            }
        )

        print(
            f"\nRefitted {province}"
        )

        print(
            "  Family:",
            shared_family,
        )

        print(
            "  Factual parameters:",
            factual_parameters,
        )

        print(
            "  Counterfactual parameters:",
            counterfactual_parameters,
        )

    summary_table = pd.DataFrame(
        summary_records
    )

    return (
        fit_store,
        summary_table,
    )


# ============================================================
# 11. Plot-range helpers
# ============================================================

def calculate_local_plot_limits(
    factual_values,
    counterfactual_values,
    factual_distribution,
    counterfactual_distribution,
    event_value,
):
    """
    Calculate a stable x-axis range that includes the samples,
    the observed event, and central fitted-distribution ranges.
    """
    combined_values = np.concatenate(
        [
            factual_values,
            counterfactual_values,
            np.array(
                [event_value],
                dtype=np.float64,
            ),
        ]
    )

    lower_candidates = [
        float(
            np.min(
                combined_values
            )
        ),
        float(
            np.quantile(
                combined_values,
                PLOT_LOWER_QUANTILE,
            )
        ),
        float(event_value),
    ]

    upper_candidates = [
        float(
            np.max(
                combined_values
            )
        ),
        float(
            np.quantile(
                combined_values,
                PLOT_UPPER_QUANTILE,
            )
        ),
        float(event_value),
    ]

    for fitted_distribution in [
        factual_distribution,
        counterfactual_distribution,
    ]:
        try:
            fitted_lower = float(
                fitted_distribution.ppf(
                    PLOT_LOWER_QUANTILE
                )
            )

            fitted_upper = float(
                fitted_distribution.ppf(
                    PLOT_UPPER_QUANTILE
                )
            )

            if np.isfinite(fitted_lower):
                lower_candidates.append(
                    fitted_lower
                )

            if np.isfinite(fitted_upper):
                upper_candidates.append(
                    fitted_upper
                )

        except Exception:
            pass

    lower = min(
        lower_candidates
    )

    upper = max(
        upper_candidates
    )

    width = upper - lower

    if (
        not np.isfinite(width)
        or width <= 0.0
    ):
        width = 1.0

    padding = (
        PLOT_PADDING_FRACTION
        * width
    )

    return (
        lower - padding,
        upper + padding,
    )


def calculate_global_plot_limits(
    fit_store,
    event_values,
):
    """
    Obtain a shared x-axis range across all five provinces.
    """
    lower_limits = []
    upper_limits = []

    for province in SELECTED_PROVINCES:
        local_lower, local_upper = (
            calculate_local_plot_limits(
                factual_values=(
                    fit_store[
                        province
                    ][
                        "factual"
                    ][
                        "data"
                    ]
                ),
                counterfactual_values=(
                    fit_store[
                        province
                    ][
                        "counterfactual"
                    ][
                        "data"
                    ]
                ),
                factual_distribution=(
                    fit_store[
                        province
                    ][
                        "factual"
                    ][
                        "dist"
                    ]
                ),
                counterfactual_distribution=(
                    fit_store[
                        province
                    ][
                        "counterfactual"
                    ][
                        "dist"
                    ]
                ),
                event_value=event_values[
                    province
                ],
            )
        )

        lower_limits.append(
            local_lower
        )

        upper_limits.append(
            local_upper
        )

    return (
        min(lower_limits),
        max(upper_limits),
    )


# ============================================================
# 12. PDF diagnostic figure
# ============================================================

def plot_fitted_distributions(
    fit_store,
    event_values,
    shared_family,
):
    """
    Plot empirical histograms, fitted PDFs, and the real
    September 2022 event anomaly.
    """
    number_of_panels = len(
        PROVINCE_INFO
    )

    figure, axes = plt.subplots(
        number_of_panels,
        1,
        figsize=PDF_FIGSIZE,
        sharex=True,
    )

    if number_of_panels == 1:
        axes = [axes]

    global_lower, global_upper = (
        calculate_global_plot_limits(
            fit_store,
            event_values,
        )
    )

    x_grid = np.linspace(
        global_lower,
        global_upper,
        N_X_POINTS,
    )

    bins = np.linspace(
        global_lower,
        global_upper,
        N_HISTOGRAM_BINS + 1,
    )

    for panel_index, (
        province_key,
        information,
    ) in enumerate(
        PROVINCE_INFO.items()
    ):
        axis = axes[
            panel_index
        ]

        province = information[
            "csv_name"
        ]

        factual_values = fit_store[
            province
        ]["factual"]["data"]

        counterfactual_values = fit_store[
            province
        ]["counterfactual"]["data"]

        factual_distribution = fit_store[
            province
        ]["factual"]["dist"]

        counterfactual_distribution = fit_store[
            province
        ]["counterfactual"]["dist"]

        event_value = event_values[
            province
        ]

        # Panel label
        axis.text(
            0.01,
            0.97,
            information["panel"],
            transform=axis.transAxes,
            fontsize=15,
            fontweight="bold",
            verticalalignment="top",
            horizontalalignment="left",
        )

        # Province name
        axis.text(
            0.985,
            0.93,
            information[
                "display_name"
            ],
            transform=axis.transAxes,
            fontsize=14,
            fontweight="bold",
            verticalalignment="top",
            horizontalalignment="right",
        )

        # Empirical histograms
        axis.hist(
            factual_values,
            bins=bins,
            density=True,
            alpha=0.25,
            color=FACTUAL_COLOR,
            edgecolor="none",
            label="Factual empirical",
        )

        axis.hist(
            counterfactual_values,
            bins=bins,
            density=True,
            alpha=0.25,
            color=COUNTERFACTUAL_COLOR,
            edgecolor="none",
            label="Counterfactual empirical",
        )

        # Fitted PDFs
        factual_pdf = (
            factual_distribution.pdf(
                x_grid
            )
        )

        counterfactual_pdf = (
            counterfactual_distribution.pdf(
                x_grid
            )
        )

        factual_pdf = np.where(
            np.isfinite(
                factual_pdf
            ),
            factual_pdf,
            np.nan,
        )

        counterfactual_pdf = np.where(
            np.isfinite(
                counterfactual_pdf
            ),
            counterfactual_pdf,
            np.nan,
        )

        axis.plot(
            x_grid,
            factual_pdf,
            color=FACTUAL_COLOR,
            linewidth=2.2,
            label="Factual fitted",
        )

        axis.plot(
            x_grid,
            counterfactual_pdf,
            color=COUNTERFACTUAL_COLOR,
            linewidth=2.2,
            label="Counterfactual fitted",
        )

        # Actual observed September 2022 event
        axis.axvline(
            event_value,
            color=EVENT_COLOR,
            linestyle="--",
            linewidth=2.0,
            alpha=0.95,
            label=(
                "Observed 2022" # September
            ),
        )

        # Set limits before obtaining y-range for annotation.
        axis.set_xlim(
            global_lower,
            global_upper,
        )

        axis.relim()
        axis.autoscale_view(
            scalex=False,
            scaley=True,
        )

        y_maximum = axis.get_ylim()[1]


        x_offset = 0.025 * (
            global_upper - global_lower
        )

        axis.text(
            event_value + x_offset,   # move right
            y_maximum * 0.72,         # move down
            f"Event = {event_value:.2f}",
            rotation=90,
            color=EVENT_COLOR,
            fontsize=12,
            verticalalignment="top",
            horizontalalignment="left",
        )

        axis.set_ylabel(
            "Density",
            fontsize=13,
        )

        axis.tick_params(
            axis="both",
            labelsize=11,
        )

        axis.grid(
            True,
            linestyle="--",
            linewidth=0.5,
            alpha=0.45,
        )

        if panel_index == 0:
            axis.legend(
                frameon=False,
                fontsize=12,
                ncol=3,
                loc="upper left",
                bbox_to_anchor=(
                    0.08,
                    1.03,
                ),
            )

    axes[-1].set_xlabel(
        "Hydropower standardised anomaly", # Sep 2022 monthly
        fontsize=13,
    )

    # figure.suptitle(
    #     "Factual and counterfactual fitted distributions\n"
    #     f"Shared family: {shared_family}",
    #     fontsize=15,
    #     y=1.002,
    # )

    plt.tight_layout()

    output_path = (
        OUTPUT_DIRECTORY
        / (
            "fit_confirmation_shared_family_"
            "with_Sep2022_event.png"
        )
    )

    figure.savefig(
        output_path,
        dpi=FIGURE_DPI,
        bbox_inches="tight",
    )

    plt.show()
    plt.close(
        figure
    )

    print(
        "\nSaved PDF diagnostic figure:"
    )

    print(
        output_path
    )




# ============================================================
# 15. Save fitting outputs
# ============================================================

def save_fitting_outputs(
    ranked_fit_table,
    shared_summary,
    best_by_dataset,
    shared_family,
    fit_store,
    parameter_summary,
    observed_event_table,
):
    """Save all tables and fitted distribution objects."""
    all_fit_path = (
        OUTPUT_DIRECTORY
        / "all_fit_tables_10datasets.csv"
    )

    shared_summary_path = (
        OUTPUT_DIRECTORY
        / "shared_family_ranking.csv"
    )

    best_by_dataset_path = (
        OUTPUT_DIRECTORY
        / "best_distribution_by_dataset.csv"
    )

    parameter_summary_path = (
        OUTPUT_DIRECTORY
        / "shared_family_fitted_parameters.csv"
    )

    event_table_path = (
        OUTPUT_DIRECTORY
        / "observed_Sep2022_event_values.csv"
    )

    pickle_path = (
        OUTPUT_DIRECTORY
        / "fit_store_shared_family.pkl"
    )

    ranked_fit_table.to_csv(
        all_fit_path,
        index=False,
    )

    shared_summary.to_csv(
        shared_summary_path,
    )

    best_by_dataset.to_csv(
        best_by_dataset_path,
        index=False,
    )

    parameter_summary.to_csv(
        parameter_summary_path,
        index=False,
    )

    observed_event_table.to_csv(
        event_table_path,
        index=False,
    )

    serialised_object = {
        "shared_family":
        shared_family,
        "selected_provinces":
        SELECTED_PROVINCES,
        "event_year":
        EVENT_YEAR,
        "event_month":
        EVENT_MONTH,
        "observed_events":
        dict(
            zip(
                observed_event_table[
                    "province"
                ],
                observed_event_table[
                    "event_value"
                ],
            )
        ),
        "fits":
        fit_store,
    }

    with open(
        pickle_path,
        "wb",
    ) as output_file:
        pickle.dump(
            serialised_object,
            output_file,
        )

    print(
        "\n" + "=" * 65
    )
    print(
        "Saved fitting outputs"
    )
    print(
        "=" * 65
    )

    for path in [
        all_fit_path,
        shared_summary_path,
        best_by_dataset_path,
        parameter_summary_path,
        event_table_path,
        pickle_path,
    ]:
        print(
            path
        )


# ============================================================
# 16. Main
# ============================================================

def main():
    assert_directory_exists(
        OBSERVED_HYDRO_DIRECTORY
    )

    # --------------------------------------------------------
    # Load 525-member factual/counterfactual datasets
    # --------------------------------------------------------
    factual = load_ensemble_csv(
        FACTUAL_CSV,
        expected_scenario="factual",
    )

    counterfactual = load_ensemble_csv(
        COUNTERFACTUAL_CSV,
        expected_scenario="counterfactual",
    )

    validate_ensemble_datasets(
        factual,
        counterfactual,
    )

    # --------------------------------------------------------
    # Load actual observed September 2022 anomalies
    # --------------------------------------------------------
    (
        observed_event_table,
        event_values,
    ) = load_all_observed_events()

    # --------------------------------------------------------
    # Fit every candidate family to all ten datasets
    # --------------------------------------------------------
    (
        all_fit_table,
        preliminary_fit_objects,
    ) = fit_all_candidates(
        factual,
        counterfactual,
    )

    # --------------------------------------------------------
    # Select one shared family
    # --------------------------------------------------------
    (
        shared_family,
        ranked_fit_table,
        shared_summary,
        best_by_dataset,
    ) = select_shared_family(
        all_fit_table
    )

    # --------------------------------------------------------
    # Refit all ten samples using the shared family
    # --------------------------------------------------------
    (
        fit_store,
        parameter_summary,
    ) = refit_using_shared_family(
        factual,
        counterfactual,
        shared_family,
    )

    # --------------------------------------------------------
    # Save fit tables, parameters, events, and fitted objects
    # --------------------------------------------------------
    save_fitting_outputs(
        ranked_fit_table=ranked_fit_table,
        shared_summary=shared_summary,
        best_by_dataset=best_by_dataset,
        shared_family=shared_family,
        fit_store=fit_store,
        parameter_summary=parameter_summary,
        observed_event_table=observed_event_table,
    )

    # --------------------------------------------------------
    # PDF plot with observed event threshold
    # --------------------------------------------------------
    plot_fitted_distributions(
        fit_store=fit_store,
        event_values=event_values,
        shared_family=shared_family,
    )


    print(
        "\n" + "=" * 65
    )
    print(
        "Attribution Step 1 completed successfully."
    )
    print(
        "=" * 65
    )

    print(
        "Selected shared distribution family:",
        shared_family,
    )


if __name__ == "__main__":
    main()
```

```python
# Distribution fit function comparison, event threshold, attribution percentile threshold labelled
# Province order matched to CDD plot:
# Guizhou, Sichuan, Hunan, Hubei, Chongqing

import os
import pickle
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy import stats

warnings.filterwarnings("ignore", category=RuntimeWarning)

# ============================================================
# 1. User settings
# ============================================================

# Saved fit store from your hydro Step 1 fitting workflow
FIT_STORE_PATH = (
    "/home/users/weijia_6/Yangtze_code/gen_result/"
    "attribution_hydro/step1_distribution_fitting/"
    "fit_store_shared_family.pkl"
)

# Province CSVs containing observed hydro anomaly values
OBSERVED_HYDRO_DIRECTORY = (
    "/home/users/weijia_6/Yangtze_code/gen_data/"
    "hydropower_anomaly"
)

# Province metadata
# Reordered to match CDD plot:
# Guizhou -> Sichuan -> Hunan -> Hubei -> Chongqing
PROVINCE_INFO = {
    "guizhou": {
        "csv_name": "Guizhou Province",
        "display_name": "Guizhou",
        "panel": "(a)",
    },
    "sichuan": {
        "csv_name": "Sichuan Province",
        "display_name": "Sichuan",
        "panel": "(b)",
    },
    "hunan": {
        "csv_name": "Hunan Province",
        "display_name": "Hunan",
        "panel": "(c)",
    },
    "hubei": {
        "csv_name": "Hubei Province",
        "display_name": "Hubei",
        "panel": "(d)",
    },
    "chongqing": {
        "csv_name": "Chongqing Municipality",
        "display_name": "Chongqing",
        "panel": "(e)",
    },
}

SELECTED_PROVINCES = [
    info["csv_name"] for info in PROVINCE_INFO.values()
]

# Observed event definition
EVENT_YEAR = 2022
EVENT_MONTH = 9
OBSERVED_YEAR_COLUMN = "year"
OBSERVED_MONTH_COLUMN = "month"
OBSERVED_ANOMALY_COLUMN = "std_anomaly"

# Ensemble CSV columns in fit_store
EXPECTED_SCENARIOS = {"factual", "counterfactual"}

# Output directory
OUTPUT_DIRECTORY = Path(
    "/home/users/weijia_6/Yangtze_code/gen_result/"
    "attribution_hydro/step1_distribution_fitting"
)
OUTPUT_DIRECTORY.mkdir(parents=True, exist_ok=True)

# Selected percentile thresholds from your hydro attribution step
SELECTED_PERCENTILE = {
    "Guizhou Province": 5.0,
    "Sichuan Province": 5.0,
    "Hunan Province": 5.0,
    "Hubei Province": 5.0,
    "Chongqing Municipality": 5.0,
}

SELECTED_THRESHOLD = {
    "Guizhou Province":  -0.852805,
    "Sichuan Province": -0.974837,
    "Hunan Province": -0.986112,
    "Hubei Province": -0.838897,
    "Chongqing Municipality": -1.004917,
}


# Figure settings
FACTUAL_COLOR = "#CC6677"
COUNTERFACTUAL_COLOR = "#4477AA"
EVENT_COLOR = "#555555"
ATTR_COLOR = "#666666"

FIGURE_DPI = 300
PDF_FIGSIZE = (11, 3.2 * len(PROVINCE_INFO))

N_HISTOGRAM_BINS = 32
N_X_POINTS = 600

PLOT_LOWER_QUANTILE = 0.001
PLOT_UPPER_QUANTILE = 0.999
PLOT_PADDING_FRACTION = 0.08

plt.rcParams.update(
    {
        "font.family": "DejaVu Sans",
        "font.size": 13,
        "axes.unicode_minus": False,
        "axes.linewidth": 1.0,
        "xtick.direction": "out",
        "ytick.direction": "out",
    }
)

# ============================================================
# 2. General helpers
# ============================================================

def assert_file_exists(path):
    if not os.path.isfile(path):
        raise FileNotFoundError(f"File not found:\n{path}")


def assert_directory_exists(path):
    if not os.path.isdir(path):
        raise FileNotFoundError(f"Directory not found:\n{path}")


def clean_data(values):
    values = np.asarray(values, dtype=np.float64)
    return values[np.isfinite(values)]


def build_observed_file_path(province_key):
    filename = f"{province_key}_hydro_anomaly_and_std_in_reg.csv"
    return os.path.join(OBSERVED_HYDRO_DIRECTORY, filename)


# ============================================================
# 3. Load the saved fit store
# ============================================================

def load_fit_store(path):
    assert_file_exists(path)
    with open(path, "rb") as f:
        saved = pickle.load(f)

    if isinstance(saved, dict) and "fits" in saved:
        return saved

    return {
        "shared_family": None,
        "selected_provinces": SELECTED_PROVINCES,
        "event_year": EVENT_YEAR,
        "event_month": EVENT_MONTH,
        "observed_events": {},
        "fits": saved,
    }


# ============================================================
# 4. Read the real September 2022 observed hydro anomaly
# ============================================================

def load_one_observed_province(province_key, province_name):
    """
    Read one historical province CSV and extract the actual
    September 2022 hydropower standardized anomaly.
    """
    input_path = build_observed_file_path(province_key)
    assert_file_exists(input_path)

    data = pd.read_csv(input_path, encoding="utf-8-sig")
    data.columns = data.columns.astype(str).str.strip()

    required_columns = {
        OBSERVED_YEAR_COLUMN,
        OBSERVED_MONTH_COLUMN,
        OBSERVED_ANOMALY_COLUMN,
    }

    missing_columns = required_columns - set(data.columns)
    if missing_columns:
        raise KeyError(
            f"{province_name} observed CSV is missing columns: "
            f"{sorted(missing_columns)}\n"
            f"File: {input_path}\n"
            f"Available columns: {list(data.columns)}"
        )

    data = data.copy()
    for column in required_columns:
        data[column] = pd.to_numeric(data[column], errors="coerce")

    data = data.replace([np.inf, -np.inf], np.nan)

    event_rows = data.loc[
        (data[OBSERVED_YEAR_COLUMN] == EVENT_YEAR)
        & (data[OBSERVED_MONTH_COLUMN] == EVENT_MONTH)
    ].copy()

    if event_rows.empty:
        raise ValueError(
            f"No {EVENT_YEAR}-{EVENT_MONTH:02d} record was found "
            f"for {province_name}.\nFile: {input_path}"
        )

    finite_event_rows = event_rows.loc[
        np.isfinite(event_rows[OBSERVED_ANOMALY_COLUMN])
    ]

    if finite_event_rows.empty:
        raise ValueError(
            f"The {EVENT_YEAR}-{EVENT_MONTH:02d} "
            f"{OBSERVED_ANOMALY_COLUMN} value is missing for {province_name}."
        )

    if len(finite_event_rows) > 1:
        unique_values = np.unique(finite_event_rows[OBSERVED_ANOMALY_COLUMN].values)
        if len(unique_values) > 1:
            raise ValueError(
                f"Multiple different September 2022 event values were found "
                f"for {province_name}: {unique_values}"
            )
        print(
            f"[WARNING] Duplicate September 2022 rows were found for "
            f"{province_name}; using the last row."
        )

    event_value = float(finite_event_rows[OBSERVED_ANOMALY_COLUMN].iloc[-1])

    return {
        "province_key": province_key,
        "province": province_name,
        "event_year": EVENT_YEAR,
        "event_month": EVENT_MONTH,
        "event_value": event_value,
        "source_file": input_path,
    }


def load_all_observed_events():
    event_records = []

    print("\n" + "=" * 65)
    print("Observed September 2022 event values")
    print("=" * 65)

    for province_key, information in PROVINCE_INFO.items():
        record = load_one_observed_province(
            province_key=province_key,
            province_name=information["csv_name"],
        )
        event_records.append(record)
        print(f"{record['province']}: {record['event_value']:.4f}")

    event_table = pd.DataFrame(event_records)
    event_values = dict(zip(event_table["province"], event_table["event_value"]))
    return event_table, event_values


# ============================================================
# 5. Extract samples from fit_store
# ============================================================

def extract_province_sample(fit_store, province, scenario):
    sample = fit_store[province][scenario]["data"]
    return clean_data(sample)


def validate_fit_store(fit_store):
    for province in SELECTED_PROVINCES:
        if province not in fit_store:
            raise KeyError(f"Missing province in fit_store: {province}")
        for scenario in EXPECTED_SCENARIOS:
            if scenario not in fit_store[province]:
                raise KeyError(f"Missing scenario '{scenario}' for {province}")
            if "data" not in fit_store[province][scenario]:
                raise KeyError(f"Missing data for {province} / {scenario}")
            if "dist" not in fit_store[province][scenario]:
                raise KeyError(f"Missing dist for {province} / {scenario}")


# ============================================================
# 6. Plot-range helpers
# ============================================================

def calculate_local_plot_limits(
    factual_values,
    counterfactual_values,
    factual_distribution,
    counterfactual_distribution,
    event_value,
    selected_value,
):
    combined_values = np.concatenate(
        [
            factual_values,
            counterfactual_values,
            np.array([event_value, selected_value], dtype=np.float64),
        ]
    )

    lower_candidates = [
        float(np.min(combined_values)),
        float(np.quantile(combined_values, PLOT_LOWER_QUANTILE)),
        float(event_value),
        float(selected_value),
    ]

    upper_candidates = [
        float(np.max(combined_values)),
        float(np.quantile(combined_values, PLOT_UPPER_QUANTILE)),
        float(event_value),
        float(selected_value),
    ]

    for fitted_distribution in [factual_distribution, counterfactual_distribution]:
        try:
            fitted_lower = float(fitted_distribution.ppf(PLOT_LOWER_QUANTILE))
            fitted_upper = float(fitted_distribution.ppf(PLOT_UPPER_QUANTILE))

            if np.isfinite(fitted_lower):
                lower_candidates.append(fitted_lower)
            if np.isfinite(fitted_upper):
                upper_candidates.append(fitted_upper)
        except Exception:
            pass

    lower = min(lower_candidates)
    upper = max(upper_candidates)

    width = upper - lower
    if (not np.isfinite(width)) or width <= 0.0:
        width = 1.0

    padding = PLOT_PADDING_FRACTION * width
    return (lower - padding, upper + padding)


def calculate_global_plot_limits(fit_store, event_values):
    lower_limits = []
    upper_limits = []

    for province in SELECTED_PROVINCES:
        local_lower, local_upper = calculate_local_plot_limits(
            factual_values=fit_store[province]["factual"]["data"],
            counterfactual_values=fit_store[province]["counterfactual"]["data"],
            factual_distribution=fit_store[province]["factual"]["dist"],
            counterfactual_distribution=fit_store[province]["counterfactual"]["dist"],
            event_value=event_values[province],
            selected_value=SELECTED_THRESHOLD[province],
        )
        lower_limits.append(local_lower)
        upper_limits.append(local_upper)

    return (min(lower_limits), max(upper_limits))


# ============================================================
# 7. Plot the fitted distributions with both thresholds
# ============================================================

def plot_fitted_distributions(fit_store, event_values):
    number_of_panels = len(PROVINCE_INFO)
    figure, axes = plt.subplots(
        number_of_panels,
        1,
        figsize=PDF_FIGSIZE,
        sharex=True,
    )

    if number_of_panels == 1:
        axes = [axes]

    global_lower, global_upper = calculate_global_plot_limits(fit_store, event_values)

    x_grid = np.linspace(global_lower, global_upper, N_X_POINTS)
    bins = np.linspace(global_lower, global_upper, N_HISTOGRAM_BINS + 1)

    for panel_index, (province_key, information) in enumerate(PROVINCE_INFO.items()):
        axis = axes[panel_index]
        province = information["csv_name"]

        factual_values = fit_store[province]["factual"]["data"]
        counterfactual_values = fit_store[province]["counterfactual"]["data"]
        factual_distribution = fit_store[province]["factual"]["dist"]
        counterfactual_distribution = fit_store[province]["counterfactual"]["dist"]

        event_value = event_values[province]
        selected_value = SELECTED_THRESHOLD[province]
        pct = SELECTED_PERCENTILE[province]

        axis.text(
            0.01,
            0.97,
            information["panel"],
            transform=axis.transAxes,
            fontsize=15,
            fontweight="bold",
            verticalalignment="top",
            horizontalalignment="left",
        )

        axis.text(
            0.985,
            0.55,
            information["display_name"],
            transform=axis.transAxes,
            fontsize=14,
            fontweight="bold",
            verticalalignment="top",
            horizontalalignment="right",
        )

        axis.hist(
            factual_values,
            bins=bins,
            density=True,
            alpha=0.25,
            color=FACTUAL_COLOR,
            edgecolor="none",
            label="Factual empirical",
        )
        axis.hist(
            counterfactual_values,
            bins=bins,
            density=True,
            alpha=0.25,
            color=COUNTERFACTUAL_COLOR,
            edgecolor="none",
            label="Counterfactual empirical",
        )

        factual_pdf = np.where(
            np.isfinite(factual_distribution.pdf(x_grid)),
            factual_distribution.pdf(x_grid),
            np.nan,
        )
        counterfactual_pdf = np.where(
            np.isfinite(counterfactual_distribution.pdf(x_grid)),
            counterfactual_distribution.pdf(x_grid),
            np.nan,
        )

        axis.plot(
            x_grid,
            factual_pdf,
            color=FACTUAL_COLOR,
            linewidth=2.2,
            label="Factual fitted",
        )
        axis.plot(
            x_grid,
            counterfactual_pdf,
            color=COUNTERFACTUAL_COLOR,
            linewidth=2.2,
            label="Counterfactual fitted",
        )

        axis.axvline(
            event_value,
            color=EVENT_COLOR,
            linestyle="--",
            linewidth=2.0,
            alpha=0.95,
            label="Observed 2022",
        )

        axis.axvline(
            selected_value,
            color=ATTR_COLOR,
            linestyle=":",
            linewidth=2.2,
            alpha=0.95,
            label=f"Selected {pct:.0f}th percentile",
        )

        axis.set_xlim(global_lower, global_upper)
        axis.relim()
        axis.autoscale_view(scalex=False, scaley=True)
        y_maximum = axis.get_ylim()[1]

        x_offset = 0.01 * (global_upper - global_lower)

        axis.text(
            event_value + x_offset,
            y_maximum * 0.72,
            f"Event = {event_value:.2f}",
            rotation=90,
            color=EVENT_COLOR,
            fontsize=12,
            verticalalignment="top",
            horizontalalignment="left",
            fontweight="bold",
        )

        axis.text(
            selected_value + x_offset,
            y_maximum * 0.72,
            f"{pct:.0f}th pct = {selected_value:.2f}",
            rotation=90,
            color=ATTR_COLOR,
            fontsize=12,
            verticalalignment="top",
            horizontalalignment="left",
            fontweight="bold",
        )

        axis.set_ylabel("Density", fontsize=13)
        axis.tick_params(axis="both", labelsize=11)
        axis.grid(True, linestyle="--", linewidth=0.5, alpha=0.45)

        if panel_index == 0:
            axis.legend(
                frameon=False,
                fontsize=12,
                ncol=3,
                loc="upper left",
                bbox_to_anchor=(0.08, 1.03),
            )

    axes[-1].set_xlabel("Hydropower standardised anomaly", fontsize=13)

    plt.tight_layout()

    output_path = (
        OUTPUT_DIRECTORY
        / "fit_confirmation_shared_family_with_Sep2022_event_and_selected_threshold.png"
    )

    figure.savefig(
        output_path,
        dpi=FIGURE_DPI,
        bbox_inches="tight",
    )

    plt.show()
    plt.close(figure)

    print("\nSaved hydro diagnostic figure:")
    print(output_path)


# ============================================================
# 8. Main
# ============================================================

def main():
    assert_directory_exists(OBSERVED_HYDRO_DIRECTORY)

    saved = load_fit_store(FIT_STORE_PATH)
    fit_store = saved["fits"]
    validate_fit_store(fit_store)

    observed_event_table, event_values = load_all_observed_events()

    print("\nSelected percentile thresholds used in the plot:")
    for province in SELECTED_PROVINCES:
        print(
            f"{province}: "
            f"{SELECTED_PERCENTILE[province]:.0f}th pct = "
            f"{SELECTED_THRESHOLD[province]:.6f}"
        )

    plot_fitted_distributions(
        fit_store=fit_store,
        event_values=event_values,
    )

    print("\nAttribution Step 1 completed successfully.")


if __name__ == "__main__":
    main()
```

## Step 2 Risk ratio & Relative intensity change vs return periods

```python
# ============================================================
# Hydropower attribution
# Step 2: Risk ratio and relative intensity change (RIC)
# versus return period
#
# Lower-tail event:
#     unusually low September hydropower standardized anomaly
#
# Definitions
# -----------
# Factual return anomaly:
#     x_F(T) = F_F^{-1}(1 / T)
#
# Counterfactual return anomaly:
#     x_CF(T) = F_CF^{-1}(1 / T)
#
# Risk ratio at the factual return threshold:
#     RR(T) = P_F[X <= x_F(T)] / P_CF[X <= x_F(T)]
#           = (1/T) / F_CF[x_F(T)]
#
# Relative intensity change:
#     RIC(T) = 100 * (x_F(T) - x_CF(T)) / |x_F(T)|
#
# Interpretation
# --------------
# RR > 1:
#     Low-hydropower events are more likely in the factual world.
#
# RIC < 0:
#     Factual return anomaly is lower and therefore more severe.
# ============================================================

from pathlib import Path
import pickle
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

warnings.filterwarnings("ignore", category=RuntimeWarning)

# ============================================================
# 1. User settings
# ============================================================

FIT_STORE_PATH = Path(
    "/home/users/weijia_6/Yangtze_code/gen_result/"
    "attribution_hydro/step1_distribution_fitting/"
    "fit_store_shared_family.pkl"
)

OUTPUT_DIRECTORY = Path(
    "/home/users/weijia_6/Yangtze_code/gen_result/"
    "attribution_hydro/step2_rr_ric"
)

OUTPUT_DIRECTORY.mkdir(parents=True, exist_ok=True)

PROVINCES = [
    "Guizhou Province",
    "Chongqing Municipality",
    "Sichuan Province",
    "Hubei Province",
    "Hunan Province",
]

PROVINCE_ABBREVIATIONS = {
    "Guizhou Province": "GZ",
    "Chongqing Municipality": "CQ",
    "Sichuan Province": "SC",
    "Hubei Province": "HB",
    "Hunan Province": "HN",
}

PROVINCE_COLORS = {
    "Guizhou Province": "#CC6677",
    "Chongqing Municipality": "#D55E00",
    "Sichuan Province": "#009E73",
    "Hubei Province": "#0072B2",
    "Hunan Province": "#000000",
}

# ============================================================
# 2. Return-period settings
# ============================================================

# RETURN_PERIOD_MIN = 5.0 #1.01 
RETURN_PERIOD_MIN = 3.0
RETURN_PERIOD_MAX = 100.0
N_RETURN_PERIODS = 250

# We start the RIC curve at the 5-year return period because the shorter return periods are too close to zero in the denominator
# which makes the percentage change unstable and visually dominated by noise rather than a meaningful intensity signal.

RETURN_PERIOD_GRID = np.logspace(
    np.log10(RETURN_PERIOD_MIN),
    np.log10(RETURN_PERIOD_MAX),
    N_RETURN_PERIODS,
)

RETURN_PERIOD_TICKS = np.array(
    [3, 5, 10, 20, 50, 100],
    dtype=np.float64,
)

# RETURN_PERIOD_TICKS = np.array([5, 10, 20, 50, 100], dtype=np.float64)

# ============================================================
# 3. Bootstrap settings
# ============================================================

N_BOOTSTRAPS = 1000
RANDOM_SEED = 42
MIN_BOOTSTRAP_SUCCESS_FRACTION = 0.90

# ============================================================
# 4. Numerical safeguards
# ============================================================

RR_CAP = 1e6

# RIC is unstable when the factual return anomaly is close to 0.
RIC_DENOMINATOR_EPSILON = 0.25
RIC_LOWER_CAP = -500.0
RIC_UPPER_CAP = 500.0

# ============================================================
# 5. Figure settings
# ============================================================

FIGURE_SIZE = (14, 12)
FIGURE_DPI = 300

LABEL_SIZE = 18
TICK_SIZE = 15
LEGEND_SIZE = 14

LINE_WIDTH = 2.5
BAND_ALPHA = 0.20

RR_YMIN = 0.1
RR_YMAX = 100.0

RIC_YMIN = -100.0
RIC_YMAX = 50.0

# ============================================================
# 6. Load Step 1 fit store
# ============================================================

if not FIT_STORE_PATH.exists():
    raise FileNotFoundError(f"Cannot find Step 1 fit store:\n{FIT_STORE_PATH}")

with FIT_STORE_PATH.open("rb") as file_object:
    loaded_object = pickle.load(file_object)

# Support both Step 1 pickle structures:
# New:
#   {"shared_family": ..., "fits": fit_store, ...}
# Old:
#   fit_store directly
if isinstance(loaded_object, dict) and "fits" in loaded_object:
    fit_store = loaded_object["fits"]
    stored_shared_family = loaded_object.get("shared_family")
else:
    fit_store = loaded_object
    stored_shared_family = None

missing_provinces = [p for p in PROVINCES if p not in fit_store]
if missing_provinces:
    raise KeyError("Missing provinces in fit store: " + ", ".join(missing_provinces))

shared_families = {fit_store[p]["shared_family"] for p in PROVINCES}
if len(shared_families) != 1:
    raise ValueError(
        f"All provinces must use the same distribution family. Found: {sorted(shared_families)}"
    )

SHARED_FAMILY = next(iter(shared_families))

if stored_shared_family is not None and stored_shared_family != SHARED_FAMILY:
    raise ValueError(
        "Stored wrapper family and province fit families differ."
    )

if not hasattr(stats, SHARED_FAMILY):
    raise AttributeError(f"scipy.stats does not contain '{SHARED_FAMILY}'.")

distribution_family = getattr(stats, SHARED_FAMILY)
random_generator = np.random.default_rng(RANDOM_SEED)

print("Using shared distribution family:", SHARED_FAMILY)

# ============================================================
# 7. Helper functions
# ============================================================

def clean_sample(values):
    values = np.asarray(values, dtype=np.float64)
    values = values[np.isfinite(values)]
    if values.size < 10:
        raise ValueError("Too few finite ensemble values.")
    return values

def lower_tail_return_value(fitted_distribution, return_period):
    """
    x(T) = F^{-1}(1/T)
    """
    return_period = np.asarray(return_period, dtype=np.float64)
    if np.any(return_period <= 1.0):
        raise ValueError("Return periods must be greater than one.")

    cumulative_probability = 1.0 / return_period
    with np.errstate(over="ignore", under="ignore", divide="ignore", invalid="ignore"):
        return_value = fitted_distribution.ppf(cumulative_probability)
    return np.asarray(return_value, dtype=np.float64)

def lower_tail_risk_ratio(factual_distribution, counterfactual_distribution, return_period):
    """
    RR(T) = (1/T) / F_CF[x_F(T)]
    """
    return_period = np.asarray(return_period, dtype=np.float64)
    factual_threshold = lower_tail_return_value(factual_distribution, return_period)

    output = np.full(return_period.shape, np.nan, dtype=np.float64)

    valid_threshold = np.isfinite(factual_threshold)
    if not np.any(valid_threshold):
        return output

    valid_indices = np.where(valid_threshold)[0]

    with np.errstate(over="ignore", under="ignore", divide="ignore", invalid="ignore"):
        log_probability_factual = -np.log(return_period[valid_threshold])
        log_probability_counterfactual = counterfactual_distribution.logcdf(
            factual_threshold[valid_threshold]
        )
        log_risk_ratio = log_probability_factual - log_probability_counterfactual

    finite = np.isfinite(log_risk_ratio)
    if finite.any():
        finite_indices = valid_indices[finite]
        output[finite_indices] = np.exp(np.minimum(log_risk_ratio[finite], np.log(RR_CAP)))

    positive_infinite = np.isposinf(log_risk_ratio)
    if positive_infinite.any():
        output[valid_indices[positive_infinite]] = RR_CAP

    return np.clip(output, 1e-12, RR_CAP)

def relative_intensity_change(factual_distribution, counterfactual_distribution, return_period):
    """
    RIC(T) = 100 * (x_F(T) - x_CF(T)) / |x_F(T)|
    """
    factual_return = lower_tail_return_value(factual_distribution, return_period)
    counterfactual_return = lower_tail_return_value(counterfactual_distribution, return_period)

    denominator = np.abs(factual_return)
    valid = (
        np.isfinite(factual_return)
        & np.isfinite(counterfactual_return)
        & (denominator >= RIC_DENOMINATOR_EPSILON)
    )

    output = np.full(factual_return.shape, np.nan, dtype=np.float64)
    with np.errstate(over="ignore", under="ignore", divide="ignore", invalid="ignore"):
        output[valid] = (
            100.0
            * (factual_return[valid] - counterfactual_return[valid])
            / denominator[valid]
        )

    return np.clip(output, RIC_LOWER_CAP, RIC_UPPER_CAP)

def fit_fixed_family(sample):
    sample = clean_sample(sample)
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        try:
            parameters = distribution_family.fit(sample)
            frozen_distribution = distribution_family(*parameters)
        except Exception:
            return None, None

    parameters = np.asarray(parameters, dtype=np.float64)
    if not np.all(np.isfinite(parameters)):
        return None, None

    scale = parameters[-1]
    if not np.isfinite(scale) or scale <= 0.0:
        return None, None

    return frozen_distribution, parameters

def summarise_bootstrap_band(bootstrap_array):
    bootstrap_array = np.asarray(bootstrap_array, dtype=np.float64)
    if bootstrap_array.ndim != 2:
        raise ValueError("Expected bootstrap array with dimensions (bootstrap, return_period).")

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        median = np.nanpercentile(bootstrap_array, 50.0, axis=0)
        lower = np.nanpercentile(bootstrap_array, 2.5, axis=0)
        upper = np.nanpercentile(bootstrap_array, 97.5, axis=0)
        valid_count = np.sum(np.isfinite(bootstrap_array), axis=0)

    return median, lower, upper, valid_count

def format_return_anomaly(value):
    if not np.isfinite(value):
        return "NA"
    return f"{value:.2f}"

# ============================================================
# 8. Parametric bootstrap
# ============================================================

RR_MEDIAN = {}
RR_LOWER = {}
RR_UPPER = {}
RR_VALID_COUNT = {}

RIC_MEDIAN = {}
RIC_LOWER = {}
RIC_UPPER = {}
RIC_VALID_COUNT = {}

ORIGINAL_CURVES = {}
BOOTSTRAP_SUMMARY_ROWS = []
BOOTSTRAP_PARAMETER_ROWS = []

for province_number, province in enumerate(PROVINCES, start=1):
    print(f"\n[{province_number}/{len(PROVINCES)}] Processing {province}")

    factual_sample = clean_sample(fit_store[province]["factual"]["data"])
    counterfactual_sample = clean_sample(fit_store[province]["counterfactual"]["data"])

    n_factual = factual_sample.size
    n_counterfactual = counterfactual_sample.size

    print("  Factual ensemble size:", n_factual)
    print("  Counterfactual ensemble size:", n_counterfactual)

    original_factual_distribution = fit_store[province]["factual"]["dist"]
    original_counterfactual_distribution = fit_store[province]["counterfactual"]["dist"]

    ORIGINAL_CURVES[province] = {
        "rr": lower_tail_risk_ratio(
            original_factual_distribution,
            original_counterfactual_distribution,
            RETURN_PERIOD_GRID,
        ),
        "ric": relative_intensity_change(
            original_factual_distribution,
            original_counterfactual_distribution,
            RETURN_PERIOD_GRID,
        ),
    }

    rr_bootstrap = np.full((N_BOOTSTRAPS, N_RETURN_PERIODS), np.nan, dtype=np.float64)
    ric_bootstrap = np.full((N_BOOTSTRAPS, N_RETURN_PERIODS), np.nan, dtype=np.float64)

    successful_bootstraps = 0

    for bootstrap_index in range(N_BOOTSTRAPS):
        simulated_factual = original_factual_distribution.rvs(
            size=n_factual,
            random_state=random_generator,
        )
        simulated_counterfactual = original_counterfactual_distribution.rvs(
            size=n_counterfactual,
            random_state=random_generator,
        )

        bootstrap_factual_distribution, factual_parameters = fit_fixed_family(simulated_factual)
        bootstrap_counterfactual_distribution, counterfactual_parameters = fit_fixed_family(simulated_counterfactual)

        if bootstrap_factual_distribution is None or bootstrap_counterfactual_distribution is None:
            continue

        rr_bootstrap[bootstrap_index, :] = lower_tail_risk_ratio(
            bootstrap_factual_distribution,
            bootstrap_counterfactual_distribution,
            RETURN_PERIOD_GRID,
        )

        ric_bootstrap[bootstrap_index, :] = relative_intensity_change(
            bootstrap_factual_distribution,
            bootstrap_counterfactual_distribution,
            RETURN_PERIOD_GRID,
        )

        successful_bootstraps += 1

        parameter_record = {
            "province": province,
            "bootstrap": bootstrap_index,
            "distribution": SHARED_FAMILY,
        }

        for parameter_index, value in enumerate(factual_parameters):
            parameter_record[f"factual_param_{parameter_index}"] = value

        for parameter_index, value in enumerate(counterfactual_parameters):
            parameter_record[f"counterfactual_param_{parameter_index}"] = value

        BOOTSTRAP_PARAMETER_ROWS.append(parameter_record)

    success_fraction = successful_bootstraps / N_BOOTSTRAPS
    if success_fraction < MIN_BOOTSTRAP_SUCCESS_FRACTION:
        raise RuntimeError(
            f"Only {successful_bootstraps}/{N_BOOTSTRAPS} fits succeeded for {province}."
        )

    RR_MEDIAN[province], RR_LOWER[province], RR_UPPER[province], RR_VALID_COUNT[province] = summarise_bootstrap_band(rr_bootstrap)
    RIC_MEDIAN[province], RIC_LOWER[province], RIC_UPPER[province], RIC_VALID_COUNT[province] = summarise_bootstrap_band(ric_bootstrap)

    print("  Successful bootstrap fits:", f"{successful_bootstraps}/{N_BOOTSTRAPS}")

    BOOTSTRAP_SUMMARY_ROWS.append(
        {
            "province": province,
            "distribution": SHARED_FAMILY,
            "factual_sample_size": n_factual,
            "counterfactual_sample_size": n_counterfactual,
            "requested_bootstraps": N_BOOTSTRAPS,
            "successful_bootstraps": successful_bootstraps,
            "success_fraction": success_fraction,
            "minimum_valid_rr_replicates": int(np.nanmin(RR_VALID_COUNT[province])),
            "minimum_valid_ric_replicates": int(np.nanmin(RIC_VALID_COUNT[province])),
        }
    )

# ============================================================
# 9. Return-period lookup
# ============================================================

lookup_rows = []

for province in PROVINCES:
    factual_distribution = fit_store[province]["factual"]["dist"]
    counterfactual_distribution = fit_store[province]["counterfactual"]["dist"]

    factual_return_anomalies = lower_tail_return_value(factual_distribution, RETURN_PERIOD_TICKS)
    counterfactual_return_anomalies = lower_tail_return_value(counterfactual_distribution, RETURN_PERIOD_TICKS)

    risk_ratios = lower_tail_risk_ratio(factual_distribution, counterfactual_distribution, RETURN_PERIOD_TICKS)
    relative_changes = relative_intensity_change(factual_distribution, counterfactual_distribution, RETURN_PERIOD_TICKS)

    for (
        return_period,
        factual_anomaly,
        counterfactual_anomaly,
        risk_ratio,
        relative_change,
    ) in zip(
        RETURN_PERIOD_TICKS,
        factual_return_anomalies,
        counterfactual_return_anomalies,
        risk_ratios,
        relative_changes,
    ):
        lookup_rows.append(
            {
                "province": province,
                "return_period": return_period,
                "factual_return_anomaly": factual_anomaly,
                "counterfactual_return_anomaly": counterfactual_anomaly,
                "original_fit_risk_ratio": risk_ratio,
                "original_fit_ric": relative_change,
            }
        )

lookup_table = pd.DataFrame(lookup_rows)

# ============================================================
# 10. Save result tables
# ============================================================

curve_rows = []

for province in PROVINCES:
    for index, return_period in enumerate(RETURN_PERIOD_GRID):
        curve_rows.append(
            {
                "province": province,
                "return_period": return_period,

                "original_fit_rr": ORIGINAL_CURVES[province]["rr"][index],
                "rr_bootstrap_median": RR_MEDIAN[province][index],
                "rr_lower_95": RR_LOWER[province][index],
                "rr_upper_95": RR_UPPER[province][index],
                "rr_valid_bootstraps": RR_VALID_COUNT[province][index],

                "original_fit_ric": ORIGINAL_CURVES[province]["ric"][index],
                "ric_bootstrap_median": RIC_MEDIAN[province][index],
                "ric_lower_95": RIC_LOWER[province][index],
                "ric_upper_95": RIC_UPPER[province][index],
                "ric_valid_bootstraps": RIC_VALID_COUNT[province][index],
            }
        )

curve_table = pd.DataFrame(curve_rows)
curve_path = OUTPUT_DIRECTORY / "hydro_rr_ric_bootstrap_curves.csv"
curve_table.to_csv(curve_path, index=False)

lookup_path = OUTPUT_DIRECTORY / "hydro_return_period_anomaly_lookup.csv"
lookup_table.to_csv(lookup_path, index=False)

bootstrap_summary_table = pd.DataFrame(BOOTSTRAP_SUMMARY_ROWS)
summary_path = OUTPUT_DIRECTORY / "hydro_rr_ric_bootstrap_summary.csv"
bootstrap_summary_table.to_csv(summary_path, index=False)

bootstrap_parameter_table = pd.DataFrame(BOOTSTRAP_PARAMETER_ROWS)
parameter_path = OUTPUT_DIRECTORY / "hydro_rr_ric_bootstrap_parameters.csv"
bootstrap_parameter_table.to_csv(parameter_path, index=False)

print("\nSaved curves:")
print(curve_path)
print("\nSaved return-period lookup:")
print(lookup_path)
print("\nSaved bootstrap summary:")
print(summary_path)
print("\nSaved bootstrap parameters:")
print(parameter_path)

# ============================================================
# 11. Build multiline x-axis labels
# ============================================================

multiline_tick_labels = []
for return_period in RETURN_PERIOD_TICKS:
    lines = [f"{return_period:g} yr"]
    for province in PROVINCES:
        matching = lookup_table.loc[
            (lookup_table["province"] == province)
            & np.isclose(lookup_table["return_period"], return_period),
            "factual_return_anomaly",
        ]
        if matching.empty:
            anomaly_value = np.nan
        else:
            anomaly_value = matching.iloc[0]

        lines.append(
            f"{PROVINCE_ABBREVIATIONS[province]} {format_return_anomaly(anomaly_value)}"
        )
    multiline_tick_labels.append("\n".join(lines))

# ============================================================
# 12. Two-panel figure
# ============================================================

plt.rcParams.update(
    {
        "axes.labelsize": LABEL_SIZE,
        "xtick.labelsize": TICK_SIZE,
        "ytick.labelsize": TICK_SIZE,
        "legend.fontsize": LEGEND_SIZE,
    }
)

figure, axes = plt.subplots(
    2,
    1,
    figsize=FIGURE_SIZE,
    sharex=True,
    gridspec_kw={"hspace": 0.10},
)

axis_rr = axes[0]
axis_ric = axes[1]

# ------------------------------------------------------------
# Panel (a): Risk ratio
# ------------------------------------------------------------

axis_rr.text(
    0.01,
    0.91,
    "(a)",
    transform=axis_rr.transAxes,
    fontsize=20,
    fontweight="bold",
)

for province in PROVINCES:
    color = PROVINCE_COLORS[province]

    axis_rr.plot(
        RETURN_PERIOD_GRID,
        RR_MEDIAN[province],
        color=color,
        linewidth=LINE_WIDTH,
        label=province,
    )

    axis_rr.fill_between(
        RETURN_PERIOD_GRID,
        RR_LOWER[province],
        RR_UPPER[province],
        color=color,
        alpha=BAND_ALPHA,
        linewidth=0,
    )

axis_rr.axhline(
    1.0,
    color="0.35",
    linestyle=":",
    linewidth=1.2,
)

axis_rr.set_xscale("log")
axis_rr.set_yscale("log")
axis_rr.set_ylabel("Risk ratio")
axis_rr.set_xlim(RETURN_PERIOD_TICKS.min(), RETURN_PERIOD_TICKS.max())
axis_rr.set_ylim(RR_YMIN, RR_YMAX)
axis_rr.grid(True, which="both", linestyle="--", alpha=0.45)
axis_rr.tick_params(axis="x", labelbottom=False)

# ------------------------------------------------------------
# Panel (b): Relative intensity change
# ------------------------------------------------------------

axis_ric.text(
    0.01,
    0.91,
    "(b)",
    transform=axis_ric.transAxes,
    fontsize=20,
    fontweight="bold",
)

for province in PROVINCES:
    color = PROVINCE_COLORS[province]

    axis_ric.plot(
        RETURN_PERIOD_GRID,
        RIC_MEDIAN[province],
        color=color,
        linewidth=LINE_WIDTH,
    )

    axis_ric.fill_between(
        RETURN_PERIOD_GRID,
        RIC_LOWER[province],
        RIC_UPPER[province],
        color=color,
        alpha=BAND_ALPHA,
        linewidth=0,
    )

axis_ric.axhline(
    0.0,
    color="0.35",
    linestyle=":",
    linewidth=1.2,
)

axis_ric.set_xscale("log")
axis_ric.set_ylabel("Relative intensity change (%)")
axis_ric.set_xlabel(
    "Return period and corresponding factual "
    "hydropower standardised anomaly",
    labelpad=12,
)
axis_ric.set_xlim(RETURN_PERIOD_TICKS.min(), RETURN_PERIOD_TICKS.max())
axis_ric.set_ylim(RIC_YMIN, RIC_YMAX)

axis_ric.set_xticks(RETURN_PERIOD_TICKS)
axis_ric.set_xticklabels(
    multiline_tick_labels,
    fontsize=11,
    linespacing=1.25,
)

axis_ric.tick_params(axis="x", pad=8)
axis_ric.grid(True, which="both", linestyle="--", alpha=0.45)

legend_handles, legend_labels = axis_rr.get_legend_handles_labels()
axis_ric.legend(
    legend_handles,
    legend_labels,
    loc="upper right",
    ncol=3,
    frameon=False,
    fontsize=LEGEND_SIZE,
)

figure.subplots_adjust(bottom=0.22)

figure_path = OUTPUT_DIRECTORY / "hydro_rr_ric_two_panel.png"
figure.savefig(figure_path, dpi=FIGURE_DPI, bbox_inches="tight")

print("\nSaved two-panel figure:")
print(figure_path)

plt.show()
plt.close(figure)

print("\nStep 2 completed successfully.")
print("Shared family:", SHARED_FAMILY)
```

## Step 3 Spatial visualisation - point estimate result

```python
# ============================================================
# Hydropower attribution
# Step 3: Percentile-based spatial point estimates
#
# Method:
#   HadGEM fitted-distribution method only.
#
# Panels:
#   (a) Relative intensity change
#   (b) Risk ratio
#
# Lower-tail percentile selection
# --------------------------------
# For every province:
#
#   1. Test factual percentiles from the lowest tail upward:
#          P0, P5, P10, ..., P100
#
#   2. Select the first factual percentile threshold for which
#      at least MIN_COUNTERFACTUAL_EMPIRICAL_EVENTS members of
#      the counterfactual ensemble are at or below the threshold.
#
# Attribution definitions
# -----------------------
# Selected factual threshold:
#     x_thr = empirical factual percentile
#
# Lower-tail fitted probabilities:
#     p_F  = F_F(x_thr)
#     p_CF = F_CF(x_thr)
#
# Risk ratio:
#     RR = p_F / p_CF
#
# Matched-probability return anomalies:
#     x_F  = F_F^{-1}(p_F)
#     x_CF = F_CF^{-1}(p_F)
#
# Relative intensity change:
#     RIC = 100 * (x_F - x_CF) / |x_F|
#
# Interpretation:
#     RR > 1:
#         low-hydropower events are more likely in factual.
#
#     RIC < 0:
#         factual hydropower anomaly is lower and more severe.
#
# Important:
#   The selected percentile, threshold, and return period are
#   saved in the output CSV but are not printed on the figure.
# ============================================================

from pathlib import Path
import pickle
import warnings

import geopandas as gpd
import matplotlib.cm as cm
import matplotlib.colors as colors
import matplotlib.patheffects as PathEffects
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


warnings.filterwarnings(
    "ignore",
    category=RuntimeWarning,
)


# ============================================================
# 1. Paths
# ============================================================

FIT_STORE_PATH = Path(
    "/home/users/weijia_6/Yangtze_code/gen_result/"
    "attribution_hydro/step1_distribution_fitting/"
    "fit_store_shared_family.pkl"
)

CHINA_SHP_FILE = Path(
    "/home/users/weijia_6/Yangtze_code/"
    "geoBoundaries-CHN-ADM1-all/"
    "geoBoundaries-CHN-ADM1.shp"
)

YANGTZE_SHP_FILE = Path(
    "/home/users/weijia_6/Yangtze_code/"
    "liuyu/liuyu.shp"
)

OUTPUT_DIRECTORY = Path(
    "/home/users/weijia_6/Yangtze_code/gen_result/"
    "attribution_hydro/"
    "step3_percentile_spatial_point_estimates"
)

OUTPUT_DIRECTORY.mkdir(
    parents=True,
    exist_ok=True,
)

OUTPUT_FIGURE = (
    OUTPUT_DIRECTORY
    / "hydro_percentile_RIC_RR_maps.png"
)

OUTPUT_SUMMARY = (
    OUTPUT_DIRECTORY
    / "hydro_percentile_RIC_RR_point_estimates.csv"
)


# ============================================================
# 2. Province settings
# ============================================================

PROVINCES = [
    "Sichuan Province",
    "Chongqing Municipality",
    "Hubei Province",
    "Hunan Province",
    "Guizhou Province",
]

SHORT_NAMES = {
    "Sichuan Province": "Sichuan",
    "Chongqing Municipality": "Chongqing",
    "Hubei Province": "Hubei",
    "Hunan Province": "Hunan",
    "Guizhou Province": "Guizhou",
}

PROVINCE_ABBREVIATIONS = {
    "Sichuan Province": "SC",
    "Chongqing Municipality": "CQ",
    "Hubei Province": "HB",
    "Hunan Province": "HN",
    "Guizhou Province": "GZ",
}

SHAPEFILE_NAME_COLUMN = "shapeName"


# ============================================================
# 3. Percentile threshold settings
# ============================================================

# Lower-tail equivalent of the CDD Method 2 sequence.
#
# CDD upper tail:
#     P100, P95, P90, ...
#
# Hydropower lower tail:
#     P0, P5, P10, ...
CANDIDATE_PERCENTILES = list(
    range(
        0,
        101,
        5,
    )
)

# Number of empirical counterfactual values required at or
# below the selected factual threshold.
#
# 1 reproduces the simple range-overlap condition.
# 5 is more robust and avoids relying on one extreme member.
MIN_COUNTERFACTUAL_EMPIRICAL_EVENTS = 5

PROBABILITY_FLOOR = 1e-300
PROBABILITY_CEILING = 1.0 - 1e-12

# RIC is unstable if the factual matched return anomaly is
# almost zero. With a lower-tail percentile, this should rarely
# matter, but the safeguard is retained.
RIC_DENOMINATOR_EPSILON = 0.05


# ============================================================
# 4. Plot settings
# ============================================================

FIGURE_SIZE = (
    14,
    7.5,
)

FIGURE_DPI = 400

RIC_CMAP = "summer"
RR_CMAP = "cividis"

# Risk-ratio colour scale.
RR_MIN = 1.0
RR_MAX = 10.0

# RIC colour range.
#
# Set RIC_USE_DYNAMIC_RANGE=True to derive the lower limit from
# the five calculated values and keep zero as the upper limit.
#
# Set it to False to use the fixed limits below.
RIC_USE_DYNAMIC_RANGE = True

RIC_FIXED_MIN = -100.0
RIC_FIXED_MAX = 0.0

MAP_EDGE_COLOR = "black"
MAP_EDGE_WIDTH = 0.8

BASIN_EDGE_COLOR = "black"
BASIN_EDGE_WIDTH = 1.1

PROVINCE_NAME_SIZE = 12
VALUE_TEXT_SIZE = 11

PANEL_LABEL_SIZE = 18
PANEL_TITLE_SIZE = 15

COLORBAR_LABEL_SIZE = 15
COLORBAR_TICK_SIZE = 12


plt.rcParams.update(
    {
        "font.family": "DejaVu Sans",
        "axes.unicode_minus": False,
        "axes.labelsize": 14,
        "xtick.labelsize": 12,
        "ytick.labelsize": 12,
    }
)


# ============================================================
# 5. Input checks
# ============================================================

def require_file(path):
    """Require an input file to exist."""
    if not path.is_file():
        raise FileNotFoundError(
            f"File not found:\n{path}"
        )


require_file(
    FIT_STORE_PATH
)

require_file(
    CHINA_SHP_FILE
)

USE_YANGTZE_BOUNDARY = (
    YANGTZE_SHP_FILE.is_file()
)

if not USE_YANGTZE_BOUNDARY:
    print(
        "[WARNING] Yangtze basin shapefile not found."
    )

    print(
        "The figure will be generated without the basin boundary."
    )


# ============================================================
# 6. Load Step 1 fitted distributions
# ============================================================

with FIT_STORE_PATH.open("rb") as file_object:
    loaded_object = pickle.load(
        file_object
    )


# Support the current wrapped structure:
#
# {
#     "shared_family": ...,
#     "fits": fit_store,
#     ...
# }
#
# and an older structure containing fit_store directly.
if (
    isinstance(loaded_object, dict)
    and "fits" in loaded_object
):
    fit_store = loaded_object[
        "fits"
    ]

    shared_family_wrapper = loaded_object.get(
        "shared_family"
    )

else:
    fit_store = loaded_object
    shared_family_wrapper = None


missing_provinces = [
    province
    for province in PROVINCES
    if province not in fit_store
]

if missing_provinces:
    raise KeyError(
        "Missing provinces in fit store: "
        f"{missing_provinces}"
    )


shared_families = {
    fit_store[province][
        "shared_family"
    ]
    for province in PROVINCES
}

if len(shared_families) != 1:
    raise ValueError(
        "All provinces must use the same shared family. "
        f"Found: {sorted(shared_families)}"
    )


SHARED_FAMILY = next(
    iter(shared_families)
)


if (
    shared_family_wrapper is not None
    and shared_family_wrapper
    != SHARED_FAMILY
):
    raise ValueError(
        "The shared family in the pickle wrapper differs from "
        "the family stored for the province fits."
    )


print(
    "Using shared fitted family:",
    SHARED_FAMILY,
)


# ============================================================
# 7. General helpers
# ============================================================

def clean_array(values):
    """Return a finite one-dimensional array."""
    values = np.asarray(
        values,
        dtype=np.float64,
    )

    values = values[
        np.isfinite(values)
    ]

    if values.size == 0:
        raise ValueError(
            "No finite ensemble values were found."
        )

    return values


def safe_probability(probability):
    """Clip a probability to an open interval."""
    probability = float(
        probability
    )

    if not np.isfinite(
        probability
    ):
        return np.nan

    return float(
        np.clip(
            probability,
            PROBABILITY_FLOOR,
            PROBABILITY_CEILING,
        )
    )


# ============================================================
# 8. Select lower-tail percentile threshold
# ============================================================

def select_lower_tail_threshold(
    factual_values,
    counterfactual_values,
    min_counterfactual_events=MIN_COUNTERFACTUAL_EMPIRICAL_EVENTS,
    min_factual_events=1,
):
    """
    Select the lowest factual percentile threshold that still has enough
    empirical counterfactual support.

    A threshold is accepted when:
      - the factual side has at least min_factual_events at/below it, and
      - the counterfactual side has at least min_counterfactual_events at/below it.
    """
    factual_values = clean_array(factual_values)
    counterfactual_values = clean_array(counterfactual_values)

    if factual_values.size == 0 or counterfactual_values.size == 0:
        raise ValueError("Empty factual or counterfactual data.")

    for percentile in CANDIDATE_PERCENTILES:
        threshold = float(np.nanpercentile(factual_values, percentile))

        if not np.isfinite(threshold):
            continue

        factual_event_count = int(np.sum(factual_values <= threshold))
        counterfactual_event_count = int(np.sum(counterfactual_values <= threshold))

        if (
            factual_event_count >= min_factual_events
            and counterfactual_event_count >= min_counterfactual_events
        ):
            return {
                "selected_percentile": float(percentile),
                "threshold": threshold,
                "factual_event_count": factual_event_count,
                "counterfactual_event_count": counterfactual_event_count,
                "empirical_factual_probability": float(np.mean(factual_values <= threshold)),
                "empirical_counterfactual_probability": float(np.mean(counterfactual_values <= threshold)),
            }

    warnings.warn(
        "No candidate factual percentile met the empirical overlap requirement. "
        "The lowest candidate percentile will be used.",
        RuntimeWarning,
    )

    percentile = float(CANDIDATE_PERCENTILES[-1])
    threshold = float(np.nanpercentile(factual_values, percentile))

    return {
        "selected_percentile": percentile,
        "threshold": threshold,
        "factual_event_count": int(np.sum(factual_values <= threshold)),
        "counterfactual_event_count": int(np.sum(counterfactual_values <= threshold)),
        "empirical_factual_probability": float(np.mean(factual_values <= threshold)),
        "empirical_counterfactual_probability": float(np.mean(counterfactual_values <= threshold)),
    }


# ============================================================
# 9. Calculate percentile-based point estimates
# ============================================================

def compute_percentile_point_estimates(
    province,
):
    """
    Calculate lower-tail RR and RIC for one province.
    """
    factual_values = clean_array(
        fit_store[
            province
        ]["factual"]["data"]
    )

    counterfactual_values = clean_array(
        fit_store[
            province
        ]["counterfactual"]["data"]
    )

    factual_distribution = (
        fit_store[
            province
        ]["factual"]["dist"]
    )

    counterfactual_distribution = (
        fit_store[
            province
        ]["counterfactual"]["dist"]
    )

    threshold_result = (
        select_lower_tail_threshold(
            factual_values,
            counterfactual_values,
        )
    )

    selected_percentile = (
        threshold_result[
            "selected_percentile"
        ]
    )

    threshold = threshold_result[
        "threshold"
    ]

    # --------------------------------------------------------
    # Fitted lower-tail probabilities at the empirical factual
    # threshold.
    # --------------------------------------------------------

    fitted_probability_factual = safe_probability(
        factual_distribution.cdf(
            threshold
        )
    )

    fitted_probability_counterfactual = safe_probability(
        counterfactual_distribution.cdf(
            threshold
        )
    )

    # --------------------------------------------------------
    # Risk ratio
    # --------------------------------------------------------

    if (
        not np.isfinite(
            fitted_probability_factual
        )
        or fitted_probability_factual <= 0.0
    ):
        factual_return_period = np.nan
        risk_ratio = np.nan

    else:
        factual_return_period = (
            1.0
            / fitted_probability_factual
        )

        if (
            not np.isfinite(
                fitted_probability_counterfactual
            )
            or fitted_probability_counterfactual <= 0.0
        ):
            risk_ratio = np.inf

        else:
            risk_ratio = (
                fitted_probability_factual
                / fitted_probability_counterfactual
            )

    # --------------------------------------------------------
    # Matched-probability intensity comparison
    #
    # Both worlds are evaluated using the factual fitted
    # probability.
    # --------------------------------------------------------

    if np.isfinite(
        fitted_probability_factual
    ):
        factual_return_anomaly = float(
            factual_distribution.ppf(
                fitted_probability_factual
            )
        )

        counterfactual_return_anomaly = float(
            counterfactual_distribution.ppf(
                fitted_probability_factual
            )
        )

    else:
        factual_return_anomaly = np.nan
        counterfactual_return_anomaly = np.nan

    # Absolute intensity change is also saved for diagnostics.
    if (
        np.isfinite(
            factual_return_anomaly
        )
        and np.isfinite(
            counterfactual_return_anomaly
        )
    ):
        absolute_intensity_change = (
            factual_return_anomaly
            - counterfactual_return_anomaly
        )

    else:
        absolute_intensity_change = np.nan

    # --------------------------------------------------------
    # Relative intensity change
    #
    # The absolute factual anomaly is used in the denominator
    # so that a lower/more severe factual anomaly gives a
    # negative RIC.
    # --------------------------------------------------------

    if (
        np.isfinite(
            absolute_intensity_change
        )
        and np.isfinite(
            factual_return_anomaly
        )
        and abs(
            factual_return_anomaly
        ) >= RIC_DENOMINATOR_EPSILON
    ):
        relative_intensity_change = (
            100.0
            * absolute_intensity_change
            / abs(
                factual_return_anomaly
            )
        )

    else:
        relative_intensity_change = np.nan

    threshold_reconstruction_error = (
        factual_return_anomaly
        - threshold
        if np.isfinite(
            factual_return_anomaly
        )
        else np.nan
    )

    return {
        "province":
        province,

        "province_abbreviation":
        PROVINCE_ABBREVIATIONS[
            province
        ],

        "shared_family":
        SHARED_FAMILY,

        "selected_factual_percentile":
        selected_percentile,

        "selected_threshold":
        threshold,

        "factual_empirical_event_count":
        threshold_result[
            "factual_event_count"
        ],

        "counterfactual_empirical_event_count":
        threshold_result[
            "counterfactual_event_count"
        ],

        "empirical_factual_lower_tail_probability":
        threshold_result[
            "empirical_factual_probability"
        ],

        "empirical_counterfactual_lower_tail_probability":
        threshold_result[
            "empirical_counterfactual_probability"
        ],

        "fitted_factual_lower_tail_probability":
        fitted_probability_factual,

        "fitted_counterfactual_lower_tail_probability":
        fitted_probability_counterfactual,

        "factual_return_period":
        factual_return_period,

        "risk_ratio":
        risk_ratio,

        "factual_return_anomaly":
        factual_return_anomaly,

        "counterfactual_return_anomaly":
        counterfactual_return_anomaly,

        "absolute_intensity_change":
        absolute_intensity_change,

        "relative_intensity_change_percent":
        relative_intensity_change,

        "threshold_reconstruction_error":
        threshold_reconstruction_error,
    }


point_estimate_records = [
    compute_percentile_point_estimates(
        province
    )
    for province in PROVINCES
]


point_estimates = (
    pd.DataFrame(
        point_estimate_records
    )
    .set_index(
        "province"
    )
)


point_estimates.to_csv(
    OUTPUT_SUMMARY
)


print(
    "\nPercentile-based point estimates:"
)

print(
    point_estimates[
        [
            "selected_factual_percentile",
            "selected_threshold",
            "counterfactual_empirical_event_count",
            "factual_return_period",
            "relative_intensity_change_percent",
            "risk_ratio",
        ]
    ].to_string()
)


print(
    "\nSaved point-estimate table:"
)

print(
    OUTPUT_SUMMARY
)


# ============================================================
# 10. Validation checks
# ============================================================

maximum_reconstruction_error = float(
    point_estimates[
        "threshold_reconstruction_error"
    ]
    .abs()
    .max()
)


print(
    "\nMaximum threshold reconstruction error:",
    maximum_reconstruction_error,
)


if maximum_reconstruction_error > 1e-5:
    warnings.warn(
        "One or more selected thresholds were not exactly "
        "reconstructed by the factual fitted distribution.",
        RuntimeWarning,
    )


if (
    point_estimates[
        "risk_ratio"
    ]
    .isna()
    .any()
):
    warnings.warn(
        "At least one risk-ratio estimate is missing.",
        RuntimeWarning,
    )


if (
    point_estimates[
        "relative_intensity_change_percent"
    ]
    .isna()
    .any()
):
    warnings.warn(
        "At least one RIC estimate is missing. Check whether "
        "the selected factual anomaly is too close to zero.",
        RuntimeWarning,
    )


# ============================================================
# 11. Load and prepare map geometries
# ============================================================

china = gpd.read_file(
    CHINA_SHP_FILE
)


if SHAPEFILE_NAME_COLUMN not in china.columns:
    raise KeyError(
        f"Column '{SHAPEFILE_NAME_COLUMN}' was not found.\n"
        f"Available columns: {list(china.columns)}"
    )


if china.crs is None:
    raise ValueError(
        "China shapefile has no CRS."
    )


china = china.to_crs(
    "EPSG:4326"
)


china = china.loc[
    ~china[
        SHAPEFILE_NAME_COLUMN
    ].isin(
        [
            "Hong Kong Special Administrative Region",
            "Macau Special Administrative Region",
            "Taiwan Province",
        ]
    )
].copy()


selected_geometry = china.loc[
    china[
        SHAPEFILE_NAME_COLUMN
    ].isin(
        PROVINCES
    )
].copy()


missing_geometries = sorted(
    set(PROVINCES)
    - set(
        selected_geometry[
            SHAPEFILE_NAME_COLUMN
        ]
    )
)


if missing_geometries:
    raise ValueError(
        "Missing province geometries: "
        f"{missing_geometries}"
    )


invalid_geometry = (
    ~selected_geometry.geometry.is_valid
)


if invalid_geometry.any():
    selected_geometry.loc[
        invalid_geometry,
        "geometry",
    ] = (
        selected_geometry.loc[
            invalid_geometry,
            "geometry",
        ]
        .geometry
        .buffer(0)
    )


map_data = selected_geometry.merge(
    point_estimates,
    left_on=SHAPEFILE_NAME_COLUMN,
    right_index=True,
    how="left",
)


# Clip RR only for colour display.
# True values remain in the text labels and output CSV.
map_data[
    "risk_ratio_plot"
] = (
    map_data[
        "risk_ratio"
    ]
    .replace(
        [np.inf, -np.inf],
        np.nan,
    )
    .clip(
        lower=RR_MIN,
        upper=RR_MAX,
    )
)


if USE_YANGTZE_BOUNDARY:
    yangtze = gpd.read_file(
        YANGTZE_SHP_FILE
    )

    if yangtze.crs is None:
        raise ValueError(
            "Yangtze basin shapefile has no CRS."
        )

    yangtze = yangtze.to_crs(
        "EPSG:4326"
    )

else:
    yangtze = None


# ============================================================
# 12. Determine RIC colour range
# ============================================================

ric_values = pd.to_numeric(
    map_data[
        "relative_intensity_change_percent"
    ],
    errors="coerce",
)

ric_values = (
    ric_values
    .replace(
        [np.inf, -np.inf],
        np.nan,
    )
    .dropna()
)


if ric_values.empty:
    raise ValueError(
        "No valid RIC values are available for mapping."
    )


if RIC_USE_DYNAMIC_RANGE:
    observed_ric_min = float(
        ric_values.min()
    )

    observed_ric_max = float(
        ric_values.max()
    )

    # Preserve zero as the no-change reference.
    ric_vmin = min(
        observed_ric_min,
        -1.0,
    )

    ric_vmax = max(
        observed_ric_max,
        0.0,
    )

    # If all values happen to be identical, widen the range.
    if np.isclose(
        ric_vmin,
        ric_vmax,
    ):
        ric_vmin -= 1.0
        ric_vmax += 1.0

else:
    ric_vmin = RIC_FIXED_MIN
    ric_vmax = RIC_FIXED_MAX


ric_norm = colors.Normalize(
    vmin=ric_vmin,
    vmax=ric_vmax,
)


rr_norm = colors.LogNorm(
    vmin=RR_MIN,
    vmax=RR_MAX,
)


print(
    "\nMap colour ranges:"
)

print(
    f"  RIC: {ric_vmin:.2f}% to {ric_vmax:.2f}%"
)

print(
    f"  RR: {RR_MIN:g} to {RR_MAX:g}, logarithmic"
)


# ============================================================
# 13. Annotation offsets
# ============================================================

# Separate positions for province names and numerical values.
#
# Each tuple is:
#     longitude offset, latitude offset
NAME_OFFSETS = {
    "Sichuan Province": (
        0.0,
        0.45,
    ),
    "Chongqing Municipality": (
        0.20,
        -0.15,
    ),
    "Hubei Province": (
        0.0,
        0.20,
    ),
    "Hunan Province": (
        0.0,
        0.00,
    ),
    "Guizhou Province": (
        -0.10,
        -0.05,
    ),
}


VALUE_OFFSETS = {
    "Sichuan Province": (
        0.0,
        -0.55,
    ),
    "Chongqing Municipality": (
        0.20,
        -0.85,
    ),
    "Hubei Province": (
        0.0,
        -0.60,
    ),
    "Hunan Province": (
        0.0,
        -0.75,
    ),
    "Guizhou Province": (
        -0.10,
        -0.80,
    ),
}


# ============================================================
# 14. Annotation helpers
# ============================================================

def geometry_label_point(geometry):
    """Return a point guaranteed to lie inside a polygon."""
    return geometry.representative_point()


def format_metric_value(
    value,
    metric,
):
    """Format RIC and RR map values."""
    if pd.isna(
        value
    ):
        return "NA"

    if np.isposinf(
        value
    ):
        return "∞"

    if metric == "ric":
        return f"{value:.2f}%" # return f"{value:.1f}%"

    if metric == "rr":
        if value >= 100:
            return f"{value:.0f}"

        if value >= 10:
            return f"{value:.1f}"

        return f"{value:.2f}"

    return str(
        value
    )


def add_province_names(
    axis,
    geometry_data,
):
    """Add short province names."""
    for _, row in geometry_data.iterrows():
        province = row[
            SHAPEFILE_NAME_COLUMN
        ]

        point = geometry_label_point(
            row.geometry
        )

        offset_x, offset_y = (
            NAME_OFFSETS.get(
                province,
                (0.0, 0.0),
            )
        )

        axis.text(
            point.x + offset_x,
            point.y + offset_y,
            SHORT_NAMES[
                province
            ],
            fontsize=PROVINCE_NAME_SIZE,
            color="white",
            horizontalalignment="center",
            verticalalignment="center",
            fontweight="bold",
            path_effects=[
                PathEffects.withStroke(
                    linewidth=2.2,
                    foreground="black",
                )
            ],
            zorder=10,
        )


def add_metric_values(
    axis,
    geometry_data,
    value_column,
    metric,
):
    """
    Add only the RIC or RR point estimate.

    No threshold, percentile, or return-period information is
    printed in the figure.
    """
    for _, row in geometry_data.iterrows():
        province = row[
            SHAPEFILE_NAME_COLUMN
        ]

        point = geometry_label_point(
            row.geometry
        )

        offset_x, offset_y = (
            VALUE_OFFSETS.get(
                province,
                (0.0, -0.7),
            )
        )

        value = row[
            value_column
        ]

        axis.text(
            point.x + offset_x,
            point.y + offset_y,
            format_metric_value(
                value,
                metric,
            ),
            fontsize=VALUE_TEXT_SIZE,
            color="black",
            horizontalalignment="center",
            verticalalignment="center",
            fontweight="bold",
            path_effects=[
                PathEffects.withStroke(
                    linewidth=2.2,
                    foreground="white",
                )
            ],
            zorder=11,
        )


# def plot_map_panel(
#     axis,
#     geometry_data,
#     colour_column,
#     label_column,
#     cmap,
#     norm,
#     # title,
#     panel_label,
#     metric,
# ):
#     """Plot one spatial attribution panel."""
#     geometry_data.plot(
#         column=colour_column,
#         cmap=cmap,
#         norm=norm,
#         linewidth=MAP_EDGE_WIDTH,
#         edgecolor=MAP_EDGE_COLOR,
#         legend=False,
#         ax=axis,
#         missing_kwds={
#             "color": "lightgrey",
#             "edgecolor": MAP_EDGE_COLOR,
#             "hatch": "///",
#         },
#     )

#     if yangtze is not None:
#         yangtze.boundary.plot(
#             ax=axis,
#             color=BASIN_EDGE_COLOR,
#             linewidth=BASIN_EDGE_WIDTH,
#             zorder=5,
#         )

#     add_province_names(
#         axis,
#         geometry_data,
#     )

#     add_metric_values(
#         axis,
#         geometry_data,
#         value_column=label_column,
#         metric=metric,
#     )

#     axis.text(
#         0.02, # 0.015
#         1.05, # 0.985
#         panel_label,
#         transform=axis.transAxes,
#         fontsize=PANEL_LABEL_SIZE,
#         fontweight="bold",
#         horizontalalignment="left",
#         verticalalignment="top",
#         zorder=20,
#     )

#     # axis.set_title(
#     #     title,
#     #     fontsize=PANEL_TITLE_SIZE,
#     #     pad=10,
#     # )

#     axis.axis(
#         "off"
#     )


def plot_map_panel(
    axis,
    geometry_data,
    colour_column,
    label_column,
    cmap,
    norm,
    panel_label,
    metric,
    show_province_names=False,
):
    """Plot one spatial attribution panel."""
    geometry_data.plot(
        column=colour_column,
        cmap=cmap,
        norm=norm,
        linewidth=MAP_EDGE_WIDTH,
        edgecolor=MAP_EDGE_COLOR,
        legend=False,
        ax=axis,
        missing_kwds={
            "color": "lightgrey",
            "edgecolor": MAP_EDGE_COLOR,
            "hatch": "///",
        },
    )

    if yangtze is not None:
        yangtze.boundary.plot(
            ax=axis,
            color=BASIN_EDGE_COLOR,
            linewidth=BASIN_EDGE_WIDTH,
            zorder=5,
        )

    if show_province_names:
        add_province_names(axis, geometry_data)

    add_metric_values(
        axis,
        geometry_data,
        value_column=label_column,
        metric=metric,
    )

    axis.text(
        0.02,
        1.05,
        panel_label,
        transform=axis.transAxes,
        fontsize=PANEL_LABEL_SIZE,
        fontweight="bold",
        horizontalalignment="left",
        verticalalignment="top",
        zorder=20,
    )

    axis.axis("off")


# ============================================================
# 15. Two-panel map
# ============================================================

figure, axes = plt.subplots(
    1,
    2,
    figsize=FIGURE_SIZE,
)


# ------------------------------------------------------------
# Panel (a): Relative intensity change
# ------------------------------------------------------------

plot_map_panel(
    axis=axes[0],
    geometry_data=map_data,
    colour_column="relative_intensity_change_percent",
    label_column="relative_intensity_change_percent",
    cmap=RIC_CMAP,
    norm=ric_norm,
    panel_label="(c)",
    metric="ric",
    show_province_names=True,   # only first subfigure
)

# ------------------------------------------------------------
# Panel (b): Risk ratio
# ------------------------------------------------------------

plot_map_panel(
    axis=axes[1],
    geometry_data=map_data,
    colour_column="risk_ratio_plot",
    label_column="risk_ratio",
    cmap=RR_CMAP,
    norm=rr_norm,
    panel_label="(d)",
    metric="rr",
    show_province_names=False,  # no names here
)
# ============================================================
# 16. Colour bars
# ============================================================

figure.subplots_adjust(
    left=0.03,
    right=0.98,
    top=0.91,
    bottom=0.14,
    wspace=0.05,
)


# ------------------------------------------------------------
# RIC colourbar — left
# ------------------------------------------------------------

ric_mappable = cm.ScalarMappable(
    norm=ric_norm,
    cmap=RIC_CMAP,
)

ric_mappable.set_array(
    []
)

ric_colorbar = figure.colorbar(
    ric_mappable,
    ax=axes[0],
    orientation="horizontal",
    fraction=0.055,
    pad=0.08,
)

ric_colorbar.set_label(
    "Relative intensity change (%)",
    fontsize=COLORBAR_LABEL_SIZE,
)

ric_colorbar.ax.tick_params(
    labelsize=COLORBAR_TICK_SIZE,
)


# ------------------------------------------------------------
# RR colourbar — right
# ------------------------------------------------------------

rr_mappable = cm.ScalarMappable(
    norm=rr_norm,
    cmap=RR_CMAP,
)

rr_mappable.set_array(
    []
)

rr_colorbar = figure.colorbar(
    rr_mappable,
    ax=axes[1],
    orientation="horizontal",
    fraction=0.055,
    pad=0.08,
)

rr_colorbar.set_label(
    "Risk ratio",
    fontsize=COLORBAR_LABEL_SIZE,
)

rr_colorbar.ax.tick_params(
    labelsize=COLORBAR_TICK_SIZE,
)

rr_ticks = np.array(
    [1, 2, 5, 10],
    dtype=np.float64,
)

rr_ticks = rr_ticks[
    (rr_ticks >= RR_MIN)
    & (rr_ticks <= RR_MAX)
]

rr_colorbar.set_ticks(
    rr_ticks
)

rr_colorbar.set_ticklabels(
    [
        f"{tick:g}"
        for tick in rr_ticks
    ]
)


# ============================================================
# 17. Save figure
# ============================================================

figure.savefig(
    OUTPUT_FIGURE,
    dpi=FIGURE_DPI,
    bbox_inches="tight",
)

print(
    "\nSaved spatial point-estimate figure:"
)

print(
    OUTPUT_FIGURE
)

plt.show()
plt.close(
    figure
)


print(
    "\nStep 3 completed successfully."
)

print(
    "Selected percentile rule:"
)

print(
    "First low factual percentile with at least "
    f"{MIN_COUNTERFACTUAL_EMPIRICAL_EVENTS} "
    "counterfactual ensemble members at or below the threshold."
)

print(
    "\nRIC definition:"
)

print(
    "100 * (factual matched anomaly - counterfactual matched "
    "anomaly) / abs(factual matched anomaly)"
)
```

```python
# Check probabilistic result table

# ============================================================
# Hydropower attribution
# Step 3 probabilistic results:
# percentile-based lower-tail RR and RIC with parametric
# bootstrap uncertainty
#
# Point-estimate threshold selection:
#   Select the first low factual percentile P0, P5, P10, ...
#   having at least MIN_COUNTERFACTUAL_EMPIRICAL_EVENTS
#   counterfactual ensemble values at or below the threshold.
#
# Bootstrap method:
#   1. Simulate factual and counterfactual samples from the
#      original fitted distributions.
#   2. Refit the same shared distribution family.
#   3. Recalculate the factual threshold using the original
#      selected percentile.
#   4. Calculate lower-tail RR and RIC.
#   5. Summarise bootstrap median and pointwise 95% interval.
#
# Definitions:
#
#   p_F  = F_F(x_threshold)
#   p_CF = F_CF(x_threshold)
#
#   RR = p_F / p_CF
#
#   x_F  = F_F^{-1}(p_F)
#   x_CF = F_CF^{-1}(p_F)
#
#   RIC = 100 * (x_F - x_CF) / abs(x_F)
#
# Interpretation:
#   RR > 1:
#       low hydropower anomaly is more likely in factual.
#
#   RIC < 0:
#       factual anomaly is lower and more severe.
# ============================================================

from pathlib import Path
import pickle
import warnings

import numpy as np
import pandas as pd
from scipy import stats


warnings.filterwarnings(
    "ignore",
    category=RuntimeWarning,
)


# ============================================================
# 1. User settings
# ============================================================

FIT_STORE_PATH = Path(
    "/home/users/weijia_6/Yangtze_code/gen_result/"
    "attribution_hydro/step1_distribution_fitting/"
    "fit_store_shared_family.pkl"
)

OUTPUT_DIRECTORY = Path(
    "/home/users/weijia_6/Yangtze_code/gen_result/"
    "attribution_hydro/"
    "step3_percentile_spatial_point_estimates"
)

OUTPUT_DIRECTORY.mkdir(
    parents=True,
    exist_ok=True,
)


OUTPUT_SUMMARY = (
    OUTPUT_DIRECTORY
    / "hydro_percentile_RR_RIC_probabilistic_summary.csv"
)

OUTPUT_BOOTSTRAP_VALUES = (
    OUTPUT_DIRECTORY
    / "hydro_percentile_RR_RIC_bootstrap_values.csv"
)

OUTPUT_BOOTSTRAP_PARAMETERS = (
    OUTPUT_DIRECTORY
    / "hydro_percentile_RR_RIC_bootstrap_parameters.csv"
)

OUTPUT_BOOTSTRAP_DIAGNOSTICS = (
    OUTPUT_DIRECTORY
    / "hydro_percentile_RR_RIC_bootstrap_diagnostics.csv"
)


PROVINCES = [
    "Sichuan Province",
    "Chongqing Municipality",
    "Hubei Province",
    "Hunan Province",
    "Guizhou Province",
]


# ============================================================
# 2. Percentile threshold settings
# ============================================================

# Lower-tail equivalent of the CDD Method 2 sequence.
CANDIDATE_PERCENTILES = list(
    range(
        0,
        101,
        5,
    )
)

# Require at least this number of empirical counterfactual
# values at or below the original selected threshold.
MIN_COUNTERFACTUAL_EMPIRICAL_EVENTS = 5


# ============================================================
# 3. Bootstrap settings
# ============================================================

N_BOOTSTRAPS = 1000
RANDOM_SEED = 42

MIN_BOOTSTRAP_SUCCESS_FRACTION = 0.90


# ============================================================
# 4. Numerical safeguards
# ============================================================

PROBABILITY_FLOOR = 1e-300
PROBABILITY_CEILING = 1.0 - 1e-12

RR_CAP = 1e6

# RIC is undefined or unstable when the factual matched anomaly
# is close to zero.
RIC_DENOMINATOR_EPSILON = 0.05

# Display/storage safeguards for very large RIC values.
RIC_MIN = -500.0
RIC_MAX = 500.0


# ============================================================
# 5. Load Step 1 fitted distributions
# ============================================================

if not FIT_STORE_PATH.is_file():
    raise FileNotFoundError(
        f"Cannot find Step 1 fit store:\n"
        f"{FIT_STORE_PATH}"
    )


with FIT_STORE_PATH.open("rb") as file_object:
    loaded_object = pickle.load(
        file_object
    )


# Support both pickle structures:
#
# Current:
# {
#     "shared_family": ...,
#     "fits": fit_store,
#     ...
# }
#
# Older:
# fit_store directly
if (
    isinstance(loaded_object, dict)
    and "fits" in loaded_object
):
    fit_store = loaded_object[
        "fits"
    ]

    stored_shared_family = loaded_object.get(
        "shared_family"
    )

else:
    fit_store = loaded_object
    stored_shared_family = None


missing_provinces = [
    province
    for province in PROVINCES
    if province not in fit_store
]

if missing_provinces:
    raise KeyError(
        "Missing provinces in fit store: "
        + ", ".join(missing_provinces)
    )


shared_families = {
    fit_store[province][
        "shared_family"
    ]
    for province in PROVINCES
}

if len(shared_families) != 1:
    raise ValueError(
        "All provinces must use one shared distribution "
        f"family. Found: {sorted(shared_families)}"
    )


SHARED_FAMILY = next(
    iter(shared_families)
)


if (
    stored_shared_family is not None
    and stored_shared_family
    != SHARED_FAMILY
):
    raise ValueError(
        "The wrapper shared family differs from the family "
        "stored for the province fits."
    )


if not hasattr(
    stats,
    SHARED_FAMILY,
):
    raise AttributeError(
        f"scipy.stats does not contain distribution "
        f"'{SHARED_FAMILY}'."
    )


distribution_family = getattr(
    stats,
    SHARED_FAMILY,
)

random_generator = np.random.default_rng(
    RANDOM_SEED
)


print(
    "Using shared distribution family:",
    SHARED_FAMILY,
)


# ============================================================
# 6. General helpers
# ============================================================

def clean_array(values):
    """Return finite one-dimensional values."""
    values = np.asarray(
        values,
        dtype=np.float64,
    )

    values = values[
        np.isfinite(values)
    ]

    if values.size < 10:
        raise ValueError(
            "Too few finite ensemble values."
        )

    return values


def safe_probability(probability):
    """Clip probability to a numerically stable open interval."""
    probability = float(
        probability
    )

    if not np.isfinite(
        probability
    ):
        return np.nan

    return float(
        np.clip(
            probability,
            PROBABILITY_FLOOR,
            PROBABILITY_CEILING,
        )
    )


def fit_fixed_family(sample):
    """
    Refit the selected shared distribution family to one
    bootstrap sample.
    """
    sample = clean_array(
        sample
    )

    with warnings.catch_warnings():
        warnings.simplefilter(
            "ignore",
            RuntimeWarning,
        )

        try:
            parameters = distribution_family.fit(
                sample
            )

            fitted_distribution = distribution_family(
                *parameters
            )

        except Exception:
            return None, None

    parameters = np.asarray(
        parameters,
        dtype=np.float64,
    )

    if not np.all(
        np.isfinite(parameters)
    ):
        return None, None

    # The final scipy continuous-distribution parameter is scale.
    scale = parameters[-1]

    if (
        not np.isfinite(scale)
        or scale <= 0.0
    ):
        return None, None

    return (
        fitted_distribution,
        parameters,
    )


# ============================================================
# 7. Select original percentile threshold
# ============================================================

def select_lower_tail_threshold(
    factual_values,
    counterfactual_values,
):
    """
    Select the rarest low factual percentile with the required
    empirical counterfactual overlap.
    """
    factual_values = clean_array(
        factual_values
    )

    counterfactual_values = clean_array(
        counterfactual_values
    )

    for percentile in CANDIDATE_PERCENTILES:
        threshold = float(
            np.nanpercentile(
                factual_values,
                percentile,
            )
        )

        if not np.isfinite(
            threshold
        ):
            continue

        factual_event_count = int(
            np.sum(
                factual_values
                <= threshold
            )
        )

        counterfactual_event_count = int(
            np.sum(
                counterfactual_values
                <= threshold
            )
        )

        if (
            counterfactual_event_count
            >= MIN_COUNTERFACTUAL_EMPIRICAL_EVENTS
        ):
            return {
                "selected_percentile":
                float(percentile),

                "threshold":
                threshold,

                "factual_event_count":
                factual_event_count,

                "counterfactual_event_count":
                counterfactual_event_count,

                "empirical_factual_probability":
                float(
                    np.mean(
                        factual_values
                        <= threshold
                    )
                ),

                "empirical_counterfactual_probability":
                float(
                    np.mean(
                        counterfactual_values
                        <= threshold
                    )
                ),
            }

    warnings.warn(
        "No candidate percentile satisfied the empirical "
        "counterfactual-overlap requirement. Using the highest "
        "candidate percentile.",
        RuntimeWarning,
    )

    percentile = float(
        CANDIDATE_PERCENTILES[-1]
    )

    threshold = float(
        np.nanpercentile(
            factual_values,
            percentile,
        )
    )

    return {
        "selected_percentile":
        percentile,

        "threshold":
        threshold,

        "factual_event_count":
        int(
            np.sum(
                factual_values
                <= threshold
            )
        ),

        "counterfactual_event_count":
        int(
            np.sum(
                counterfactual_values
                <= threshold
            )
        ),

        "empirical_factual_probability":
        float(
            np.mean(
                factual_values
                <= threshold
            )
        ),

        "empirical_counterfactual_probability":
        float(
            np.mean(
                counterfactual_values
                <= threshold
            )
        ),
    }


# ============================================================
# 8. Calculate RR and RIC at one threshold
# ============================================================

def calculate_metrics_at_threshold(
    factual_distribution,
    counterfactual_distribution,
    threshold,
):
    """
    Calculate lower-tail RR and RIC at a supplied threshold.
    """
    probability_factual = safe_probability(
        factual_distribution.cdf(
            threshold
        )
    )

    probability_counterfactual = safe_probability(
        counterfactual_distribution.cdf(
            threshold
        )
    )

    if (
        not np.isfinite(
            probability_factual
        )
        or probability_factual <= 0.0
    ):
        return {
            "factual_probability": np.nan,
            "counterfactual_probability": np.nan,
            "factual_return_period": np.nan,
            "risk_ratio": np.nan,
            "factual_return_anomaly": np.nan,
            "counterfactual_return_anomaly": np.nan,
            "absolute_intensity_change": np.nan,
            "relative_intensity_change_percent": np.nan,
        }

    factual_return_period = (
        1.0
        / probability_factual
    )

    if (
        not np.isfinite(
            probability_counterfactual
        )
        or probability_counterfactual <= 0.0
    ):
        risk_ratio = RR_CAP

    else:
        risk_ratio = (
            probability_factual
            / probability_counterfactual
        )

        risk_ratio = float(
            np.clip(
                risk_ratio,
                1e-12,
                RR_CAP,
            )
        )

    # Matched factual probability in both worlds.
    factual_return_anomaly = float(
        factual_distribution.ppf(
            probability_factual
        )
    )

    counterfactual_return_anomaly = float(
        counterfactual_distribution.ppf(
            probability_factual
        )
    )

    if (
        np.isfinite(
            factual_return_anomaly
        )
        and np.isfinite(
            counterfactual_return_anomaly
        )
    ):
        absolute_intensity_change = (
            factual_return_anomaly
            - counterfactual_return_anomaly
        )

    else:
        absolute_intensity_change = np.nan

    if (
        np.isfinite(
            absolute_intensity_change
        )
        and np.isfinite(
            factual_return_anomaly
        )
        and abs(
            factual_return_anomaly
        ) >= RIC_DENOMINATOR_EPSILON
    ):
        relative_intensity_change = (
            100.0
            * absolute_intensity_change
            / abs(
                factual_return_anomaly
            )
        )

        relative_intensity_change = float(
            np.clip(
                relative_intensity_change,
                RIC_MIN,
                RIC_MAX,
            )
        )

    else:
        relative_intensity_change = np.nan

    return {
        "factual_probability":
        probability_factual,

        "counterfactual_probability":
        probability_counterfactual,

        "factual_return_period":
        factual_return_period,

        "risk_ratio":
        risk_ratio,

        "factual_return_anomaly":
        factual_return_anomaly,

        "counterfactual_return_anomaly":
        counterfactual_return_anomaly,

        "absolute_intensity_change":
        absolute_intensity_change,

        "relative_intensity_change_percent":
        relative_intensity_change,
    }


# ============================================================
# 9. Bootstrap summary helper
# ============================================================

def summarise_bootstrap(values):
    """
    Return bootstrap median, 2.5th percentile, 97.5th
    percentile, and valid count.
    """
    values = np.asarray(
        values,
        dtype=np.float64,
    )

    finite_values = values[
        np.isfinite(values)
    ]

    if finite_values.size == 0:
        return {
            "median": np.nan,
            "lower_95": np.nan,
            "upper_95": np.nan,
            "valid_count": 0,
        }

    return {
        "median": float(
            np.nanpercentile(
                finite_values,
                50.0,
            )
        ),

        "lower_95": float(
            np.nanpercentile(
                finite_values,
                2.5,
            )
        ),

        "upper_95": float(
            np.nanpercentile(
                finite_values,
                97.5,
            )
        ),

        "valid_count": int(
            finite_values.size
        ),
    }


# ============================================================
# 10. Point estimates and parametric bootstrap
# ============================================================

summary_records = []
bootstrap_value_records = []
bootstrap_parameter_records = []
diagnostic_records = []


for province_number, province in enumerate(
    PROVINCES,
    start=1,
):
    print(
        f"\n[{province_number}/{len(PROVINCES)}] "
        f"Processing {province}"
    )

    factual_values = clean_array(
        fit_store[
            province
        ]["factual"]["data"]
    )

    counterfactual_values = clean_array(
        fit_store[
            province
        ]["counterfactual"]["data"]
    )

    original_factual_distribution = (
        fit_store[
            province
        ]["factual"]["dist"]
    )

    original_counterfactual_distribution = (
        fit_store[
            province
        ]["counterfactual"]["dist"]
    )

    n_factual = factual_values.size
    n_counterfactual = (
        counterfactual_values.size
    )

    # --------------------------------------------------------
    # Original percentile selection
    # --------------------------------------------------------

    threshold_selection = (
        select_lower_tail_threshold(
            factual_values,
            counterfactual_values,
        )
    )

    selected_percentile = (
        threshold_selection[
            "selected_percentile"
        ]
    )

    original_threshold = (
        threshold_selection[
            "threshold"
        ]
    )

    print(
        "  Selected factual percentile:",
        selected_percentile,
    )

    print(
        "  Original selected threshold:",
        original_threshold,
    )

    print(
        "  Counterfactual empirical event count:",
        threshold_selection[
            "counterfactual_event_count"
        ],
    )

    # --------------------------------------------------------
    # Original point estimates
    # --------------------------------------------------------

    original_metrics = (
        calculate_metrics_at_threshold(
            factual_distribution=(
                original_factual_distribution
            ),
            counterfactual_distribution=(
                original_counterfactual_distribution
            ),
            threshold=original_threshold,
        )
    )

    print(
        "  Point RR:",
        original_metrics[
            "risk_ratio"
        ],
    )

    print(
        "  Point RIC:",
        original_metrics[
            "relative_intensity_change_percent"
        ],
    )

    # --------------------------------------------------------
    # Bootstrap storage
    # --------------------------------------------------------

    rr_bootstrap = np.full(
        N_BOOTSTRAPS,
        np.nan,
        dtype=np.float64,
    )

    ric_bootstrap = np.full(
        N_BOOTSTRAPS,
        np.nan,
        dtype=np.float64,
    )

    absolute_change_bootstrap = np.full(
        N_BOOTSTRAPS,
        np.nan,
        dtype=np.float64,
    )

    threshold_bootstrap = np.full(
        N_BOOTSTRAPS,
        np.nan,
        dtype=np.float64,
    )

    return_period_bootstrap = np.full(
        N_BOOTSTRAPS,
        np.nan,
        dtype=np.float64,
    )

    successful_fits = 0

    # --------------------------------------------------------
    # Parametric bootstrap
    # --------------------------------------------------------

    for bootstrap_index in range(
        N_BOOTSTRAPS
    ):
        simulated_factual = (
            original_factual_distribution.rvs(
                size=n_factual,
                random_state=random_generator,
            )
        )

        simulated_counterfactual = (
            original_counterfactual_distribution.rvs(
                size=n_counterfactual,
                random_state=random_generator,
            )
        )

        (
            bootstrap_factual_distribution,
            factual_parameters,
        ) = fit_fixed_family(
            simulated_factual
        )

        (
            bootstrap_counterfactual_distribution,
            counterfactual_parameters,
        ) = fit_fixed_family(
            simulated_counterfactual
        )

        if (
            bootstrap_factual_distribution is None
            or bootstrap_counterfactual_distribution
            is None
        ):
            continue

        # Preserve the originally selected percentile, but
        # recalculate its threshold in the bootstrap factual
        # sample.
        bootstrap_threshold = float(
            np.nanpercentile(
                simulated_factual,
                selected_percentile,
            )
        )

        if not np.isfinite(
            bootstrap_threshold
        ):
            continue

        bootstrap_metrics = (
            calculate_metrics_at_threshold(
                factual_distribution=(
                    bootstrap_factual_distribution
                ),
                counterfactual_distribution=(
                    bootstrap_counterfactual_distribution
                ),
                threshold=bootstrap_threshold,
            )
        )

        rr_bootstrap[
            bootstrap_index
        ] = bootstrap_metrics[
            "risk_ratio"
        ]

        ric_bootstrap[
            bootstrap_index
        ] = bootstrap_metrics[
            "relative_intensity_change_percent"
        ]

        absolute_change_bootstrap[
            bootstrap_index
        ] = bootstrap_metrics[
            "absolute_intensity_change"
        ]

        threshold_bootstrap[
            bootstrap_index
        ] = bootstrap_threshold

        return_period_bootstrap[
            bootstrap_index
        ] = bootstrap_metrics[
            "factual_return_period"
        ]

        successful_fits += 1

        bootstrap_value_records.append(
            {
                "province":
                province,

                "bootstrap":
                bootstrap_index,

                "selected_percentile":
                selected_percentile,

                "bootstrap_threshold":
                bootstrap_threshold,

                "factual_probability":
                bootstrap_metrics[
                    "factual_probability"
                ],

                "counterfactual_probability":
                bootstrap_metrics[
                    "counterfactual_probability"
                ],

                "factual_return_period":
                bootstrap_metrics[
                    "factual_return_period"
                ],

                "risk_ratio":
                bootstrap_metrics[
                    "risk_ratio"
                ],

                "factual_return_anomaly":
                bootstrap_metrics[
                    "factual_return_anomaly"
                ],

                "counterfactual_return_anomaly":
                bootstrap_metrics[
                    "counterfactual_return_anomaly"
                ],

                "absolute_intensity_change":
                bootstrap_metrics[
                    "absolute_intensity_change"
                ],

                "relative_intensity_change_percent":
                bootstrap_metrics[
                    "relative_intensity_change_percent"
                ],
            }
        )

        parameter_record = {
            "province":
            province,

            "bootstrap":
            bootstrap_index,

            "distribution":
            SHARED_FAMILY,
        }

        for parameter_index, parameter_value in enumerate(
            factual_parameters
        ):
            parameter_record[
                f"factual_param_{parameter_index}"
            ] = parameter_value

        for parameter_index, parameter_value in enumerate(
            counterfactual_parameters
        ):
            parameter_record[
                f"counterfactual_param_{parameter_index}"
            ] = parameter_value

        bootstrap_parameter_records.append(
            parameter_record
        )

    # --------------------------------------------------------
    # Bootstrap quality check
    # --------------------------------------------------------

    success_fraction = (
        successful_fits
        / N_BOOTSTRAPS
    )

    if (
        success_fraction
        < MIN_BOOTSTRAP_SUCCESS_FRACTION
    ):
        raise RuntimeError(
            f"{province}: only "
            f"{successful_fits}/{N_BOOTSTRAPS} "
            "bootstrap fits succeeded."
        )

    rr_summary = summarise_bootstrap(
        rr_bootstrap
    )

    ric_summary = summarise_bootstrap(
        ric_bootstrap
    )

    absolute_change_summary = (
        summarise_bootstrap(
            absolute_change_bootstrap
        )
    )

    threshold_summary = summarise_bootstrap(
        threshold_bootstrap
    )

    return_period_summary = (
        summarise_bootstrap(
            return_period_bootstrap
        )
    )

    print(
        "  Successful bootstrap fits:",
        f"{successful_fits}/{N_BOOTSTRAPS}",
    )

    print(
        "  RR bootstrap median and 95% interval:",
        rr_summary["median"],
        rr_summary["lower_95"],
        rr_summary["upper_95"],
    )

    print(
        "  RIC bootstrap median and 95% interval:",
        ric_summary["median"],
        ric_summary["lower_95"],
        ric_summary["upper_95"],
    )

    # --------------------------------------------------------
    # Province summary row
    # --------------------------------------------------------

    summary_records.append(
        {
            "Province":
            province,

            "Distribution":
            SHARED_FAMILY,

            "Selected_percentile":
            selected_percentile,

            "Original_threshold":
            original_threshold,

            "Original_factual_empirical_event_count":
            threshold_selection[
                "factual_event_count"
            ],

            "Original_counterfactual_empirical_event_count":
            threshold_selection[
                "counterfactual_event_count"
            ],

            "Original_empirical_factual_probability":
            threshold_selection[
                "empirical_factual_probability"
            ],

            "Original_empirical_counterfactual_probability":
            threshold_selection[
                "empirical_counterfactual_probability"
            ],

            "Threshold_bootstrap_median":
            threshold_summary[
                "median"
            ],

            "Threshold_2.5%":
            threshold_summary[
                "lower_95"
            ],

            "Threshold_97.5%":
            threshold_summary[
                "upper_95"
            ],

            "Factual_return_period_point":
            original_metrics[
                "factual_return_period"
            ],

            "Factual_return_period_bootstrap_median":
            return_period_summary[
                "median"
            ],

            "Factual_return_period_2.5%":
            return_period_summary[
                "lower_95"
            ],

            "Factual_return_period_97.5%":
            return_period_summary[
                "upper_95"
            ],

            # Point estimate from the original fitted
            # distributions.
            "RR_point":
            original_metrics[
                "risk_ratio"
            ],

            # Bootstrap distribution summary.
            "RR_bootstrap_median":
            rr_summary[
                "median"
            ],

            "RR_2.5%":
            rr_summary[
                "lower_95"
            ],

            "RR_97.5%":
            rr_summary[
                "upper_95"
            ],

            "RR_valid_bootstraps":
            rr_summary[
                "valid_count"
            ],

            "RIC_point":
            original_metrics[
                "relative_intensity_change_percent"
            ],

            "RIC_bootstrap_median":
            ric_summary[
                "median"
            ],

            "RIC_2.5%":
            ric_summary[
                "lower_95"
            ],

            "RIC_97.5%":
            ric_summary[
                "upper_95"
            ],

            "RIC_valid_bootstraps":
            ric_summary[
                "valid_count"
            ],

            "Absolute_intensity_change_point":
            original_metrics[
                "absolute_intensity_change"
            ],

            "Absolute_intensity_change_bootstrap_median":
            absolute_change_summary[
                "median"
            ],

            "Absolute_intensity_change_2.5%":
            absolute_change_summary[
                "lower_95"
            ],

            "Absolute_intensity_change_97.5%":
            absolute_change_summary[
                "upper_95"
            ],

            "Requested_bootstraps":
            N_BOOTSTRAPS,

            "Successful_bootstraps":
            successful_fits,

            "Bootstrap_success_fraction":
            success_fraction,
        }
    )

    diagnostic_records.append(
        {
            "Province":
            province,

            "Distribution":
            SHARED_FAMILY,

            "Factual_sample_size":
            n_factual,

            "Counterfactual_sample_size":
            n_counterfactual,

            "Selected_percentile":
            selected_percentile,

            "Requested_bootstraps":
            N_BOOTSTRAPS,

            "Successful_bootstraps":
            successful_fits,

            "Bootstrap_success_fraction":
            success_fraction,

            "Valid_RR_bootstraps":
            rr_summary[
                "valid_count"
            ],

            "Valid_RIC_bootstraps":
            ric_summary[
                "valid_count"
            ],

            "RIC_denominator_epsilon":
            RIC_DENOMINATOR_EPSILON,
        }
    )


# ============================================================
# 11. Build and save summary tables
# ============================================================

df_m2_summary = pd.DataFrame(
    summary_records
).set_index(
    "Province"
)


df_bootstrap_values = pd.DataFrame(
    bootstrap_value_records
)


df_bootstrap_parameters = pd.DataFrame(
    bootstrap_parameter_records
)


df_bootstrap_diagnostics = pd.DataFrame(
    diagnostic_records
).set_index(
    "Province"
)


df_m2_summary.to_csv(
    OUTPUT_SUMMARY
)

df_bootstrap_values.to_csv(
    OUTPUT_BOOTSTRAP_VALUES,
    index=False,
)

df_bootstrap_parameters.to_csv(
    OUTPUT_BOOTSTRAP_PARAMETERS,
    index=False,
)

df_bootstrap_diagnostics.to_csv(
    OUTPUT_BOOTSTRAP_DIAGNOSTICS
)


# ============================================================
# 12. Print compact probabilistic summary
# ============================================================

summary_table = pd.DataFrame(
    index=PROVINCES
)


summary_table[
    "M2_selected_percentile"
] = df_m2_summary[
    "Selected_percentile"
]


summary_table[
    "M2_threshold"
] = df_m2_summary[
    "Original_threshold"
]


summary_table[
    "M2_RR_point"
] = df_m2_summary[
    "RR_point"
]

summary_table[
    "M2_RR_bootstrap_median"
] = df_m2_summary[
    "RR_bootstrap_median"
]

summary_table[
    "M2_RR_2.5%"
] = df_m2_summary[
    "RR_2.5%"
]

summary_table[
    "M2_RR_97.5%"
] = df_m2_summary[
    "RR_97.5%"
]


summary_table[
    "M2_RIC_point"
] = df_m2_summary[
    "RIC_point"
]

summary_table[
    "M2_RIC_bootstrap_median"
] = df_m2_summary[
    "RIC_bootstrap_median"
]

summary_table[
    "M2_RIC_2.5%"
] = df_m2_summary[
    "RIC_2.5%"
]

summary_table[
    "M2_RIC_97.5%"
] = df_m2_summary[
    "RIC_97.5%"
]


summary_table[
    "M2_successful_bootstraps"
] = df_m2_summary[
    "Successful_bootstraps"
]


summary_table.index.name = "Province"


summary_table_print = (
    summary_table
    .copy()
    .round(3)
)


print(
    "\n" + "=" * 90
)

print(
    "Hydropower percentile-based probabilistic attribution summary"
)

print(
    "=" * 90
)

print(
    summary_table_print.to_string()
)


compact_summary_path = (
    OUTPUT_DIRECTORY
    / "province_level_M2_RR_RIC_probabilistic_summary.csv"
)

summary_table.to_csv(
    compact_summary_path
)


print(
    "\nSaved full summary:"
)

print(
    OUTPUT_SUMMARY
)

print(
    "\nSaved compact summary:"
)

print(
    compact_summary_path
)

print(
    "\nSaved bootstrap values:"
)

print(
    OUTPUT_BOOTSTRAP_VALUES
)

print(
    "\nSaved bootstrap parameters:"
)

print(
    OUTPUT_BOOTSTRAP_PARAMETERS
)

print(
    "\nSaved bootstrap diagnostics:"
)

print(
    OUTPUT_BOOTSTRAP_DIAGNOSTICS
)


print(
    "\nStep 3 probabilistic analysis completed successfully."
)
```

## Step 4 Return period plots

```python
# ============================================================
# Hydropower attribution
# Step 4: Factual and counterfactual return-period curves
#
# Five-province square-panel layout:
#
#     Sichuan     Chongqing     Hubei
#           Hunan          Guizhou
#
# Lower-tail return anomaly:
#
#     x(T) = F^{-1}(1 / T)
#
# Larger return periods correspond to rarer and lower
# hydropower standardized anomalies.
#
# Curves:
#     Red  = factual climate
#     Blue = counterfactual climate
#
# Shading:
#     Pointwise 95% parametric-bootstrap interval
#
# Horizontal dashed line:
#     Observed September 2022 hydropower standardized anomaly
# ============================================================

from __future__ import annotations

from pathlib import Path
import pickle
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy import stats


warnings.filterwarnings(
    "ignore",
    category=RuntimeWarning,
)


# ============================================================
# 1. Paths
# ============================================================

FIT_STORE_PATH = Path(
    "/home/users/weijia_6/Yangtze_code/gen_result/"
    "attribution_hydro/step1_distribution_fitting/"
    "fit_store_shared_family.pkl"
)

OBSERVED_HYDRO_DIRECTORY = Path(
    "/home/users/weijia_6/Yangtze_code/"
    "gen_data/hydropower_anomaly"
)

OUTPUT_DIRECTORY = Path(
    "/home/users/weijia_6/Yangtze_code/gen_result/"
    "attribution_hydro/step4_return_period_curves"
)

OUTPUT_DIRECTORY.mkdir(
    parents=True,
    exist_ok=True,
)

OUTPUT_FIGURE = (
    OUTPUT_DIRECTORY
    / "hydro_five_province_return_period_curves_square.png"
)

OUTPUT_CURVES = (
    OUTPUT_DIRECTORY
    / "hydro_return_period_bootstrap_curves.csv"
)

OUTPUT_SUMMARY = (
    OUTPUT_DIRECTORY
    / "hydro_return_period_bootstrap_summary.csv"
)


# ============================================================
# 2. Province settings
# ============================================================

PROVINCES = [
    "Sichuan Province",
    "Chongqing Municipality",
    "Hubei Province",
    "Hunan Province",
    "Guizhou Province",
]

PROVINCE_SHORT = {
    "Sichuan Province": "Sichuan",
    "Chongqing Municipality": "Chongqing",
    "Hubei Province": "Hubei",
    "Hunan Province": "Hunan",
    "Guizhou Province": "Guizhou",
}

PROVINCE_FILE_KEYS = {
    "Sichuan Province": "sichuan",
    "Chongqing Municipality": "chongqing",
    "Hubei Province": "hubei",
    "Hunan Province": "hunan",
    "Guizhou Province": "guizhou",
}

PANEL_LABELS = {
    "Sichuan Province": "(a)",
    "Chongqing Municipality": "(b)",
    "Hubei Province": "(c)",
    "Hunan Province": "(d)",
    "Guizhou Province": "(e)",
}


# ============================================================
# 3. Event settings
# ============================================================

EVENT_YEAR = 2022
EVENT_MONTH = 9

OBSERVED_ANOMALY_COLUMN = "std_anomaly"


# ============================================================
# 4. Return-period settings
# ============================================================

# Do not begin exactly at T = 1 because p = 1/T would equal 1.
RETURN_PERIOD_MIN = 1.01
RETURN_PERIOD_MAX = 10000.0
N_RETURN_PERIODS = 400

RETURN_PERIOD_GRID = np.logspace(
    np.log10(RETURN_PERIOD_MIN),
    np.log10(RETURN_PERIOD_MAX),
    N_RETURN_PERIODS,
)

# Reduced tick set for compact square panels.
RETURN_PERIOD_TICKS = np.array(
    [
        2,
        10,
        100,
        1000,
        10000,
    ],
    dtype=np.float64,
)


# ============================================================
# 5. Bootstrap settings
# ============================================================

N_BOOTSTRAPS = 1000
RANDOM_SEED = 42

MIN_BOOTSTRAP_SUCCESS_FRACTION = 0.90


# ============================================================
# 6. Plot settings
# ============================================================

# Slightly taller figure to accommodate square plotting boxes.
FIGURE_SIZE = (
    15,
    10.5,
)

FIGURE_DPI = 350

FACTUAL_COLOR = "#d62728"
COUNTERFACTUAL_COLOR = "#1f77b4"
EVENT_COLOR = "0.30"

LINE_WIDTH = 2.5
EVENT_LINE_WIDTH = 1.5
BAND_ALPHA = 0.20

PANEL_LABEL_SIZE = 19
PROVINCE_NAME_SIZE = 17

TICK_SIZE = 15
X_LABEL_SIZE = 20
LEGEND_SIZE = 17

GRID_ALPHA = 0.35

# Background behind province name.
PROVINCE_BOX_ALPHA = 0.72


plt.rcParams.update(
    {
        "xtick.labelsize": TICK_SIZE,
        "ytick.labelsize": TICK_SIZE,
        "legend.fontsize": LEGEND_SIZE,
        "axes.unicode_minus": False,
    }
)


# ============================================================
# 7. Optional fixed y-axis limits
# ============================================================

# False:
#     determine y-axis limits separately for each province.
#
# True:
#     use CUSTOM_Y_LIMITS below.
USE_CUSTOM_Y_LIMITS = False


CUSTOM_Y_LIMITS = {
    "Sichuan Province": (
        -3.5,
        2.0,
    ),
    "Chongqing Municipality": (
        -3.5,
        2.0,
    ),
    "Hubei Province": (
        -3.5,
        2.0,
    ),
    "Hunan Province": (
        -3.5,
        2.0,
    ),
    "Guizhou Province": (
        -3.5,
        2.0,
    ),
}


# ============================================================
# 8. Input checks
# ============================================================

if not FIT_STORE_PATH.is_file():
    raise FileNotFoundError(
        f"Cannot find Step 1 fit store:\n"
        f"{FIT_STORE_PATH}"
    )

if not OBSERVED_HYDRO_DIRECTORY.is_dir():
    raise FileNotFoundError(
        "Cannot find observed hydropower directory:\n"
        f"{OBSERVED_HYDRO_DIRECTORY}"
    )


# ============================================================
# 9. Load Step 1 fitted distributions
# ============================================================

with FIT_STORE_PATH.open(
    "rb"
) as file_object:
    loaded_object = pickle.load(
        file_object
    )


# Support both pickle structures:
#
# Wrapped:
#     {
#         "shared_family": ...,
#         "fits": fit_store,
#         "observed_events": ...
#     }
#
# Direct:
#     fit_store
if (
    isinstance(
        loaded_object,
        dict,
    )
    and "fits" in loaded_object
):
    fit_store = loaded_object[
        "fits"
    ]

    stored_shared_family = loaded_object.get(
        "shared_family"
    )

    stored_observed_events = loaded_object.get(
        "observed_events",
        {},
    )

else:
    fit_store = loaded_object
    stored_shared_family = None
    stored_observed_events = {}


missing_provinces = [
    province
    for province in PROVINCES
    if province not in fit_store
]

if missing_provinces:
    raise KeyError(
        "Missing provinces in fit store: "
        + ", ".join(
            missing_provinces
        )
    )


shared_families = {
    fit_store[
        province
    ]["shared_family"]
    for province in PROVINCES
}

if len(
    shared_families
) != 1:
    raise ValueError(
        "All provinces must use one shared fitted family. "
        f"Found: {sorted(shared_families)}"
    )


SHARED_FAMILY = next(
    iter(
        shared_families
    )
)


if (
    stored_shared_family is not None
    and stored_shared_family
    != SHARED_FAMILY
):
    raise ValueError(
        "The wrapper distribution family differs from the "
        "province-level fit records."
    )


if not hasattr(
    stats,
    SHARED_FAMILY,
):
    raise AttributeError(
        f"scipy.stats does not contain "
        f"'{SHARED_FAMILY}'."
    )


distribution_family = getattr(
    stats,
    SHARED_FAMILY,
)

random_generator = np.random.default_rng(
    RANDOM_SEED
)


print(
    "Using shared distribution family:",
    SHARED_FAMILY,
)


# ============================================================
# 10. Load observed September 2022 event values
# ============================================================

def load_observed_event(
    province,
):
    """
    Load the actual September 2022 hydropower standardized
    anomaly for one province.
    """
    if (
        province
        in stored_observed_events
        and np.isfinite(
            stored_observed_events[
                province
            ]
        )
    ):
        return float(
            stored_observed_events[
                province
            ]
        )

    province_key = PROVINCE_FILE_KEYS[
        province
    ]

    input_file = (
        OBSERVED_HYDRO_DIRECTORY
        / (
            f"{province_key}_"
            "hydro_anomaly_and_std_in_reg.csv"
        )
    )

    if not input_file.is_file():
        raise FileNotFoundError(
            "Observed hydropower file not found for "
            f"{province}:\n{input_file}"
        )

    data = pd.read_csv(
        input_file,
        encoding="utf-8-sig",
    )

    data.columns = (
        data.columns
        .astype(str)
        .str.strip()
    )

    required_columns = {
        "year",
        "month",
        OBSERVED_ANOMALY_COLUMN,
    }

    missing_columns = (
        required_columns
        - set(
            data.columns
        )
    )

    if missing_columns:
        raise KeyError(
            f"{province} file is missing columns: "
            f"{sorted(missing_columns)}"
        )

    data[
        "year"
    ] = pd.to_numeric(
        data["year"],
        errors="coerce",
    )

    data[
        "month"
    ] = pd.to_numeric(
        data["month"],
        errors="coerce",
    )

    data[
        OBSERVED_ANOMALY_COLUMN
    ] = pd.to_numeric(
        data[
            OBSERVED_ANOMALY_COLUMN
        ],
        errors="coerce",
    )

    event_rows = data.loc[
        (
            data["year"]
            == EVENT_YEAR
        )
        & (
            data["month"]
            == EVENT_MONTH
        )
        & np.isfinite(
            data[
                OBSERVED_ANOMALY_COLUMN
            ]
        )
    ]

    if event_rows.empty:
        raise ValueError(
            f"No valid {EVENT_YEAR}-{EVENT_MONTH:02d} "
            f"event value found for {province}."
        )

    return float(
        event_rows[
            OBSERVED_ANOMALY_COLUMN
        ].iloc[-1]
    )


OBSERVED_EVENTS = {
    province:
    load_observed_event(
        province
    )
    for province in PROVINCES
}


print(
    "\nObserved September 2022 anomalies:"
)

for province in PROVINCES:
    print(
        f"  {province}: "
        f"{OBSERVED_EVENTS[province]:.4f}"
    )


# ============================================================
# 11. Helper functions
# ============================================================

def clean_array(
    values,
):
    """Return finite one-dimensional values."""
    values = np.asarray(
        values,
        dtype=np.float64,
    )

    values = values[
        np.isfinite(
            values
        )
    ]

    if values.size < 10:
        raise ValueError(
            "Too few finite ensemble values."
        )

    return values


def lower_tail_return_value(
    fitted_distribution,
    return_period,
):
    """
    Calculate the lower-tail return anomaly:

        x(T) = F^{-1}(1/T)
    """
    return_period = np.asarray(
        return_period,
        dtype=np.float64,
    )

    if np.any(
        return_period <= 1.0
    ):
        raise ValueError(
            "Return periods must be greater than one."
        )

    cumulative_probability = (
        1.0
        / return_period
    )

    with np.errstate(
        over="ignore",
        under="ignore",
        divide="ignore",
        invalid="ignore",
    ):
        return_value = (
            fitted_distribution.ppf(
                cumulative_probability
            )
        )

    return np.asarray(
        return_value,
        dtype=np.float64,
    )


def fit_shared_family(
    sample,
):
    """
    Refit the fixed shared distribution family.
    """
    sample = clean_array(
        sample
    )

    with warnings.catch_warnings():
        warnings.simplefilter(
            "ignore",
            RuntimeWarning,
        )

        try:
            fitted_parameters = (
                distribution_family.fit(
                    sample
                )
            )

            fitted_distribution = (
                distribution_family(
                    *fitted_parameters
                )
            )

        except Exception:
            return None, None

    fitted_parameters = np.asarray(
        fitted_parameters,
        dtype=np.float64,
    )

    if not np.all(
        np.isfinite(
            fitted_parameters
        )
    ):
        return None, None

    scale = fitted_parameters[
        -1
    ]

    if (
        not np.isfinite(
            scale
        )
        or scale <= 0.0
    ):
        return None, None

    return (
        fitted_distribution,
        fitted_parameters,
    )


def summarise_bootstrap(
    bootstrap_array,
):
    """
    Calculate pointwise bootstrap median and 95% interval.
    """
    bootstrap_array = np.asarray(
        bootstrap_array,
        dtype=np.float64,
    )

    if bootstrap_array.ndim != 2:
        raise ValueError(
            "Expected bootstrap array with shape "
            "(n_bootstraps, n_return_periods)."
        )

    with warnings.catch_warnings():
        warnings.simplefilter(
            "ignore",
            RuntimeWarning,
        )

        median = np.nanpercentile(
            bootstrap_array,
            50.0,
            axis=0,
        )

        lower = np.nanpercentile(
            bootstrap_array,
            2.5,
            axis=0,
        )

        upper = np.nanpercentile(
            bootstrap_array,
            97.5,
            axis=0,
        )

        valid_count = np.sum(
            np.isfinite(
                bootstrap_array
            ),
            axis=0,
        )

    return (
        median,
        lower,
        upper,
        valid_count,
    )


def automatic_y_limits(
    curve,
    event_value,
):
    """
    Calculate province-specific y-axis limits.
    """
    values = np.concatenate(
        [
            curve[
                "factual_lower"
            ],
            curve[
                "factual_upper"
            ],
            curve[
                "counterfactual_lower"
            ],
            curve[
                "counterfactual_upper"
            ],
            np.array(
                [
                    event_value
                ],
                dtype=np.float64,
            ),
        ]
    )

    values = values[
        np.isfinite(
            values
        )
    ]

    if values.size == 0:
        return (
            -3.0,
            2.0,
        )

    minimum = float(
        np.nanmin(
            values
        )
    )

    maximum = float(
        np.nanmax(
            values
        )
    )

    data_range = (
        maximum
        - minimum
    )

    if (
        not np.isfinite(
            data_range
        )
        or data_range <= 0.0
    ):
        padding = 0.5

    else:
        padding = (
            0.08
            * data_range
        )

    return (
        minimum - padding,
        maximum + padding,
    )


# ============================================================
# 12. Compute return-period curves
# ============================================================

def compute_return_period_curves(
    province,
):
    """
    Compute factual and counterfactual lower-tail return curves
    with parametric-bootstrap uncertainty.
    """
    print(
        f"\nProcessing {province}"
    )

    factual_data = clean_array(
        fit_store[
            province
        ]["factual"]["data"]
    )

    counterfactual_data = clean_array(
        fit_store[
            province
        ]["counterfactual"]["data"]
    )

    n_factual = factual_data.size
    n_counterfactual = (
        counterfactual_data.size
    )

    original_factual_distribution = (
        fit_store[
            province
        ]["factual"]["dist"]
    )

    original_counterfactual_distribution = (
        fit_store[
            province
        ]["counterfactual"]["dist"]
    )

    factual_original = lower_tail_return_value(
        original_factual_distribution,
        RETURN_PERIOD_GRID,
    )

    counterfactual_original = lower_tail_return_value(
        original_counterfactual_distribution,
        RETURN_PERIOD_GRID,
    )

    factual_bootstrap = np.full(
        (
            N_BOOTSTRAPS,
            N_RETURN_PERIODS,
        ),
        np.nan,
        dtype=np.float64,
    )

    counterfactual_bootstrap = np.full(
        (
            N_BOOTSTRAPS,
            N_RETURN_PERIODS,
        ),
        np.nan,
        dtype=np.float64,
    )

    successful_bootstraps = 0

    for bootstrap_index in range(
        N_BOOTSTRAPS
    ):
        simulated_factual = (
            original_factual_distribution.rvs(
                size=n_factual,
                random_state=random_generator,
            )
        )

        simulated_counterfactual = (
            original_counterfactual_distribution.rvs(
                size=n_counterfactual,
                random_state=random_generator,
            )
        )

        (
            bootstrap_factual_distribution,
            _,
        ) = fit_shared_family(
            simulated_factual
        )

        (
            bootstrap_counterfactual_distribution,
            _,
        ) = fit_shared_family(
            simulated_counterfactual
        )

        if (
            bootstrap_factual_distribution
            is None
            or bootstrap_counterfactual_distribution
            is None
        ):
            continue

        factual_bootstrap[
            bootstrap_index,
            :,
        ] = lower_tail_return_value(
            bootstrap_factual_distribution,
            RETURN_PERIOD_GRID,
        )

        counterfactual_bootstrap[
            bootstrap_index,
            :,
        ] = lower_tail_return_value(
            bootstrap_counterfactual_distribution,
            RETURN_PERIOD_GRID,
        )

        successful_bootstraps += 1

    success_fraction = (
        successful_bootstraps
        / N_BOOTSTRAPS
    )

    if (
        success_fraction
        < MIN_BOOTSTRAP_SUCCESS_FRACTION
    ):
        raise RuntimeError(
            f"{province}: only "
            f"{successful_bootstraps}/{N_BOOTSTRAPS} "
            "bootstrap fits succeeded."
        )

    (
        factual_median,
        factual_lower,
        factual_upper,
        factual_valid_count,
    ) = summarise_bootstrap(
        factual_bootstrap
    )

    (
        counterfactual_median,
        counterfactual_lower,
        counterfactual_upper,
        counterfactual_valid_count,
    ) = summarise_bootstrap(
        counterfactual_bootstrap
    )

    print(
        "  Ensemble sizes:",
        n_factual,
        n_counterfactual,
    )

    print(
        "  Successful bootstraps:",
        f"{successful_bootstraps}/{N_BOOTSTRAPS}",
    )

    return {
        "province":
        province,

        "return_period":
        RETURN_PERIOD_GRID,

        "factual_original":
        factual_original,

        "factual_median":
        factual_median,

        "factual_lower":
        factual_lower,

        "factual_upper":
        factual_upper,

        "factual_valid_count":
        factual_valid_count,

        "counterfactual_original":
        counterfactual_original,

        "counterfactual_median":
        counterfactual_median,

        "counterfactual_lower":
        counterfactual_lower,

        "counterfactual_upper":
        counterfactual_upper,

        "counterfactual_valid_count":
        counterfactual_valid_count,

        "factual_sample_size":
        n_factual,

        "counterfactual_sample_size":
        n_counterfactual,

        "successful_bootstraps":
        successful_bootstraps,

        "success_fraction":
        success_fraction,
    }


CURVE_STORE = {
    province:
    compute_return_period_curves(
        province
    )
    for province in PROVINCES
}


# ============================================================
# 13. Save curve and diagnostic tables
# ============================================================

curve_records = []


for province in PROVINCES:
    curve = CURVE_STORE[
        province
    ]

    for index, return_period in enumerate(
        RETURN_PERIOD_GRID
    ):
        curve_records.append(
            {
                "province":
                province,

                "return_period":
                return_period,

                "factual_original":
                curve[
                    "factual_original"
                ][index],

                "factual_bootstrap_median":
                curve[
                    "factual_median"
                ][index],

                "factual_lower_95":
                curve[
                    "factual_lower"
                ][index],

                "factual_upper_95":
                curve[
                    "factual_upper"
                ][index],

                "factual_valid_bootstraps":
                curve[
                    "factual_valid_count"
                ][index],

                "counterfactual_original":
                curve[
                    "counterfactual_original"
                ][index],

                "counterfactual_bootstrap_median":
                curve[
                    "counterfactual_median"
                ][index],

                "counterfactual_lower_95":
                curve[
                    "counterfactual_lower"
                ][index],

                "counterfactual_upper_95":
                curve[
                    "counterfactual_upper"
                ][index],

                "counterfactual_valid_bootstraps":
                curve[
                    "counterfactual_valid_count"
                ][index],
            }
        )


curve_table = pd.DataFrame(
    curve_records
)

curve_table.to_csv(
    OUTPUT_CURVES,
    index=False,
)


summary_records = []


for province in PROVINCES:
    curve = CURVE_STORE[
        province
    ]

    summary_records.append(
        {
            "province":
            province,

            "distribution":
            SHARED_FAMILY,

            "factual_sample_size":
            curve[
                "factual_sample_size"
            ],

            "counterfactual_sample_size":
            curve[
                "counterfactual_sample_size"
            ],

            "requested_bootstraps":
            N_BOOTSTRAPS,

            "successful_bootstraps":
            curve[
                "successful_bootstraps"
            ],

            "bootstrap_success_fraction":
            curve[
                "success_fraction"
            ],

            "minimum_valid_factual_bootstraps":
            int(
                np.nanmin(
                    curve[
                        "factual_valid_count"
                    ]
                )
            ),

            "minimum_valid_counterfactual_bootstraps":
            int(
                np.nanmin(
                    curve[
                        "counterfactual_valid_count"
                    ]
                )
            ),

            "observed_2022_anomaly":
            OBSERVED_EVENTS[
                province
            ],
        }
    )


summary_table = pd.DataFrame(
    summary_records
)

summary_table.to_csv(
    OUTPUT_SUMMARY,
    index=False,
)


print(
    "\nSaved return-period curves:"
)

print(
    OUTPUT_CURVES
)

print(
    "\nSaved bootstrap summary:"
)

print(
    OUTPUT_SUMMARY
)


# ============================================================
# 14. Plotting helper
# ============================================================

def plot_return_period_panel(
    axis,
    curve,
    province,
):
    """Plot one square province return-period panel."""
    return_period = curve[
        "return_period"
    ]

    # --------------------------------------------------------
    # Factual
    # --------------------------------------------------------

    axis.fill_between(
        return_period,
        curve[
            "factual_lower"
        ],
        curve[
            "factual_upper"
        ],
        color=FACTUAL_COLOR,
        alpha=BAND_ALPHA,
        linewidth=0,
    )

    axis.plot(
        return_period,
        curve[
            "factual_median"
        ],
        color=FACTUAL_COLOR,
        linewidth=LINE_WIDTH,
        label="Factual",
    )

    # --------------------------------------------------------
    # Counterfactual
    # --------------------------------------------------------

    axis.fill_between(
        return_period,
        curve[
            "counterfactual_lower"
        ],
        curve[
            "counterfactual_upper"
        ],
        color=COUNTERFACTUAL_COLOR,
        alpha=BAND_ALPHA,
        linewidth=0,
    )

    axis.plot(
        return_period,
        curve[
            "counterfactual_median"
        ],
        color=COUNTERFACTUAL_COLOR,
        linewidth=LINE_WIDTH,
        label="Counterfactual",
    )

    # --------------------------------------------------------
    # Observed September 2022 event
    # --------------------------------------------------------

    axis.axhline(
        OBSERVED_EVENTS[
            province
        ],
        color=EVENT_COLOR,
        linestyle="--",
        linewidth=EVENT_LINE_WIDTH,
        alpha=0.85,
        label="Observed 2022", # Sep
    )

    # --------------------------------------------------------
    # Axis configuration
    # --------------------------------------------------------

    axis.set_xscale(
        "log"
    )

    axis.set_xlim(
        RETURN_PERIOD_MIN,
        RETURN_PERIOD_MAX,
    )

    # Make physical plotting box square: width / height = 1.
    axis.set_box_aspect(
        1
    )

    axis.set_xticks(
        RETURN_PERIOD_TICKS
    )

    axis.set_xticklabels(
        [
            f"{value:g}"
            for value in RETURN_PERIOD_TICKS
        ],
        fontsize=TICK_SIZE,
    )

    axis.tick_params(
        axis="y",
        labelsize=TICK_SIZE,
    )

    axis.tick_params(
        axis="both",
        which="major",
        length=5.5,
        width=1.1,
    )

    axis.tick_params(
        axis="both",
        which="minor",
        length=3.0,
        width=0.8,
    )

    axis.grid(
        True,
        which="both",
        linestyle="--",
        alpha=GRID_ALPHA,
    )

    # --------------------------------------------------------
    # Panel label at upper-left
    # --------------------------------------------------------

    axis.text(
        0.025,
        0.965,
        PANEL_LABELS[
            province
        ],
        transform=axis.transAxes,
        fontsize=PANEL_LABEL_SIZE,
        fontweight="bold",
        horizontalalignment="left",
        verticalalignment="top",
        zorder=20,
    )

    # --------------------------------------------------------
    # Province name inside upper-right
    # --------------------------------------------------------

    axis.text(
        0.965,
        0.965,
        PROVINCE_SHORT[
            province
        ],
        transform=axis.transAxes,
        fontsize=PROVINCE_NAME_SIZE,
        fontweight="bold",
        horizontalalignment="right",
        verticalalignment="top",
        bbox={
            "facecolor": "white",
            "edgecolor": "none",
            "alpha": PROVINCE_BOX_ALPHA,
            "pad": 2.5,
        },
        zorder=20,
    )

    # --------------------------------------------------------
    # Y-axis limits
    # --------------------------------------------------------

    if USE_CUSTOM_Y_LIMITS:
        axis.set_ylim(
            CUSTOM_Y_LIMITS[
                province
            ]
        )

    else:
        y_minimum, y_maximum = (
            automatic_y_limits(
                curve,
                OBSERVED_EVENTS[
                    province
                ],
            )
        )

        axis.set_ylim(
            y_minimum,
            y_maximum,
        )


# ============================================================
# 15. Five-panel square figure
# ============================================================

figure = plt.figure(
    figsize=FIGURE_SIZE
)


# A 2 x 6 GridSpec allows:
#
# Top:
#     [0:2] [2:4] [4:6]
#
# Bottom:
#        [1:3] [3:5]
#
# Smaller wspace and hspace keep the square panels close.
grid = figure.add_gridspec(
    nrows=2,
    ncols=6,
    height_ratios=[
        1.0,
        1.0,
    ],
    wspace=0.18,
    hspace=0.18,
)


axes = {
    "Sichuan Province":
    figure.add_subplot(
        grid[
            0,
            0:2,
        ]
    ),

    "Chongqing Municipality":
    figure.add_subplot(
        grid[
            0,
            2:4,
        ]
    ),

    "Hubei Province":
    figure.add_subplot(
        grid[
            0,
            4:6,
        ]
    ),

    "Hunan Province":
    figure.add_subplot(
        grid[
            1,
            1:3,
        ]
    ),

    "Guizhou Province":
    figure.add_subplot(
        grid[
            1,
            3:5,
        ]
    ),
}


for province in PROVINCES:
    plot_return_period_panel(
        axis=axes[
            province
        ],
        curve=CURVE_STORE[
            province
        ],
        province=province,
    )


# ============================================================
# 16. Shared x-axis label
# ============================================================

figure.supxlabel(
    "Return period (years)",
    fontsize=X_LABEL_SIZE,
    fontweight="normal",
    y=0.055,
)


# No shared y-axis label.


# ============================================================
# 17. Shared legend
# ============================================================

legend_handles, legend_labels = (
    axes[
        "Sichuan Province"
    ].get_legend_handles_labels()
)


unique_legend = {}

for handle, label in zip(
    legend_handles,
    legend_labels,
):
    if label not in unique_legend:
        unique_legend[
            label
        ] = handle


figure.legend(
    unique_legend.values(),
    unique_legend.keys(),
    loc="lower center",
    bbox_to_anchor=(
        0.5,
        -0.005,
    ),
    ncol=3,
    frameon=False,
    fontsize=LEGEND_SIZE,
    handlelength=3.0,
    columnspacing=2.2,
)


# ============================================================
# 18. Final layout and save
# ============================================================

figure.subplots_adjust(
    left=0.055,
    right=0.985,
    top=0.975,
    bottom=0.145,
)


figure.savefig(
    OUTPUT_FIGURE,
    dpi=FIGURE_DPI,
    bbox_inches="tight",
)


print(
    "\nSaved five-panel square return-period figure:"
)

print(
    OUTPUT_FIGURE
)


plt.show()

plt.close(
    figure
)


print(
    "\nStep 4 completed successfully."
)

print(
    "Shared family:",
    SHARED_FAMILY,
)

print(
    "Return-level definition:"
)

print(
    "x(T) = F^{-1}(1/T), lower hydropower tail"
)
```
