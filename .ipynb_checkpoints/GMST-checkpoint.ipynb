```r
# Upload the annual csv files to file directory, and then read it (the same level as var)
# Reads a CSV file into R and stores it as a data frame called df

# df = read.csv("/home/users/weijia_6/Yangtze_code/method_1_data/population_weighted_cdd_summer_sum_1961_2022.csv", row.names = "Province") # original

# # CN05 after regridded to HadGEM
df = read.csv("/home/users/weijia_6/Yangtze_code/popweighted_CDD_summer_sum_1961_2022_regridded_to_hadgem.csv", row.names = 1)
```

```r
# This selects the entire row of the data frame df corresponding to the row name "Sichuan".
df["Sichuan",]
```

```r
# Builds a two-column data frame (year + CDD values for the selected province Sichuan)
cdd = data.frame("year" = 1961:2022,
           "CDD" = unlist(df["Sichuan",]))

# Computes the logarithm of CDD and stores it as a new column,
cdd$logcdd = log(cdd$CDD)
```

```r
# Plots the empirical distribution of log(CDD) from your data
plot(density(log(cdd$CDD)))

# Overlays a theoretical normal distribution fitted to the same data
lines(x<-seq(2, 6, 0.01), dnorm(x, mean(cdd$logcdd), sd(cdd$logcdd)), col = "blue")
```

```r
# Read gmst file and stores it as a data frame called gmst.
gmst = read.table("/home/users/weijia_6/Yangtze_code/method_1_data/gmst.txt", col.names = c("year", "GMST"))
```

```r
# Combine two data frames (gmst and cdd) into a single data frame by matching rows with the same year.
cdd = merge(gmst, cdd)
```

```r
# Create a scatterplot matrix (also called a pairs plot) for all variables in the data frame cdd.
# Check relationship between different groups
plot(cdd, lower.panel = panel.smooth)
```

```r
# Check how climate change (GMST) shift the probability distribution of the selected variable (CDD)
# "norm": Standard normal (Gaussian) distribution
# "norm_logt": Normal distribution with log-transformed tails - Better captures extreme events
# Set GMST as the climate covariate, set event year, run three fitting models

# Distribution: normal, Variable: log(CDD), Assumes log(CDD) is Gaussian, Common baseline model
mdl = fit_ns(dist = "norm", type = "shift", data = cdd, varnm = "logcdd", covnm = "GMST", ev_year = 2022)

# Distribution: normal, Variable: raw CDD, Stronger assumption (CDD is symmetric), Often less realistic for extremes
mdl_cdd = fit_ns(dist = "norm", type = "shift", data = cdd, varnm = "CDD", covnm = "GMST", ev_year = 2022)

# Distribution: norm_logt, Variable: raw CDD, Best suited for positive, skewed, extreme-sensitive data, Usually the preferred model
mdl_cdd_1 = fit_ns(dist = "norm_logt", type = "shift", data = cdd, varnm = "CDD", covnm = "GMST", ev_year = 2022)
```

```r
# Creates a PNG image file, Sets up a 1 × 2 plotting layout, Plots two trend figures side by side, Saves everything to disk

png("/home/users/weijia_6/Yangtze_code/method_1_result/fit_model.png", height = 360, width = 360*2);{
prep_window(c(1,2))
plot_trend(mdl, ylim = c(2.5, 6.5), main = "Trend in Log CDD ") # trend log CDD, shown in log CDD
# plot_trend(mdl_cdd, main = "Trend in CDD")
plot_trend(mdl_cdd_1, main = "Trend in CDD") # trend log CDD, shown in CDD
};dev.off()
```

```r
# plots how the fitted model parameter changes with the climate covariate GMST.
plot_covtrend(mdl, "GMST")
```

```r
plot(cdd$year, cdd$GMST)
```

```r
# Extract the GMST value for the year 2022 from the data frame cdd and stores it as a one-row data frame called cov_f,
# representing the factual (real-world) climate condition in the event year.
cov_f = cdd[cdd$year == 2022, "GMST", drop = F] # factual cdd
```

```r
# Create a counterfactual climate covariate by removing anthropogenic warming from the observed (factual) climate state.
cov_cf = cov_f - 1.2 # for 2022 counterfactual
```

```r
# Generate return-level (or return-period) plot comparing the factual and counterfactual climate worlds.
prep_window(h = 6, w = 6)

# nsamp controls bootstrap sampling, Number of resamples used to estimate uncertainty,
# Larger values → smoother, more stable uncertainty bands, Smaller values → faster computation. 50 for debugging, 500+ for final results
plot_returnlevels(mdl, cov_f, cov_cf, nsamp = 50)
```

```r
# Use another fitting model
prep_window(h = 6, w = 6)
plot_returnlevels(mdl_cdd_1, cov_f, cov_cf, nsamp = 50)
```

```r
# Generate result table
boot_ci(mdl, cov_f, cov_cf, nsamp = 50) # default 1000 nsamp
```

```r
# suggested! use the log based fitting, but leave the variable as raw CDD change
boot_ci(mdl_cdd_1, cov_f, cov_cf, nsamp = 50) # default 1000 nsamp
# PR: probability ratio change: 59.60 times more frequent
# dI_rel: realtive intensity change: 72.04 times more intense
```

```r
# Create a path for results
out_dir <- "/home/users/weijia_6/Yangtze_code/method_1_result"
dir.create(out_dir, showWarnings = FALSE)
```

```r
# Generate results
gmst = read.table("/home/users/weijia_6/Yangtze_code/method_1_data/gmst.txt", col.names = c("year", "GMST"))
cov_f = cdd[cdd$year == 2022, "GMST", drop = F]
cov_cf = cov_f - 1.2 # for 2022 counterfactual

for (rnm in rownames(df)){
  cdd = data.frame("year" = 1961:2022,
           "CDD" = unlist(df[rnm,]))
           if (any(is.na(cdd$CDD))) next
           if (any(cdd$CDD == 0)) next
cdd$logcdd = log(cdd$CDD)
cdd = merge(gmst, cdd)
mdl_cdd_1 = fit_ns(dist = "norm_logt", type = "shift", data = cdd, varnm = "CDD", covnm = "GMST", ev_year = 2022)

# png(paste0("/var/fit_model_", rnm, ".png"), height = 360, width = 360);{
png(paste0(out_dir, "/fit_model_", rnm, ".png"), height = 360, width = 360);{
prep_window(c(1,1))
plot_trend(mdl_cdd_1, main = (paste0("Trend in CDD ", rnm))) # trend log CDD, shown in CDD
};dev.off()

png(paste0(out_dir, "/fit_model_returnperiod_", rnm, ".png"), height = 360, width = 360);{
prep_window(h = 6, w = 6)
plot_returnlevels(mdl_cdd_1, cov_f, cov_cf, nsamp = 1000, main = (paste0(rnm))) # can be set to 500, 1000 later, might be slow due to bootstrap
};dev.off()

res = boot_ci(mdl_cdd_1, cov_f, cov_cf, nsamp = 1000) # default 1000 nsamp
write.csv(res, paste0(out_dir, "/fit_model_res_", rnm, ".csv"))
}
```

```r
# Summarise all province result

library(dplyr)
library(readr)
library(tidyr)
library(stringr)

# out_dir <- "/path/to/output"   # <-- change this

# List all CSV files
files <- list.files(out_dir, pattern = "fit_model_res_.*\\.csv", full.names = TRUE)

read_and_clean <- function(f) {
  # Read CSV with row names
  df <- read.csv(f, row.names = 1, check.names = FALSE)

  # Convert rownames to a column
  df <- df %>%
    mutate(metric = rownames(.)) %>%
    relocate(metric)

  # Extract province name WITHOUT ".csv"
  prov <- basename(f) %>%
    str_replace("fit_model_res_", "") %>%
    str_replace("\\.csv$", "")   # <-- removes .csv

  df$Province <- prov

  # Pivot wider so each metric becomes a column
  df_wide <- df %>%
    pivot_wider(
      names_from = metric,
      values_from = c(est, `2.5%`, `97.5%`),
      names_glue = "{metric}_{.value}"
    )

  return(df_wide)
}

# Apply to all files
combined <- bind_rows(lapply(files, read_and_clean))

# Save
write_csv(combined, file.path("/home/users/weijia_6/Yangtze_code/method_1_result/combined_attribution_results_clean.csv"))

combined
```
