```python
# ============================================================
# Raincloud-style 2x2 panel figure:
# Energy imbalance attributed to anthropogenic climate change
#
# Panels:
# (a) Demand anomaly: CF vs F
# (b) Hydropower generation anomaly: CF vs F
# (c) Energy imbalance anomaly: demand - generation, CF vs F
# (d) Climate-change-driven energy imbalance: F - CF from panel (c)
# ============================================================

import warnings
warnings.filterwarnings("ignore")

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from matplotlib.patches import Patch
from matplotlib.lines import Line2D


# ============================================================
# 1. User settings
# ============================================================

selected_provinces = [
    "Guizhou Province",
    "Sichuan Province",
    "Hunan Province",
    "Hubei Province",
    "Chongqing Municipality",
]

province_short = {
    "Guizhou Province": "Guizhou",
    "Sichuan Province": "Sichuan",
    "Hunan Province": "Hunan",
    "Hubei Province": "Hubei",
    "Chongqing Municipality": "Chongqing",
}

hydro_mapping = {
    "Guizhou Province": "贵州",
    "Sichuan Province": "四川",
    "Hunan Province": "湖南",
    "Hubei Province": "湖北",
    "Chongqing Municipality": "重庆",
}

# -----------------------------------------------------------------
# Updated data paths
# -----------------------------------------------------------------

hist_cdd_file = "/home/users/weijia_6/Yangtze_code/popweighted_CDD_summer_sum_1961_2022_regridded_to_hadgem.csv"

cdd_f_file = "/home/users/weijia_6/Yangtze_code/CDD26_factual_popweighted_provinces_correct.csv"
cdd_cf_file = "/home/users/weijia_6/Yangtze_code/CDD26_counterfactual_popweighted_provinces_correct.csv"

hydro_f_file = "/home/users/weijia_6/Yangtze_code/gen_result/step5_gamma_spi6/hydro_std_anom_sep_factual_525_by_province.csv"
hydro_cf_file = "/home/users/weijia_6/Yangtze_code/gen_result/step5_gamma_spi6/hydro_std_anom_sep_counterfactual_525_by_province.csv"

# Output files
out_dir = "/home/users/weijia_6/Yangtze_code/imbalance_result"
os.makedirs(out_dir, exist_ok=True)

out_fig_png = f"{out_dir}/Panel_Figure_energy_imbalance_raincloud_spaced.png"
out_fig_pdf = f"{out_dir}/Panel_Figure_energy_imbalance_raincloud_spaced.pdf"

out_summary_demand = f"{out_dir}/summary_demand_std_anomaly_2022.csv"
out_summary_generation = f"{out_dir}/summary_hydro_generation_std_anomaly_2022.csv"
out_summary_imbalance = f"{out_dir}/summary_energy_imbalance_std_2022.csv"
out_summary_delta = f"{out_dir}/summary_climate_change_energy_imbalance_2022.csv"


# ============================================================
# 2. Plot style
# ============================================================

plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "font.size": 15,
    "axes.titlesize": 18,
    "axes.labelsize": 16,
    "xtick.labelsize": 15,
    "ytick.labelsize": 15,
    "legend.fontsize": 14,
    "figure.titlesize": 20,
    "axes.linewidth": 1.0,
    "savefig.dpi": 450,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})

# Muted, print-friendly colours
CF_COLOR = "#4C78A8"       # blue for counterfactual
F_COLOR = "#F58518"         # orange for factual
DELTA_COLOR = "#2CA25F"     # green for panel (d)
POINT_COLOR = "0.20"

# Row spacing control
ROW_GAP = 1.5


# ============================================================
# 3. Helper functions
# ============================================================

def load_province_ensemble_csv(path):
    """
    Load a province x ensemble CSV with province names in the first column.
    Used for the CDD files.
    """
    df = pd.read_csv(path, index_col=0)
    df.index = df.index.astype(str).str.strip()
    df.columns = [str(c).strip() for c in df.columns]

    missing = [p for p in selected_provinces if p not in df.index]
    if missing:
        raise KeyError(f"Missing provinces in {path}: {missing}")

    df = df.loc[selected_provinces].copy()
    df = df.apply(pd.to_numeric, errors="coerce")
    return df


def load_hydro_long_csv(path, value_col="hydro_std_anom_sep", scenario=None):
    """
    Load the hydro CSV, which is in long format:
      scenario, ensemble, r, p, province, hydro_std_anom_sep, ...

    Returns a province x ensemble DataFrame.
    """
    df = pd.read_csv(path)
    df.columns = [str(c).strip() for c in df.columns]

    required = {"province", "ensemble", value_col}
    missing_cols = required - set(df.columns)
    if missing_cols:
        raise KeyError(f"{path} is missing columns: {sorted(missing_cols)}")

    df["province"] = df["province"].astype(str).str.strip()

    if scenario is not None and "scenario" in df.columns:
        df["scenario"] = df["scenario"].astype(str).str.strip().str.lower()
        scenario = str(scenario).strip().lower()
        df = df.loc[df["scenario"] == scenario].copy()

    df["ensemble"] = pd.to_numeric(df["ensemble"], errors="coerce")
    df[value_col] = pd.to_numeric(df[value_col], errors="coerce")

    df = df.loc[df["province"].isin(selected_provinces)].copy()

    # One row per province/ensemble is expected.
    wide = df.pivot_table(
        index="province",
        columns="ensemble",
        values=value_col,
        aggfunc="first",
    )

    # Preserve province order and make sure all five are present.
    missing = [p for p in selected_provinces if p not in wide.index]
    if missing:
        raise KeyError(f"Missing provinces in {path}: {missing}")

    wide = wide.loc[selected_provinces].copy()

    # Sort ensemble columns numerically and relabel them 0..n-1
    wide = wide.reindex(sorted(wide.columns), axis=1)
    wide.columns = range(wide.shape[1])

    return wide.apply(pd.to_numeric, errors="coerce")


def force_matching_columns(*dfs):
    """
    Force all DataFrames to have ensemble columns 0, 1, ..., n-1.
    """
    ncols = [df.shape[1] for df in dfs]
    if len(set(ncols)) != 1:
        raise ValueError(
            f"Input DataFrames have different ensemble sizes: {ncols}. "
            "Check whether the ensembles are comparable and paired."
        )

    n = ncols[0]
    out = []
    for df in dfs:
        tmp = df.copy()
        tmp.columns = range(n)
        out.append(tmp)
    return out


def summary_stats(df, prefix):
    """
    Summary statistics for a province x ensemble DataFrame.
    """
    return pd.DataFrame({
        f"{prefix}_mean": df.mean(axis=1),
        f"{prefix}_median": df.median(axis=1),
        f"{prefix}_2.5%": df.quantile(0.025, axis=1),
        f"{prefix}_25%": df.quantile(0.25, axis=1),
        f"{prefix}_75%": df.quantile(0.75, axis=1),
        f"{prefix}_97.5%": df.quantile(0.975, axis=1),
        f"{prefix}_P_gt_0": (df > 0).mean(axis=1),
    })


def paired_summary_stats(df_cf, df_f, name):
    """
    Summary table for CF and F distributions.
    """
    return pd.concat(
        [
            summary_stats(df_cf, f"{name}_CF"),
            summary_stats(df_f, f"{name}_F"),
        ],
        axis=1,
    )


def symmetric_xlim(*dfs, pad=0.12):
    """
    Symmetric x-limit around zero based on 0.5-99.5% range.
    """
    vals = np.concatenate([df.values.ravel() for df in dfs])
    vals = vals[np.isfinite(vals)]

    lo = np.nanquantile(vals, 0.005)
    hi = np.nanquantile(vals, 0.995)
    vmax = max(abs(lo), abs(hi))
    vmax = vmax * (1 + pad)

    return -vmax, vmax


def data_xlim(*dfs, pad=0.12):
    """
    Non-symmetric x-limit based on 0.5-99.5% range.
    Useful for panel d if values are mostly positive.
    """
    vals = np.concatenate([df.values.ravel() for df in dfs])
    vals = vals[np.isfinite(vals)]

    lo = np.nanquantile(vals, 0.005)
    hi = np.nanquantile(vals, 0.995)
    span = hi - lo

    return lo - pad * span, hi + pad * span


def add_panel_label(ax, label):
    ax.text(
        -0.10,
        1.08,
        label,
        transform=ax.transAxes,
        fontsize=17,
        fontweight="bold",
        va="top",
        ha="left"
    )


def half_violin(
    ax,
    values,
    position,
    color,
    width=0.62,
    side="right",
    alpha=0.35
):
    """
    Draw a half violin at a given y-position.
    """
    values = np.asarray(values)
    values = values[np.isfinite(values)]

    vp = ax.violinplot(
        [values],
        positions=[position],
        vert=False,
        widths=width,
        showmeans=False,
        showmedians=False,
        showextrema=False
    )

    body = vp["bodies"][0]
    verts = body.get_paths()[0].vertices

    if side == "right":
        verts[:, 1] = np.maximum(verts[:, 1], position)
    elif side == "left":
        verts[:, 1] = np.minimum(verts[:, 1], position)
    else:
        raise ValueError("side must be 'right' or 'left'")

    body.set_facecolor(color)
    body.set_edgecolor("none")
    body.set_alpha(alpha)
    body.set_zorder(1)

    return body


def draw_raincloud_paired(
    ax,
    df_cf,
    df_f,
    provinces,
    title,
    xlabel,
    panel_label=None,
    xlim=None,
    show_yticklabels=True
):
    """
    Draw paired CF/F raincloud plot.
    """
    rng = np.random.default_rng(42)

    y_centres = np.arange(len(provinces)) * ROW_GAP
    offset = 0.24
    box_width = 0.13

    ax.axvline(
        0,
        color="0.70",
        linewidth=1.2,
        linestyle="-",
        zorder=0
    )

    ax.grid(
        axis="x",
        color="0.90",
        linewidth=0.9,
        linestyle="-",
        zorder=0
    )

    for i, prov in enumerate(provinces):
        cf_vals = df_cf.loc[prov].dropna().values
        f_vals = df_f.loc[prov].dropna().values

        y_cf = y_centres[i] - offset
        y_f = y_centres[i] + offset

        half_violin(
            ax,
            cf_vals,
            y_cf,
            CF_COLOR,
            width=0.60,
            side="left",
            alpha=0.34
        )

        half_violin(
            ax,
            f_vals,
            y_f,
            F_COLOR,
            width=0.60,
            side="right",
            alpha=0.34
        )

        max_points = 220

        cf_plot = rng.choice(cf_vals, size=min(len(cf_vals), max_points), replace=False) if len(cf_vals) > max_points else cf_vals
        f_plot = rng.choice(f_vals, size=min(len(f_vals), max_points), replace=False) if len(f_vals) > max_points else f_vals

        jitter_cf = rng.normal(loc=0, scale=0.028, size=len(cf_plot))
        jitter_f = rng.normal(loc=0, scale=0.028, size=len(f_plot))

        ax.scatter(
            cf_plot,
            np.full(len(cf_plot), y_cf) + jitter_cf,
            s=10,
            color=CF_COLOR,
            alpha=0.24,
            linewidths=0,
            zorder=2
        )

        ax.scatter(
            f_plot,
            np.full(len(f_plot), y_f) + jitter_f,
            s=10,
            color=F_COLOR,
            alpha=0.24,
            linewidths=0,
            zorder=2
        )

        bp = ax.boxplot(
            [cf_vals, f_vals],
            positions=[y_cf, y_f],
            vert=False,
            widths=box_width,
            patch_artist=True,
            showfliers=False,
            whis=(2.5, 97.5),
            medianprops=dict(color="black", linewidth=1.3),
            whiskerprops=dict(color="0.35", linewidth=1.0),
            capprops=dict(color="0.35", linewidth=1.0),
            boxprops=dict(linewidth=1.0),
            zorder=4
        )

        for patch, color in zip(bp["boxes"], [CF_COLOR, F_COLOR]):
            patch.set_facecolor(color)
            patch.set_alpha(0.88)
            patch.set_edgecolor("0.25")

        ax.scatter(
            np.nanmean(cf_vals),
            y_cf,
            marker="D",
            s=34,
            color="black",
            edgecolor="white",
            linewidth=0.5,
            zorder=5
        )

        ax.scatter(
            np.nanmean(f_vals),
            y_f,
            marker="D",
            s=34,
            color="black",
            edgecolor="white",
            linewidth=0.5,
            zorder=5
        )

    ax.set_yticks(y_centres)

    if show_yticklabels:
        ax.set_yticklabels([province_short[p] for p in provinces])
    else:
        ax.set_yticklabels([])

    ax.invert_yaxis()

    if xlim is not None:
        ax.set_xlim(xlim)

    ax.set_title(title, loc="left", pad=10)
    ax.set_xlabel(xlabel, labelpad=8)

    if panel_label is not None:
        add_panel_label(ax, panel_label)

    ax.tick_params(axis="both", labelsize=12.5)

    for spine in ["top", "right"]:
        ax.spines[spine].set_visible(False)


def draw_raincloud_single(
    ax,
    df,
    provinces,
    title,
    xlabel,
    panel_label=None,
    xlim=None,
    color=DELTA_COLOR,
    show_yticklabels=True
):
    """
    Draw single raincloud plot for F - CF energy imbalance.
    """
    rng = np.random.default_rng(43)

    y_centres = np.arange(len(provinces)) * ROW_GAP
    box_width = 0.15

    ax.axvline(
        0,
        color="0.70",
        linewidth=1.2,
        linestyle="-",
        zorder=0
    )

    ax.grid(
        axis="x",
        color="0.90",
        linewidth=0.9,
        linestyle="-",
        zorder=0
    )

    for i, prov in enumerate(provinces):
        vals = df.loc[prov].dropna().values
        y = y_centres[i]

        half_violin(
            ax,
            vals,
            y,
            color,
            width=0.76,
            side="right",
            alpha=0.38
        )

        max_points = 260
        vals_plot = rng.choice(vals, size=min(len(vals), max_points), replace=False) if len(vals) > max_points else vals

        jitter = rng.normal(loc=-0.13, scale=0.040, size=len(vals_plot))

        ax.scatter(
            vals_plot,
            np.full(len(vals_plot), y) + jitter,
            s=10,
            color=color,
            alpha=0.28,
            linewidths=0,
            zorder=2
        )

        bp = ax.boxplot(
            [vals],
            positions=[y],
            vert=False,
            widths=box_width,
            patch_artist=True,
            showfliers=False,
            whis=(2.5, 97.5),
            medianprops=dict(color="black", linewidth=1.3),
            whiskerprops=dict(color="0.35", linewidth=1.0),
            capprops=dict(color="0.35", linewidth=1.0),
            boxprops=dict(linewidth=1.0),
            zorder=4
        )

        bp["boxes"][0].set_facecolor(color)
        bp["boxes"][0].set_alpha(0.88)
        bp["boxes"][0].set_edgecolor("0.25")

        ax.scatter(
            np.nanmean(vals),
            y,
            marker="D",
            s=36,
            color="black",
            edgecolor="white",
            linewidth=0.5,
            zorder=5
        )

    ax.set_yticks(y_centres)

    if show_yticklabels:
        ax.set_yticklabels([province_short[p] for p in provinces])
    else:
        ax.set_yticklabels([])

    ax.invert_yaxis()

    if xlim is not None:
        ax.set_xlim(xlim)

    ax.set_title(title, loc="left", pad=10)
    ax.set_xlabel(xlabel, labelpad=8)

    if panel_label is not None:
        add_panel_label(ax, panel_label)

    ax.tick_params(axis="both", labelsize=16)

    for spine in ["top", "right"]:
        ax.spines[spine].set_visible(False)


def prepare_cdd_reference(hist_file, provinces, ref_year_start=1970, ref_year_end=2010):
    """
    Load the long-term CDD file, select the target provinces,
    and compute the 1970-2010 climatological mean and std.
    """
    df = pd.read_csv(hist_file, index_col=0)
    df.index = df.index.astype(str).str.strip()
    df = df.loc[provinces].apply(pd.to_numeric, errors="coerce")

    year_cols = []
    for y in range(ref_year_start, ref_year_end + 1):
        candidates = [str(y), y]
        found = None
        for c in candidates:
            if c in df.columns:
                found = c
                break
        if found is None:
            raise KeyError(f"Year {y} not found in {hist_file}")
        year_cols.append(found)

    ref = df[year_cols].copy()
    ref.columns = [int(c) for c in year_cols]

    clim_mu = ref.mean(axis=1)
    clim_sd = ref.std(axis=1, ddof=1)

    if (clim_sd <= 0).any():
        bad = clim_sd[clim_sd <= 0].index.tolist()
        raise ValueError(f"Zero/negative CDD std for provinces: {bad}")

    return df, clim_mu, clim_sd


# ============================================================
# 4. Calculate demand anomaly with 1970-2010 reference
# ============================================================

df_hist_all, clim_mu, clim_sd = prepare_cdd_reference(
    hist_cdd_file,
    selected_provinces,
    ref_year_start=1970,
    ref_year_end=2010,
)

df_cdd_f = load_province_ensemble_csv(cdd_f_file)
df_cdd_cf = load_province_ensemble_csv(cdd_cf_file)

# Standardize factual/counterfactual against the 1970-2010 baseline
std_f_demand = df_cdd_f.sub(clim_mu, axis=0).div(clim_sd, axis=0)
std_cf_demand = df_cdd_cf.sub(clim_mu, axis=0).div(clim_sd, axis=0)

std_cf_demand, std_f_demand = force_matching_columns(
    std_cf_demand,
    std_f_demand,
)

summary_demand = paired_summary_stats(
    std_cf_demand,
    std_f_demand,
    "demand",
)
summary_demand.to_csv(out_summary_demand)


# ============================================================
# 5. Calculate hydropower generation anomaly
# ============================================================

std_f_gen = load_hydro_long_csv(
    hydro_f_file,
    value_col="hydro_std_anom_sep",
    scenario="factual",
)

std_cf_gen = load_hydro_long_csv(
    hydro_cf_file,
    value_col="hydro_std_anom_sep",
    scenario="counterfactual",
)

std_cf_gen, std_f_gen = force_matching_columns(
    std_cf_gen,
    std_f_gen,
)

summary_generation = paired_summary_stats(
    std_cf_gen,
    std_f_gen,
    "generation",
)
summary_generation.to_csv(out_summary_generation)


# ============================================================
# 6. Calculate energy imbalance and climate-change contribution
# ============================================================

n_demand = std_f_demand.shape[1]
n_gen = std_f_gen.shape[1]

if n_demand != n_gen:
    raise ValueError(
        f"Demand and hydropower ensemble sizes differ: "
        f"demand={n_demand}, hydropower={n_gen}. "
        "If they are not paired, use bootstrap or all-pairs sampling instead."
    )

std_cf_demand, std_f_demand, std_cf_gen, std_f_gen = force_matching_columns(
    std_cf_demand,
    std_f_demand,
    std_cf_gen,
    std_f_gen,
)

# Energy imbalance = demand - generation
imb_cf = std_cf_demand - std_cf_gen
imb_f = std_f_demand - std_f_gen

summary_imbalance = paired_summary_stats(
    imb_cf,
    imb_f,
    "imbalance",
)
summary_imbalance.to_csv(out_summary_imbalance)

# Climate-change-driven imbalance = factual - counterfactual
delta_imb = imb_f - imb_cf

summary_delta = summary_stats(delta_imb, "delta_imbalance")
summary_delta.to_csv(out_summary_delta)


# ============================================================
# 7. Province order
# ============================================================

province_order = selected_provinces


# ============================================================
# 8. Axis limits
# ============================================================

xlim_demand = symmetric_xlim(std_cf_demand, std_f_demand, pad=0.15)
xlim_gen = symmetric_xlim(std_cf_gen, std_f_gen, pad=0.20)
xlim_imb = symmetric_xlim(imb_cf, imb_f, pad=0.15)

xlim_delta = data_xlim(delta_imb, pad=0.15)
xlim_delta = (
    min(xlim_delta[0], -0.5),
    max(xlim_delta[1], 0.5)
)


# ============================================================
# 9. Plot 2x2 raincloud panel figure
# ============================================================

fig, axes = plt.subplots(
    2,
    2,
    figsize=(15.0, 11),
    constrained_layout=False
)

ax_a, ax_b, ax_c, ax_d = axes.ravel()

draw_raincloud_paired(
    ax=ax_a,
    df_cf=std_cf_demand,
    df_f=std_f_demand,
    provinces=province_order,
    title="Demand anomaly",
    xlabel="CDD26 standardised anomaly",
    panel_label="a",
    xlim=xlim_demand,
    show_yticklabels=True
)

draw_raincloud_paired(
    ax=ax_b,
    df_cf=std_cf_gen,
    df_f=std_f_gen,
    provinces=province_order,
    title="Hydropower generation anomaly",
    xlabel="Generation standardised anomaly",
    panel_label="b",
    xlim=xlim_gen,
    show_yticklabels=False
)

draw_raincloud_paired(
    ax=ax_c,
    df_cf=imb_cf,
    df_f=imb_f,
    provinces=province_order,
    title="Energy imbalance anomaly",
    xlabel="Demand - generation, standardised units",
    panel_label="c",
    xlim=xlim_imb,
    show_yticklabels=True
)

draw_raincloud_single(
    ax=ax_d,
    df=delta_imb,
    provinces=province_order,
    title="Climate-change-driven energy imbalance",
    xlabel="Factual - counterfactual imbalance",
    panel_label="d",
    xlim=xlim_delta,
    color=DELTA_COLOR,
    show_yticklabels=False
)

legend_handles = [
    Patch(
        facecolor=CF_COLOR,
        edgecolor="0.25",
        alpha=0.88,
        label="Counterfactual"
    ),
    Patch(
        facecolor=F_COLOR,
        edgecolor="0.25",
        alpha=0.88,
        label="Factual"
    ),
    Patch(
        facecolor=DELTA_COLOR,
        edgecolor="0.25",
        alpha=0.88,
        label="Factual - counterfactual"
    ),
    Line2D(
        [0], [0],
        marker="D",
        color="none",
        markerfacecolor="black",
        markeredgecolor="white",
        markeredgewidth=0.5,
        markersize=7,
        label="Mean"
    ),
    Line2D(
        [0], [0],
        color="black",
        linewidth=1.3,
        label="Median"
    )
]

fig.legend(
    handles=legend_handles,
    loc="lower center",
    ncol=5,
    frameon=False,
    bbox_to_anchor=(0.5, 0.015),
    columnspacing=1.5,
    handlelength=1.8
)

plt.subplots_adjust(
    left=0.08,
    right=0.985,
    top=0.91,
    bottom=0.14,
    wspace=0.18,
    hspace=0.46
)

fig.savefig(out_fig_png, dpi=450, bbox_inches="tight")
fig.savefig(out_fig_pdf, bbox_inches="tight")

plt.show()


# ============================================================
# 10. Print outputs
# ============================================================

print("\nSaved figure:")
print(f"  {out_fig_png}")
print(f"  {out_fig_pdf}")

print("\nSaved summary tables:")
print(f"  {out_summary_demand}")
print(f"  {out_summary_generation}")
print(f"  {out_summary_imbalance}")
print(f"  {out_summary_delta}")

print("\nClimate-change-driven energy imbalance summary:")
print(summary_delta.round(3))
```

p.s. delta_imbalance_P_gt_0 is the fraction of ensemble members in which delta_imbalance > 0 for each province.


## Interpretation

These results suggest that anthropogenic climate change generally increased the energy imbalance in all five provinces. Because delta_imbalance = imbalance_factual - imbalance_counterfactual, a positive value means the factual climate produced a larger demand - generation gap than the counterfactual climate.

The strongest signal is in Guizhou Province: the mean is 5.676, the median is 5.492, and delta_imbalance_P_gt_0 = 0.985, so almost all ensemble members are positive. That makes Guizhou the most clearly impacted province in this output.

The other four provinces also show positive median and mean values, but the effect is smaller:
Sichuan is the weakest positive case, while Hunan has the next-largest increase, and Hubei and Chongqing are in the middle.

A few details matter:

The mean and median are both positive for every province, so the central estimate is consistently above zero.

For Sichuan, Hunan, Hubei, and Chongqing, the 2.5% bootstrap quantile is negative, which means there is still some uncertainty and a small fraction of the distribution allows a slight decrease.

But the P_gt_0 values are all high (0.882 to 0.985), so most ensemble members still point to a positive climate-change contribution.

Short summary: climate change appears to have worsened the imbalance between demand and hydropower generation, with the clearest increase in Guizhou, and a more moderate but still mostly positive increase in the other four provinces.

```python
# Violin plot


# ============================================================
# Paired violin plot:
# Energy imbalance attributed to anthropogenic climate change
# ============================================================

import warnings
warnings.filterwarnings("ignore")

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from matplotlib.lines import Line2D


# ============================================================
# 1. User settings
# ============================================================

selected_provinces = [
    "Guizhou Province",
    "Sichuan Province",
    "Hunan Province",
    "Hubei Province",
    "Chongqing Municipality",
]

province_short = {
    "Guizhou Province": "Guizhou",
    "Sichuan Province": "Sichuan",
    "Hunan Province": "Hunan",
    "Hubei Province": "Hubei",
    "Chongqing Municipality": "Chongqing",
}

# CDD files
hist_cdd_file = "/home/users/weijia_6/Yangtze_code/popweighted_CDD_summer_sum_1961_2022_regridded_to_hadgem.csv"
cdd_f_file = "/home/users/weijia_6/Yangtze_code/CDD26_factual_popweighted_provinces_correct.csv"
cdd_cf_file = "/home/users/weijia_6/Yangtze_code/CDD26_counterfactual_popweighted_provinces_correct.csv"

# Hydro files (long format)
hydro_f_file = "/home/users/weijia_6/Yangtze_code/gen_result/step5_gamma_spi6/hydro_std_anom_sep_factual_525_by_province.csv"
hydro_cf_file = "/home/users/weijia_6/Yangtze_code/gen_result/step5_gamma_spi6/hydro_std_anom_sep_counterfactual_525_by_province.csv"

# Output directory
out_dir = "/home/users/weijia_6/Yangtze_code/imbalance_result"
os.makedirs(out_dir, exist_ok=True)

out_fig_png = f"{out_dir}/Panel_Figure_energy_imbalance_violin.png"
out_fig_pdf = f"{out_dir}/Panel_Figure_energy_imbalance_violin.pdf"

out_summary_demand = f"{out_dir}/summary_demand_std_anomaly_2022.csv"
out_summary_generation = f"{out_dir}/summary_hydro_generation_std_anomaly_2022.csv"
out_summary_imbalance = f"{out_dir}/summary_energy_imbalance_std_2022.csv"
out_summary_delta = f"{out_dir}/summary_climate_change_energy_imbalance_2022.csv"


# ============================================================
# 2. Plot style
# ============================================================

plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "font.size": 14,
    "axes.titlesize": 17,
    "axes.labelsize": 15,
    "xtick.labelsize": 13,
    "ytick.labelsize": 13,
    "legend.fontsize": 12,
    "figure.titlesize": 18,
    "axes.linewidth": 1.0,
    "savefig.dpi": 450,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})

CF_COLOR = "#4C78A8"
F_COLOR = "#F58518"
DELTA_COLOR = "#2CA25F"

ROW_GAP = 1.4


# ============================================================
# 3. Helper functions
# ============================================================

def load_province_ensemble_csv(path):
    """
    Load a province x ensemble CSV (province in first column).
    """
    df = pd.read_csv(path, index_col=0)
    df.index = df.index.astype(str).str.strip()
    df.columns = [str(c).strip() for c in df.columns]

    missing = [p for p in selected_provinces if p not in df.index]
    if missing:
        raise KeyError(f"Missing provinces in {path}: {missing}")

    df = df.loc[selected_provinces].copy()
    df = df.apply(pd.to_numeric, errors="coerce")
    return df


def load_hydro_long_csv(path, value_col="hydro_std_anom_sep", scenario=None):
    """
    Load long-format hydro csv and pivot to province x ensemble.
    Expected columns include province, ensemble, hydro_std_anom_sep, optional scenario.
    """
    df = pd.read_csv(path)
    df.columns = [str(c).strip() for c in df.columns]

    required = {"province", "ensemble", value_col}
    missing = required - set(df.columns)
    if missing:
        raise KeyError(f"{path} is missing columns: {sorted(missing)}")

    df["province"] = df["province"].astype(str).str.strip()

    if scenario is not None and "scenario" in df.columns:
        df["scenario"] = df["scenario"].astype(str).str.strip().str.lower()
        scenario = str(scenario).strip().lower()
        df = df.loc[df["scenario"] == scenario].copy()

    df["ensemble"] = pd.to_numeric(df["ensemble"], errors="coerce")
    df[value_col] = pd.to_numeric(df[value_col], errors="coerce")

    df = df.loc[df["province"].isin(selected_provinces)].copy()

    wide = df.pivot_table(
        index="province",
        columns="ensemble",
        values=value_col,
        aggfunc="first",
    )

    missing = [p for p in selected_provinces if p not in wide.index]
    if missing:
        raise KeyError(f"Missing provinces in {path}: {missing}")

    wide = wide.loc[selected_provinces].copy()
    wide = wide.reindex(sorted(wide.columns), axis=1)
    wide.columns = range(wide.shape[1])

    return wide.apply(pd.to_numeric, errors="coerce")


def force_matching_columns(*dfs):
    ncols = [df.shape[1] for df in dfs]
    if len(set(ncols)) != 1:
        raise ValueError(f"Different ensemble sizes: {ncols}")
    n = ncols[0]
    out = []
    for df in dfs:
        tmp = df.copy()
        tmp.columns = range(n)
        out.append(tmp)
    return out


def summary_stats(df, prefix):
    return pd.DataFrame({
        f"{prefix}_mean": df.mean(axis=1),
        f"{prefix}_median": df.median(axis=1),
        f"{prefix}_2.5%": df.quantile(0.025, axis=1),
        f"{prefix}_25%": df.quantile(0.25, axis=1),
        f"{prefix}_75%": df.quantile(0.75, axis=1),
        f"{prefix}_97.5%": df.quantile(0.975, axis=1),
        f"{prefix}_P_gt_0": (df > 0).mean(axis=1),
    })


def paired_summary_stats(df_cf, df_f, name):
    return pd.concat(
        [
            summary_stats(df_cf, f"{name}_CF"),
            summary_stats(df_f, f"{name}_F"),
        ],
        axis=1,
    )


def symmetric_xlim(*dfs, pad=0.12):
    vals = np.concatenate([df.values.ravel() for df in dfs])
    vals = vals[np.isfinite(vals)]
    lo = np.nanquantile(vals, 0.005)
    hi = np.nanquantile(vals, 0.995)
    vmax = max(abs(lo), abs(hi))
    vmax *= (1 + pad)
    return -vmax, vmax


def data_xlim(*dfs, pad=0.12):
    vals = np.concatenate([df.values.ravel() for df in dfs])
    vals = vals[np.isfinite(vals)]
    lo = np.nanquantile(vals, 0.005)
    hi = np.nanquantile(vals, 0.995)
    span = hi - lo
    return lo - pad * span, hi + pad * span


def add_panel_label(ax, label):
    ax.text(
        -0.10, 1.08, label,
        transform=ax.transAxes,
        fontsize=16,
        fontweight="bold",
        va="top",
        ha="left",
    )


def half_violin(ax, values, position, color, width=0.62, side="right", alpha=0.35):
    values = np.asarray(values)
    values = values[np.isfinite(values)]
    vp = ax.violinplot(
        [values],
        positions=[position],
        vert=False,
        widths=width,
        showmeans=False,
        showmedians=False,
        showextrema=False,
    )
    body = vp["bodies"][0]
    verts = body.get_paths()[0].vertices

    if side == "right":
        verts[:, 1] = np.maximum(verts[:, 1], position)
    elif side == "left":
        verts[:, 1] = np.minimum(verts[:, 1], position)
    else:
        raise ValueError("side must be 'right' or 'left'")

    body.set_facecolor(color)
    body.set_edgecolor("none")
    body.set_alpha(alpha)
    body.set_zorder(1)
    return body


def draw_violin_paired(
    ax,
    df_cf,
    df_f,
    provinces,
    title,
    xlabel,
    panel_label=None,
    xlim=None,
    show_yticklabels=True,
):
    rng = np.random.default_rng(42)

    y_centres = np.arange(len(provinces)) * ROW_GAP
    offset = 0.22
    box_width = 0.12

    ax.axvline(0, color="0.70", linewidth=1.2, linestyle="-", zorder=0)
    ax.grid(axis="x", color="0.90", linewidth=0.9, linestyle="-", zorder=0)

    for i, prov in enumerate(provinces):
        cf_vals = df_cf.loc[prov].dropna().values
        f_vals = df_f.loc[prov].dropna().values

        y_cf = y_centres[i] - offset
        y_f = y_centres[i] + offset

        half_violin(ax, cf_vals, y_cf, CF_COLOR, width=0.62, side="left", alpha=0.32)
        half_violin(ax, f_vals, y_f, F_COLOR, width=0.62, side="right", alpha=0.32)

        bp = ax.boxplot(
            [cf_vals, f_vals],
            positions=[y_cf, y_f],
            vert=False,
            widths=box_width,
            patch_artist=True,
            showfliers=False,
            whis=(2.5, 97.5),
            medianprops=dict(color="black", linewidth=1.2),
            whiskerprops=dict(color="0.35", linewidth=1.0),
            capprops=dict(color="0.35", linewidth=1.0),
            boxprops=dict(linewidth=1.0),
            zorder=4,
        )

        for patch, color in zip(bp["boxes"], [CF_COLOR, F_COLOR]):
            patch.set_facecolor(color)
            patch.set_alpha(0.90)
            patch.set_edgecolor("0.25")

        ax.scatter(
            np.nanmean(cf_vals),
            y_cf,
            marker="D",
            s=32,
            color="black",
            edgecolor="white",
            linewidth=0.5,
            zorder=5,
        )

        ax.scatter(
            np.nanmean(f_vals),
            y_f,
            marker="D",
            s=32,
            color="black",
            edgecolor="white",
            linewidth=0.5,
            zorder=5,
        )

    ax.set_yticks(y_centres)
    if show_yticklabels:
        ax.set_yticklabels([province_short[p] for p in provinces])
    else:
        ax.set_yticklabels([])

    ax.invert_yaxis()

    if xlim is not None:
        ax.set_xlim(xlim)

    ax.set_title(title, loc="left", pad=10)
    ax.set_xlabel(xlabel, labelpad=8)

    if panel_label is not None:
        add_panel_label(ax, panel_label)

    ax.tick_params(axis="both", labelsize=12.5)

    for spine in ["top", "right"]:
        ax.spines[spine].set_visible(False)


def draw_violin_single(
    ax,
    df,
    provinces,
    title,
    xlabel,
    panel_label=None,
    xlim=None,
    color=DELTA_COLOR,
    show_yticklabels=True,
):
    y_centres = np.arange(len(provinces)) * ROW_GAP
    box_width = 0.14

    ax.axvline(0, color="0.70", linewidth=1.2, linestyle="-", zorder=0)
    ax.grid(axis="x", color="0.90", linewidth=0.9, linestyle="-", zorder=0)

    for i, prov in enumerate(provinces):
        vals = df.loc[prov].dropna().values
        y = y_centres[i]

        half_violin(ax, vals, y, color, width=0.76, side="right", alpha=0.36)

        bp = ax.boxplot(
            [vals],
            positions=[y],
            vert=False,
            widths=box_width,
            patch_artist=True,
            showfliers=False,
            whis=(2.5, 97.5),
            medianprops=dict(color="black", linewidth=1.2),
            whiskerprops=dict(color="0.35", linewidth=1.0),
            capprops=dict(color="0.35", linewidth=1.0),
            boxprops=dict(linewidth=1.0),
            zorder=4,
        )

        bp["boxes"][0].set_facecolor(color)
        bp["boxes"][0].set_alpha(0.90)
        bp["boxes"][0].set_edgecolor("0.25")

        ax.scatter(
            np.nanmean(vals),
            y,
            marker="D",
            s=34,
            color="black",
            edgecolor="white",
            linewidth=0.5,
            zorder=5,
        )

    ax.set_yticks(y_centres)
    if show_yticklabels:
        ax.set_yticklabels([province_short[p] for p in provinces])
    else:
        ax.set_yticklabels([])

    ax.invert_yaxis()

    if xlim is not None:
        ax.set_xlim(xlim)

    ax.set_title(title, loc="left", pad=10)
    ax.set_xlabel(xlabel, labelpad=8)

    if panel_label is not None:
        add_panel_label(ax, panel_label)

    ax.tick_params(axis="both", labelsize=13)

    for spine in ["top", "right"]:
        ax.spines[spine].set_visible(False)


def prepare_cdd_reference(hist_file, provinces, ref_year_start=1970, ref_year_end=2010):
    df = pd.read_csv(hist_file, index_col=0)
    df.index = df.index.astype(str).str.strip()
    df = df.loc[provinces].apply(pd.to_numeric, errors="coerce")

    year_cols = []
    for y in range(ref_year_start, ref_year_end + 1):
        found = None
        for c in (str(y), y):
            if c in df.columns:
                found = c
                break
        if found is None:
            raise KeyError(f"Year {y} not found in {hist_file}")
        year_cols.append(found)

    ref = df[year_cols].copy()
    ref.columns = [int(c) for c in year_cols]

    clim_mu = ref.mean(axis=1)
    clim_sd = ref.std(axis=1, ddof=1)

    if (clim_sd <= 0).any():
        bad = clim_sd[clim_sd <= 0].index.tolist()
        raise ValueError(f"Zero/negative CDD std for provinces: {bad}")

    return clim_mu, clim_sd


# ============================================================
# 4. Demand anomaly (CDD) using 1970-2010 baseline
# ============================================================

clim_mu, clim_sd = prepare_cdd_reference(
    hist_cdd_file,
    selected_provinces,
    ref_year_start=1970,
    ref_year_end=2010,
)

df_cdd_f = load_province_ensemble_csv(cdd_f_file)
df_cdd_cf = load_province_ensemble_csv(cdd_cf_file)

std_f_demand = df_cdd_f.sub(clim_mu, axis=0).div(clim_sd, axis=0)
std_cf_demand = df_cdd_cf.sub(clim_mu, axis=0).div(clim_sd, axis=0)

std_cf_demand, std_f_demand = force_matching_columns(std_cf_demand, std_f_demand)
summary_demand = paired_summary_stats(std_cf_demand, std_f_demand, "demand")
summary_demand.to_csv(out_summary_demand)


# ============================================================
# 5. Hydropower generation anomaly
# ============================================================

std_f_gen = load_hydro_long_csv(hydro_f_file, value_col="hydro_std_anom_sep", scenario="factual")
std_cf_gen = load_hydro_long_csv(hydro_cf_file, value_col="hydro_std_anom_sep", scenario="counterfactual")

std_cf_gen, std_f_gen = force_matching_columns(std_cf_gen, std_f_gen)
summary_generation = paired_summary_stats(std_cf_gen, std_f_gen, "generation")
summary_generation.to_csv(out_summary_generation)


# ============================================================
# 6. Energy imbalance and climate-change contribution
# ============================================================

n_demand = std_f_demand.shape[1]
n_gen = std_f_gen.shape[1]

if n_demand != n_gen:
    raise ValueError(
        f"Demand and hydropower ensemble sizes differ: demand={n_demand}, hydro={n_gen}"
    )

std_cf_demand, std_f_demand, std_cf_gen, std_f_gen = force_matching_columns(
    std_cf_demand, std_f_demand, std_cf_gen, std_f_gen
)

imb_cf = std_cf_demand - std_cf_gen
imb_f = std_f_demand - std_f_gen

summary_imbalance = paired_summary_stats(imb_cf, imb_f, "imbalance")
summary_imbalance.to_csv(out_summary_imbalance)

delta_imb = imb_f - imb_cf
summary_delta = summary_stats(delta_imb, "delta_imbalance")
summary_delta.to_csv(out_summary_delta)


# ============================================================
# 7. Axis limits
# ============================================================

xlim_demand = symmetric_xlim(std_cf_demand, std_f_demand, pad=0.15)
xlim_gen = symmetric_xlim(std_cf_gen, std_f_gen, pad=0.20)
xlim_imb = symmetric_xlim(imb_cf, imb_f, pad=0.15)

xlim_delta = data_xlim(delta_imb, pad=0.15)
xlim_delta = (min(xlim_delta[0], -0.5), max(xlim_delta[1], 0.5))


# ============================================================
# 8. Plot 2x2 violin figure
# ============================================================

fig, axes = plt.subplots(
    2,
    2,
    figsize=(15.0, 11),
    constrained_layout=False,
)

ax_a, ax_b, ax_c, ax_d = axes.ravel()

draw_violin_paired(
    ax=ax_a,
    df_cf=std_cf_demand,
    df_f=std_f_demand,
    provinces=selected_provinces,
    title="Demand anomaly",
    xlabel="CDD26 standardised anomaly",
    panel_label="a",
    xlim=xlim_demand,
    show_yticklabels=True,
)

draw_violin_paired(
    ax=ax_b,
    df_cf=std_cf_gen,
    df_f=std_f_gen,
    provinces=selected_provinces,
    title="Hydropower generation anomaly",
    xlabel="Generation standardised anomaly",
    panel_label="b",
    xlim=xlim_gen,
    show_yticklabels=False,
)

draw_violin_paired(
    ax=ax_c,
    df_cf=imb_cf,
    df_f=imb_f,
    provinces=selected_provinces,
    title="Energy imbalance anomaly",
    xlabel="Demand - generation, standardised units",
    panel_label="c",
    xlim=xlim_imb,
    show_yticklabels=True,
)

draw_violin_single(
    ax=ax_d,
    df=delta_imb,
    provinces=selected_provinces,
    title="Climate-change-driven energy imbalance",
    xlabel="Factual - counterfactual imbalance",
    panel_label="d",
    xlim=xlim_delta,
    color=DELTA_COLOR,
    show_yticklabels=False,
)

legend_handles = [
    Patch(facecolor=CF_COLOR, edgecolor="0.25", alpha=0.90, label="Counterfactual"),
    Patch(facecolor=F_COLOR, edgecolor="0.25", alpha=0.90, label="Factual"),
    Patch(facecolor=DELTA_COLOR, edgecolor="0.25", alpha=0.90, label="Factual - counterfactual"),
    Line2D([0], [0], marker="D", color="none", markerfacecolor="black", markeredgecolor="white", markeredgewidth=0.5, markersize=7, label="Mean"),
    Line2D([0], [0], color="black", linewidth=1.3, label="Median"),
]

fig.legend(
    handles=legend_handles,
    loc="lower center",
    ncol=5,
    frameon=False,
    bbox_to_anchor=(0.5, 0.015),
    columnspacing=1.5,
    handlelength=1.8,
)

plt.subplots_adjust(
    left=0.08,
    right=0.985,
    top=0.91,
    bottom=0.14,
    wspace=0.18,
    hspace=0.46,
)

fig.savefig(out_fig_png, dpi=450, bbox_inches="tight")
fig.savefig(out_fig_pdf, bbox_inches="tight")

plt.show()

# ============================================================
# 9. Print outputs
# ============================================================

print("\nSaved figure:")
print(f"  {out_fig_png}")
print(f"  {out_fig_pdf}")

print("\nSaved summary tables:")
print(f"  {out_summary_demand}")
print(f"  {out_summary_generation}")
print(f"  {out_summary_imbalance}")
print(f"  {out_summary_delta}")

print("\nClimate-change-driven energy imbalance summary:")
print(summary_delta.round(3))
```
